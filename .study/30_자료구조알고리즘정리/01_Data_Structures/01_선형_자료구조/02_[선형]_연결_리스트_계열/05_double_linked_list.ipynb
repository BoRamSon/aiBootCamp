# 🟩 [**Doubly Linked List**] 파이썬으로 `다중 연결 리스트` 구현하기  




```python
class Node:
  def __init__(self, data=None):
    self.prev = None   # 앞을 가르키는 node 주소
    self.data = data
    self.next = None   # 뒤를 가르키는 node 주소

# 더미노드로 생각합니다.
class DoubleLinkedList:
  def __init__(self):
    self.head = Node()  # 리스트 전체의 시작을 가리킨다.
    self.tail = Node()  # 리스트 전체의 끝을 가리킨다.
    self.head.prev = self.head
    self.head.next = self.tail
    self.tail.prev = self.head
    self.tail.next = self.tail
    # initialized 상태 >>>     head <=> tail

  # ------------------------------------------------------------
  #        head                                     tail
  #        node                node                 node       
  # [pr | data | ne]  →  [pr | data | ne]  →  [pr | data | ne]
  # ------------------------------------------------------------
  def insert_head(self, data):  # 맨 앞에 넣는 것
    temp = Node(data)
    # 기존에 head에 있던 주소를 가져와야겠지?
    temp.prev = self.head
    temp.next = self.head.next
    # 2. 다음에 head 오른쪽에 있던 노드의 prev 주소도 바꿔줘야겠지?  # 🔥 나중에 순서 문제가 있었음 이거 먼저 해줬어야 합니다.
    self.head.next.prev = temp
    # 1. 이제 head를 정의해줘야겠지? 바뀌었으니깐
    self.head.next = temp
    

  def insert_tail(self, data):  # 맨 뒤에 넣는 것
    temp = Node(data)
    temp.prev = self.tail.prev
    temp.next = self.tail
    self.tail.prev.next = temp  # 🔥 이거 먼저 해줬어야 합니다.
    self.tail.prev = temp
    

  def print_prev(self):
    temp = self.tail.prev
    print(f"TAIL < ", end="")
    while temp != self.head:   # 더미노드가 서로를 link하고 있지 않다면.. temp.prev가 head를 가르킬 때까지 동작하겠다!
      print(f"{temp.data}", end=" < ")  
      temp = temp.prev   # 계속 이전으로 이동하면서 print 한다.
    print(f"HEAD")
  
  def print_next(self):
    temp = self.head.next
    print(f"HEAD > ", end="")
    while temp != self.tail: 
      print(f"{temp.data}", end=" > ")
      temp = temp.next   # 계속 뒤로 이동하면서 print 한다.
    print(f"TAIL")

  def delete_head(self):
    if self.head.next == self.tail:  # 지금 비었는지 확인하는 것입니다. 
      return
    # head.next가 가르키는 것을 한단계 건너뛰어서 linked 해주면 됩니다. 
    self.head.next = self.head.next.next  
    # 지금 head.next가 원하는대로 바뀐 상태이기 때문에 그대로 prev는 head로 해주면 됩니다.
    self.head.next.prev = self.head

  def delete_tail(self):
    if self.head.next == self.tail:  # 지금 비었는지 확인하는 것입니다. 
      return
    # tail.prev가 가르키는 것을 한단계 건너뛰어서 linked 해주면 됩니다.
    self.tail.prev = self.tail.prev.prev
    # 지금 tail.prev가 바뀐 상태이기 때문에 그대로 next는 tail로 해주면 됩니다.
    self.tail.prev.next = self.tail


dlist = DoubleLinkedList()

# -----------------------------
# head 추가해보기
print('------------------ 추가 ------------------')
dlist.insert_head("A")
dlist.insert_head("B")
dlist.insert_head("C")
dlist.insert_head("D")
dlist.insert_head("E")
dlist.insert_head("F")
dlist.insert_head("G")
dlist.insert_head("H")

# 앞에서 부터 print
dlist.print_next()


# -----------------------------
# tail도 추가해보기
dlist.insert_tail("I")

# 뒤에서부터 print
dlist.print_prev()

# 다시 print 해보기
dlist.print_next()


# -----------------------------
# head 삭제해보기
print('\n------------------ 삭제 ------------------')
dlist.delete_head()

# 앞에서 부터 print
dlist.print_next() # H 삭제


# -----------------------------
# tail 삭제해보기
dlist.delete_tail()  # I 삭제
dlist.delete_tail()  # A 삭제

# 앞에서 부터 print
dlist.print_next() 

```
