# 🟩 Stack 문제 풀어보기 2  
  > 후위 수식 계산기(Postfix Expression Evaluator)  
    


#### ⚫️ 스택(Stack) 자료구조에서 사용하는 기본 연산(메서드, 함수)  
- **is_full()**: 스택이 가득 찼는지 확인하는 작업 (가득 찼으면 True, 주로 용량 제한이 있을 때 사용)  
- **is_empty()**: 스택이 비어있는지 확인하는 작업 (비었으면 True)  
- **push(x)**: 자료 x를 스택에 넣는 작업 (삽입)  
- **pop()**: 스택에서 가장 마지막에 넣은 자료를 꺼내는 작업 (삭제)  
- **peek()**: 스택의 가장 마지막 자료를 확인하는 작업 (꺼내지 않고 보기)  


```python
# 먼저 앞서서 숫자와 연산자를 분류하는 함수를 만들었습니다.
# '숫자 넣은 stack'을 만들고 / '연산자 넣는 stack'을 만들고 분류하는 함수 만들기
# 그 다음 연산처리까지 하면 끝!!!

class Stack:
    def __init__(self, size=10):
        if size < 10:
            self.size = 10   # 최소 stack 크기를 10으로 하기 위한 것
        else : 
            self.size = size 

        self.stack = []
        for i in range(0, self.size):  # 단순 stack 빈값 채워주는 것입니다.
            self.stack.append(0)

        self.top = -1 


    # is_full
    def is_full(self):
        if self.size-1 == self.top:
            return True
        return False

    # is_empty
    def is_empty(self):
        if self.top == -1:
            return True
        return False

    # push
    # isFull상태가 아니면 top증가 시키고 그 안에 값넣기
    def push(self, data):
        if self.is_full():
            print('이미 가득 참')
            return 
        self.top += 1
        self.stack[self.top] = data  

    # 
    def pop(self):
        if self.is_empty():
            return None

        # return self.stack.pop()  # 맨 뒤에서 꺼냄 

        item = self.stack[self.top]
        self.top -= 1
        return item
    

    def peek(self):  # 스택의 가장 마지막 자료를 확인하는 작업 (꺼내지 않고 보기)  
        if self.is_empty():
            return None
        return self.stack[self.top]  # 맨 뒤 값을 반환 (삭제는 안 함)


    def print(self):
        i = 0 
        while i <= self.top:
            print(self.stack[i], end=" ")
            i+=1
        print()

```

```python
def is_digit(ch):
  if(ord(ch)>=ord('0') and ord(ch)<=ord('9')):
    return True 
  return False 

def is_operator(ch):
  if ch == "+" or ch == "-" or ch == "*" or ch == "/":
    return True
  return False

def operate(c, left, right):
  if c == "+":
    result = int(left) + int(right)
  elif c == "-":
    result = int(left) - int(right)
  elif c == "*":
    result = int(left) * int(right)
  else:
    result = int(left) // int(right)
  return result

def get_result(postfix):
  s = Stack(100)
  for c in postfix:
    if is_digit(c):
      s.push(c)
    elif is_operator(c):
      right = s.pop()
      left = s.pop()
      result = operate(c, left, right)
      s.push(result)
  return s.pop()
      
s = "4 5 2 * + 7 3 / -"

print(get_result(s))
```
