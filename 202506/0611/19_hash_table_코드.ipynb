# 🟩 자료구조 - 해시(Hash)  

값 =>  해쉬함수 ==>  해쉬테이블에 저장(특정메모리로 이동한다)  
                    (bucket 이라도도 부른다.)  
                    (배열만, 배열과 링크드리스트, 링크드리스트로만으로도 구성가능)  
  
  
- 해시테이블 인덱스 결정 방법 = hash(key) % table_size 방식으로 버킷 인덱스를 결정  
  
school   =======>  23432%100  32 -> bucket[32]    
rain     =======>  33454%100  54 -> bucket[54]    
rainbow  =======>  23412%100  32 -> bucket[12]  
                                    (hash table)  

"school" 이라는 문자열을 해시 함수에 넣었더니 23432라는 숫자가 나왔고  
이를 %100 (100으로 나눈 나머지) 하면 32  
⇒ 이 값이 bucket[32] 에 저장됨  

"rain" → 해시값 33454 → %100 → 54  
⇒ bucket[54] 에 저장  

"rainbow" → 해시값 23412 → %100 → 12  
⇒ bucket[12] 에 저장  

bucket의 크기는 전체 key값 개수 1/2 ~1/3    
bucket[100]   







```python

class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None 
    def __str__(self):  #기본 연산자 오버라이딩, 객체를 print할때 이 연산자를 overriding 
                        #해주면 원하는 대로 출력을 해준다.
                        #안하면 객체 주소값 출력 

        return  f"{self.data} {self.next}"
    
class HashTable:
    def __init__(self, cnt=10):
        self.cnt = cnt  #10개보다 작으면 최소한 bucket개수를 10개는 
                        #만들자
        if self.cnt<10:
            self.cnt = 10
        self.bucketList = [None]*cnt
        #앞에 완충장치용으로 노드를 미리 만들어서 붙여놓자
        self.bucketList = [ Node() for i in range(0, self.cnt)]
        for i in range(0, 5):
            print(self.bucketList[i])

    def getHash(self, key): #해쉬함수 붙이기 
        total=0
        for k in key:
            total += ord(k)
        return total%self.cnt     

    #해쉬테이블 구성 함수 => 해쉬테이블을 구축을 해야 검색을 빠르게 
    #처음에 해쉬테이블 구축시간이 많이 걸린다. Mysql join 이 외부에서 inner join, outer join
    #실제로 mysql 안에 옵티마이저 Nested loop join(for문 돌림), Hash join(해쉬테이블)
    #hash join 은 데이터가 대용량리고 배치처리( 일감을 한번에 모아서 처리하는 방식)<-> 온라인처리
    # (실시간처리)  에 유리하다. 보통의 경우는 실시간처리(고객이 통장을 개설해주세요), 이때는
    # nested loop join이 빠르다 
    # trade off : 균형잡기 - 다양한 알고리즘중에 적당한 걸 선택해야 한다.   

    def createTable(self, dataList=None): #생성자에서 해도 되는데 코드가 길어질거 같아서 함수를 별도로 제작
                            #생성자에서 호출하면 된다.
        #외부에서 [("book"), ("school"), ("rain")] 값을 받아온다 
        for item in dataList:
            #1.해쉬값을 가져온다
            key = self.getHash(item)
            #print( item, key)
            #2.Node(key) 를 만들어서 노드의 앞쪽에 붙인다
            bucket = Node(item)
            #3.bucketList의 앞쪽에 붙인다.
            bucket.next = self.bucketList[key].next
            self.bucketList[key].next = bucket 
            
    def printList(self):
        for item in self.bucketList:
            trace = item.next 
            while trace!=None:
                print(trace.data, end="=>")
                trace = trace.next 
            print() 

hash = HashTable()
hash.createTable([("school"), ("desk"), ("chair"), ("rain"), ("survey"),
                    ("house"), ("home"), ("doll"), ("python"), 
                    ("java"), ("html"), ("javascript")])
hash.printList()
```

```python


# Node 클래스: 연결 리스트에서 사용되는 노드 단위
class Node:
    def __init__(self, data=None):
        self.data = data        # 데이터 저장
        self.next = None        # 다음 노드를 가리키는 포인터

    def __str__(self):
        # 객체를 print()할 때 출력 형태 정의
        return f"{self.data} {self.next}"

# 해시 테이블 클래스 정의 (💕체이닝 방식)
class HashTable:
    def __init__(self, cnt=10):
        self.cnt = cnt  # 기본 버킷 개수
        if self.cnt < 10:
            self.cnt = 10  # 최소 10개는 보장

        # 연결 리스트용 dummy head 노드를 각 버킷에 하나씩 생성
        self.bucketList = [Node() for i in range(self.cnt)]

        # 디버깅용: 앞부분 몇 개 버킷의 상태를 출력
        for i in range(0, 5):
            print(self.bucketList[i])

    # 문자열 키를 해시값(인덱스)으로 변환하는 해시 함수
    def getHash(self, key):
        total = 0
        for k in key:
            total += ord(k)  # 각 문자 아스키값을 더함
        return total % self.cnt  # 테이블 크기로 나눈 나머지를 인덱스로 사용

    # 해시 테이블을 구축하는 함수
    def createTable(self, dataList=None):
        for item in dataList:  # 예: ["school", "desk", "rain", ...]
            key = self.getHash(item)  # 해시 인덱스 계산
            bucket = Node(item)       # 새 노드를 생성 (데이터 저장)
            
            # 새 노드를 현재 연결 리스트의 맨 앞에 삽입 (head insertion)
            bucket.next = self.bucketList[key].next
            self.bucketList[key].next = bucket

    # 테이블 상태를 출력하는 함수
    def printList(self):
        for item in self.bucketList:
            trace = item.next  # dummy head 다음부터 시작
            while trace != None:
                print(trace.data, end="=>")  # 연결 리스트 형식으로 출력
                trace = trace.next
            print()  # 버킷 하나 끝나면 줄바꿈



# 해시 테이블 인스턴스 생성
hash = HashTable()

# 키 목록을 테이블에 삽입
hash.createTable([
    "school", "desk", "chair", "rain", "survey",
    "house", "home", "doll", "python", "java",
    "html", "javascript"
])

# 해시 테이블 구조 출력
hash.printList()
```
