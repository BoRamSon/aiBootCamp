## 🟢 크롤링  

- 크롤링과 웹스크래핑의 차이점  
    - 크롤링: 여러 웹페이지를 돌아다니며 찾는 일 (페이지 발견하기)  
    - 웹스크래핑: 찾은 페이지에서 원하는 정보만 뽑아내는 일 (정보 추출하기)  

### 크롤링과 웹스크래핑 사용법  

크롤링과 웹스크래핑을 시작하는 방법을 쉽게 알려드릴게요.  

### 기본 준비물  

크롤링을 하려면 다음 준비물이 필요해요:  

1. **컴퓨터**  
2. **인터넷 연결**  
3. **파이썬** (가장 인기 있는 크롤링 언어)  
4. **필요한 라이브러리** (도구상자라고 생각하세요)  

### 크롤링 시작하기 - 단계별 안내  

#### 1. 파이썬 설치하기  

파이썬은 컴퓨터에게 일을 시키기 위한 언어예요. 파이썬 공식 웹사이트에서 다운로드하고 설치할 수 있어요.  

#### 2. 필요한 도구(라이브러리) 설치하기  

파이썬에 특별한 도구를 추가해야 해요. 명령창(cmd나 터미널)에서 이렇게 입력해요:  

```bash
pip install requests beautifulsoup4 pandas  
```

- 이 명령어는 다음 도구들을 설치해요:  
    - requests: 웹페이지를 가져오는 도구  
    - beautifulsoup4: 웹페이지에서 정보를 찾는 도구  
    - pandas: 모은 정보를 정리하는 도구  



#### 3. 크롤링 목표 정하기  
- 무엇을 모을지 정확히 정해야 해요. 예를 들면:  
    - 뉴스 제목 모으기  
    - 쇼핑몰 상품 가격 모으기  
    - 영화 평점 모으기  

#### 4. 웹사이트 분석하기  
- 정보를 모을 웹사이트를 찾고, 브라우저의 "개발자 도구"를 사용해 HTML 구조를 살펴봐요.  
- 개발자 도구 열기:  
    - 크롬: F12 키 또는 오른쪽 클릭 → '검사'  
    - 파이어폭스: F12 키 또는 오른쪽 클릭 → '요소 검사'  
- 이것은 마치 책의 목차를 보는 것과 같아요. 어떤 정보가 어디에 있는지 알 수 있어요.  

#### 5. 웹사이트의 규칙 확인하기  
`robots.txt` 파일을 확인하세요. 예: `www.example.com/robots.txt`  
이 파일은 웹사이트 주인이 "이 페이지는 크롤링해도 돼요, 이 페이지는 안 돼요"라고 알려주는 규칙집이에요.  

### 간단한 크롤링 과정  
1. **웹페이지 가져오기**: 원하는 웹사이트에 방문해서 HTML 코드를 가져와요.  
2. **정보 추출하기**: HTML에서 원하는 정보만 찾아내요.  
3. **정보 저장하기**: 찾은 정보를 파일(CSV, Excel 등)로 저장해요.  
4. **다음 페이지로 이동**: 필요하다면 다른 페이지로 이동해 1~3 과정을 반복해요.  

### 크롤링할 때 주의할 점  
1. **천천히 크롤링하기**: 너무 빠르게 많은 요청을 보내면 웹사이트에 부담을 줘요. 마치 도서관에서 한 번에 책 100권을 빌리려고 하는 것과 같아요.  
2. **봇처럼 행동하지 않기**: 사람처럼 행동해야 해요. 시간 간격을 두고 요청을 보내세요.  
3. **robots.txt 지키기**: 웹사이트의 규칙을 꼭 지켜야 해요.  
4. **개인정보 주의**: 개인정보가 포함된 데이터는 특별히 조심해서 다뤄야 해요.  
5. **저작권 확인**: 모든 정보가 자유롭게 사용할 수 있는 것은 아니에요.  

### 크롤링을 더 쉽게 해주는 도구들  
1. **Scrapy**: 큰 프로젝트에 좋은 크롤링 전문 도구  
2. **Selenium**: 자바스크립트가 많은 웹사이트를 크롤링할 때 유용한 도구  
3. **Octoparse, ParseHub**: 코딩 없이도 크롤링할 수 있는 도구  

### 언제 어떤 도구를 써야 할까?  
- **간단한 크롤링**: Requests + BeautifulSoup (초보자에게 좋아요)  
- **복잡한 웹사이트**: Selenium (로그인이 필요하거나 버튼을 눌러야 할 때)  
- **대규모 크롤링**: Scrapy (많은 웹사이트를 크롤링할 때)  

크롤링은 마치 블록 쌓기와 같아요. 처음엔 간단하게 시작해서 점점 더 복잡한 것을 만들어 나가면 됩니다!  

### 🟡 [1] 특정 웹사이트에 접속하기  

````python
import requests  # 웹페이지를 가져오는 도구를 불러옵니다

# 웹사이트 주소를 정합니다
url = "https://boramson.org"

# 웹사이트에 방문해서 내용을 가져옵니다 (마치 웹사이트를 방문하는 것처럼)
response = requests.get(url)

# 웹사이트가 잘 응답했는지 확인합니다 (200은 "성공"이라는 신호예요)
if response.status_code == 200:
    # 웹페이지의 내용을 출력합니다
    print("웹페이지를 성공적으로 가져왔어요!")
    print(response.text[:500])  # 처음 500자만 보여줍니다
else:
    # 문제가 생겼다면 에러 메시지를 보여줍니다
    print(f"웹페이지를 가져오는데 실패했어요. 에러 코드: {response.status_code}")
````

### 🟡 [2] 웹페이지에서 정보 추출하기  

````python
import requests  # 웹페이지 가져오기 도구
from bs4 import BeautifulSoup  # HTML을 분석하는 도구

# 크롤링할 웹사이트 주소
url = "https://boramson.org"

# 웹사이트 방문하기
response = requests.get(url)

# 웹사이트가 잘 응답했는지 확인
if response.status_code == 200:
    # HTML 내용을 가져와서 BeautifulSoup으로 분석하기 쉽게 만듭니다
    # 마치 책을 펼쳐서 목차를 보기 쉽게 정리하는 것과 같아요
    soup = BeautifulSoup(response.text, "html.parser")

    # 모든 제목(h1 태그) 찾기
    # 이것은 마치 책에서 챕터 제목만 찾는 것과 같아요
    titles = soup.find_all("h1")

    # 찾은 제목들 출력하기
    print("웹페이지에서 찾은 제목들:")
    for title in titles:
        print(title.text.strip())  # strip()은 앞뒤 공백을 제거해줍니다

    # 모든 링크(a 태그) 찾기
    # 이것은 마치 책의 참고 자료 목록을 찾는 것과 같아요
    links = soup.find_all("a")

    print("\n웹페이지에서 찾은 링크들:")
    for link in links:
        # href는 링크의 주소를 담고 있는 속성이에요
        if link.has_attr("href"):
            print(f"링크 텍스트: {link.text.strip()}, 주소: {link['href']}")
else:
    print(f"웹페이지를 가져오는데 실패했어요. 에러 코드: {response.status_code}")
````
