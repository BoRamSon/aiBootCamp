# 🟩 자료구조 stack - 후위표기식 문제풀기  

## 🟢 stack class 재활용  

````python
class MyStack:
    def __init__(self, size=10):
        if size < 10:
            self.size = 10 
        else : 
            self.size = size 
        self.stack = []
        for i in range(0, self.size):
            self.stack.append(0)
        self.top = -1   

    def is_full(self):
        if self.size-1 == self.top:
            return True
        return False

    def is_empty(self):
        if self.top == -1:
            return True
        return False

    def push(self, data):
        if self.is_full():
            print('이미 가득 참')
            return 
        self.top += 1  
        self.stack[self.top] = data  

    def pop(self):
        if self.is_empty():
            return None
        item = self.stack[self.top]
        self.top -= 1
        return item
    
    def peek(self):  
        if self.is_empty():
            return None
        return self.stack[self.top]

    def print(self):
        i = 0 
        while i <= self.top:
            print(self.stack[i], end=" ")
            i+=1
        print()

````

## 🟢 후위표기식 문제 풀기  

4 + 2 * 7 - 3 * 2 + 9  
이것을 후위표기식으로 만드시오~  

4 + (2 * 7) - (3 * 2) + 9  

<br>

### 🔄 후위 표기식으로 변환 순서  
```markdown
    2 * 7                 →     2 7 *  
    4 + (2 * 7)           →     4 2 7 * +  
    3 * 2                 →     3 2 *  
    4 2 7 * + - (3 2 *)   →     4 2 7 * + 3 2 * -    
        (진짜 앞에서만 먼저 생각해야하는구나)  
    + 9                   →     4 2 7 * + 3 2 * - 9 +  

    <최종 후위 표기식>
    4 2 7 * + 3 2 * - 9 +  
```

  - 이후 현재 연산자 push

````python
# from stack import MyStack

# infix 수식을 받아서 postfix수식으로 전환하여 반환하는 함수

# 수식 중 연산자인지 확인하는 함수
def is_operator(c):  
    return c in "+-*/"  # +, -, *, / 중 하나인지 확인

# 공백 문자인지 확인하는 함수  (사용하지 않음)
# def is_space(c):
#     return c == " "  # 문자열이 공백 한 개와 같은지 확인

# 연산자의 우선순위를 반환하는 함수
def get_priority(c):
    if c in "+-":
        return 1  # +, -의 우선순위는 1
    elif c in "*/":
        return 2  # *, /의 우선순위는 2 (더 높음)
    else:
        return 0  # 연산자가 아니면 우선순위 없음


# =======================================================

# 중위표기식을 후위표기식(Postfix)으로 변환하는 함수
def postfix(expr):
    stack = MyStack()  # 연산자를 담기 위한 스택 생성
    result = ""        # 후위표기식 결과를 담을 문자열

    # 중위표기식의 각 문자(s)를 하나씩 순회
    for s in expr:
        
        # 1. 피연산자인 경우: 숫자일 경우 결과 문자열에 바로 추가
        if s.isdigit():
            result += s  # 숫자는 그대로 출력
        
        # 2. 연산자인 경우
        elif is_operator(s):
            result += " "  # 숫자와 연산자 사이를 구분해주기 위한 공백 추가

            # 스택이 비어있지 않고, 스택의 top에 있는 연산자의 우선순위가
            # 현재 연산자보다 크거나 같으면, 스택에서 pop해서 결과에 추가
            while not stack.is_empty() and \
                    get_priority(stack.peek()) >= get_priority(s):
                result = result + " " + stack.pop()
            
            # 현재 연산자를 스택에 push
            stack.push(s)
        
        # 3. 공백이나 기타 문자는 그대로 추가 (이 부분은 후속 개선 필요)
        else:
            result += s

    # 4. 모든 문자를 순회한 뒤에도 스택에 남은 연산자가 있다면 모두 pop해서 결과에 추가
    while not stack.is_empty():
        result += " " + stack.pop()

    return result  # 후위표기식 결과 반환

# 테스트: 중위표기식 '37*5+2'을 후위표기식으로 변환
print(postfix('37*5+2'))  # 결과: '37 5 * 2 +'
````
