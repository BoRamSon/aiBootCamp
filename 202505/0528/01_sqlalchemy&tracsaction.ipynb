### ⚫️ 정의
- SQLAlchemy는 **Python에서 데이터베이스를 쉽게 다룰 수 있게 도와주는 ORM 도구(라이브러리)**입니다.
- 정확히 말하면: ✅ SQL을 직접 쓰지 않아도, Python 코드만으로 데이터베이스 테이블을 만들고, 데이터를 추가/조회/수정/삭제할 수 있는 ORM 도구 입니다.
- Sequelize:	SQL 없이도 DB를 조작할 수 있게 해주는 ORM 도구로 node.js 환경에서 사용하는 것입니다.

### ⚫️ ORM 및 다른 설명들
- ORM (Object Relation Mapping)
  - 객체 지향 프로그래밍 언어에서 사용하는 객체와 데이터베이스의 테이블(관계형 데이터) 사이의 불일치를 해결하기 위해 사용하는 기술.
  - '객체지향 언어' <-> '관계형 디비'를 연결하는 프로그램 기법
  - django(장고, 파이썬 웹프레임워크)
  - jpa - JAVA
  - 파이썬에서 SQL Alchemy 가 지원한다.
  - 전체 테이블이 10개 미만으로 사용해야합니다.
  - 테이블 3개 이상 조인 시 속도문제, 테이블간의 구조가 복잡해질수록 orm만으로는 처리가 되지 않는다.

- SPA (Single Page Appliacation)
  - 인스타그램, 페이스북 등의 개발기법으로 웹페이지를 하나만 만들어서 페이지 체인지를 하는 방식으로 화면 이동이 훨씬 매끄럽다 무한스크롤 등을 지원한다.
    - react, vue.js(우리나라 금융권), angular, polymer(유튜브)
  - 웹페이지가 하나만 만들어서 돌리는 방식
  - 인스타그램 개발하다가 만든 것이 react 라이브러리

- SQLite
  - 본래의 의미는 디비는 아니고, 사실은 파일이다.
  - 다만 쿼리형식으로 읽고 쓰기는 가능하나 네트워크가 안됨... 마치 휴대폰의 전화번호부와 같다.


### ⚫️ "SQL을 직접 쓸 줄 아는데 왜 ORM을 써야 하나요?"
| 이유            | 설명                                           |
| ------------- | -------------------------------------------- |
| **생산성 ↑**     | 반복적인 SQL을 코드로 자동 처리함 (삽입/조회 등)               |
| **유지보수 편리**   | 테이블 구조 변경 시 코드 한 곳만 수정하면 반영됨                 |
| **보안 ↑**      | 파라미터 자동 처리 → SQL 인젝션 막기 쉬움                   |
| **객체 지향적 개발** | 테이블을 클래스로 다룰 수 있어 코드 일관성 유지                  |
| **다중 DB 지원**  | SQLite, MySQL, PostgreSQL 등 변경 시 SQL 거의 안 바꿈 |

## 🟢 sqlalchemy 라이브러리를 이용하여 기초 DB 조작 class 만들기

- 🟡 명령어를 생각해보자!!!
  
  - conn
    - conn   # : -- db 연결
    - conn.commit()   # 변경사항적용

  - cursor   # : -- db 연결되었으니, 조작해!
    - conn.cursor(pymysql.cursors.DictCursor)   # dict type으로 가져올게
    - 🔥 conn.cursor(~~).execute("SELECT * FROM tb_member WHERE user_id = %s", ("boram",))    # 실제 sql 쿼리를 실행
    - conn.cursor.executemany()

    - conn.cursor.fetchall()


  - execute() 형태
    - conn.execute(sql문, 파라미터)


```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

# SQLAlchemyrk PyMySQL을 내부적으로 사용하며 pool 지원
engine = create_engine(
    "mysql+pymysql://root:비밀번호@localhost:3306/mydb",
    pool_size=10,  # 최대 연결 수
    max_overflow=5,  # 초과 시 추가 연결 수
    pool_recycle=3600,  # 재활용 시간
)

# 1. 데이터 가져오기
# 데이터베이스 연결 객체(engine)를 통해 connect()를 호출하여 연결을 시작
# 이게 끝나면 자동으로 닫힘.
with engine.connect() as conn:
    sql = """
        select empno, ename, sal
        from emp
    """
    # [방법 1] 튜플 형식으로 결과 가져오기
    # SQL 실행: text(sql)로 SQL 문자열을 실행 가능한 객체로 변환
    result = conn.execute(text(sql))
    # result.all() → 전체 결과를 리스트 형태로 가져옴 (각 행은 튜플)
    # for now in result.all():
    #     print(now)  # 예: (7369, 'SMITH', 800)

    # [방법 2] 딕셔너리 형식으로 결과 가져오기
    # 다시 SQL 실행 (주의: 위의 result는 이미 한 번 소모되었기 때문에 새로 실행해야 함)
    result = conn.execute(text(sql))
    # result.mappings() → 각 행을 딕셔너리처럼 가져올 수 있게 함
    # .all()로 전체 결과 리스트로 변환
    rows = result.mappings().all()
    # 각 row는 Mapping 타입 (dict처럼 작동), dict(row)로 실제 딕셔너리로 변환 가능
    # for row in rows:
    #     print(dict(row))  # 예: {'empno': 7369, 'ename': 'SMITH', 'sal': 800}

# 2. 전체 검색
with engine.connect() as conn:
    sql = """
        select empno, ename, sal
        from emp
    """
    result = conn.execute(text(sql))

    for now in result.all():
        print(now)


# 2. 검색어를 전달할 때
with engine.connect() as conn:
    ename = "백승빈" #나중에 input 받으삼

    sql = """
        select empno, ename, sal
        from emp
        where ename = :name
    """

    # :name
    result = conn.execute(text(sql), [{"name" : ename}])
    # print(list(result)) # 파악만 하고 주석처리
    temp = result.all() # 한번 읽으면 없어지니깐, 복사해서 계속 가지고 있어라~ 
                # 왜 자꾸 all()을 붙이는가????  =  쿼리 결과를 "전부 리스트로 가져오는 함수
    print(temp)
    # if len(temp) == 0:
    #     print("없음")
    # else:
    #     for now in temp:
    #         print(now)


# 3. insert 
with engine.connect() as conn:
    sql = """
        select ifnull(max(empno), 0) + 1
        from emp
    """
    result = conn.execute(text(sql))  # [()] 이렇게 tuple로 옵니다.

    empno = result.all()[0][0]

    sql = """
        insert into emp(empno, ename, sal)
        values(:empno, :ename, :sal)
    """

    conn.execute(text(sql), [{"empno":empno, "ename":"테스트"+str(empno), "sal":9000}])

    conn.commit()




```

## 🟢 우리의 미션! = 트랜잭션 처리를 합시다!!!!
위 test1과 test2에 대해서 Tracsaction 처리를 합니다.

- Transaction
    - connet()를 beging으로 바꾼다. 
    - 작동되는 것을 하나로 (test1동작 + test2동작) 합친다.
    - 하나라도 안되면 모두 commit(반영)되지 않는다.

```python
# Transaction
    # connet()를 beging으로 바꾼다. 
    # 작동되는 것을 하나로 (test1동작 + test2동작) 합친다.

with engine.begin() as conn:
    # -----------------
    # test1
    sql = """
        select ifnull(max(id), 0) + 1
        from test1
    """
    result = conn.execute(text(sql))  # [()] 이렇게 tuple로 옵니다.
    id = result.all()[0][0]
    sql = """
        insert into test1(id, field)
        values(:id, :field)
    """
    conn.execute(text(sql), [{"id":id, "field":"test"}])
    # begin이 commit을 자동으로 함

    # -----------------
    # test2
    sql = """
        select ifnull(max(id), 0) + 1
        from test2
    """
    result = conn.execute(text(sql))  # [()] 이렇게 tuple로 옵니다.
    id = result.all()[0][0]
    sql = """
        insert into test2(id, field)
        values(:id, :field)
    """
    conn.execute(text(sql), [{"id":id, "field":"testBegin"}])
    # begin이 commit을 자동으로 함
    
```

## 🟢 sqlalchemy 라이브러리를 통한 조작 로직들 class로 만들어보기

```python
from sqlalchemy import create_engine, text
from sqlalchemy.exc import SQLAlchemyError

class Alchemy:
    def __init__(self):
        self.engine = create_engine(
            "mysql+pymysql://root:비밀번호@localhost:3306/mydb",
            pool_size=10,  # 최대 연결 수
            max_overflow=5,  # 초과 시 추가 연결 수
            pool_recycle=3600,  # 재활용 시간
        )

    # ------------------------------------
    # 1. 모든 데이터 가져오기
    def read_all_data(self):
        # 1. 데이터 가져오기
        # 데이터베이스 연결 객체(engine)를 통해 connect()를 호출하여 연결을 시작
        # 이게 끝나면 자동으로 닫힘.
        with self.engine.connect() as conn:
            sql = """
                select empno, ename, sal
                from emp
            """
            # [방법 1] 튜플 형식으로 결과 가져오기
            # SQL 실행: text(sql)로 SQL 문자열을 실행 가능한 객체로 변환
            result = conn.execute(text(sql))
            # result.all() → 전체 결과를 리스트 형태로 가져옴 (각 행은 튜플)
            # for now in result.all():
            #     print(now)  # 예: (7369, 'SMITH', 800)

            # [방법 2] 딕셔너리 형식으로 결과 가져오기
            # 다시 SQL 실행 (주의: 위의 result는 이미 한 번 소모되었기 때문에 새로 실행해야 함)
            result = conn.execute(text(sql))
            # result.mappings() → 각 행을 딕셔너리처럼 가져올 수 있게 함
            # .all()로 전체 결과 리스트로 변환
            rows = result.mappings().all()
            # 각 row는 Mapping 타입 (dict처럼 작동), dict(row)로 실제 딕셔너리로 변환 가능
            # for row in rows:
            #     print(dict(row))  # 예: {'empno': 7369, 'ename': 'SMITH', 'sal': 800}

    def read_all_data2(self):
        with self.engine.connect() as conn:
            sql = """
                select empno, ename, sal
                from emp
            """
            result = conn.execute(text(sql))

            for now in result.all():
                print(now)

    # ------------------------------------
    # 2. 특정 데이터 가져오기 / 검색어를 전달할 때
    def read_specific_date(self):   
        with self.engine.connect() as conn:
            ename = input("찾을 이름 입력하세요. : ")

            sql = """
                select empno, ename, sal
                from emp
                where ename = :name
            """

            # :name
            result = conn.execute(text(sql), [{"name" : ename}])
            # print(list(result)) # 파악만 하고 주석처리
            # print(result.all()) # 왜 자꾸 all()을 붙이는가????  =  쿼리 결과를 "전부 리스트로 가져오는 함수
            temp = result.all() # 한번 읽으면 없어지니깐, 변수에 넣고 실행하는 동안 계속 가지고 있어라~ 
            print(temp)
            # if len(temp) == 0:
            #     print("없음")
            # else:
            #     for now in temp:
            #         print(now)

    # ------------------------------------
    # 3. insert 
    def insert_data(self):
        with self.engine.connect() as conn:
            sql = """
                select ifnull(max(empno), 0) + 1
                from emp
            """
            result = conn.execute(text(sql))  # [()] 이렇게 tuple로 옵니다.

            empno = result.all()[0][0]

            sql = """
                insert into emp(empno, ename, sal)
                values(:empno, :ename, :sal)
            """

            conn.execute(text(sql), [{"empno":empno, "ename":"테스트"+str(empno), "sal":9000}])

            conn.commit()

    # ------------------------------------
    def start(self):
        while True:
            print("1.전체보기 | 2.특정데이터검색 | 3.데이터넣기 |")
            select = input('메뉴를 선택해주세요 : ')
            if select == '1':
                self.read_all_data()
            if select == '2':
                self.read_specific_date()
            if select == '3':
                self.insert_data()
            if select == '4':
                pass
            if select == '0':
                break

if __name__ == '__main__':
    s = Alchemy()
    s.start()


```
