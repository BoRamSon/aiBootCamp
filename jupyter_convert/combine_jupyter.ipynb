```python
import json
import os
import glob

def combine_notebooks_in_folder(folder_path, output_file):
    # 폴더에서 모든 .ipynb 파일을 찾아요
    notebook_files = glob.glob(os.path.join(folder_path, "*.ipynb"))
    
    # 파일이 없으면 작업을 중단해요
    if not notebook_files:
        print(f"'{folder_path}' 폴더에서 .ipynb 파일을 찾을 수 없어요.")
        return
    
    # 파일 이름을 알파벳 순서로 정렬해요
    notebook_files.sort()
    
    print(f"다음 순서로 파일을 합칠게요:")
    for i, file in enumerate(notebook_files):
        print(f"{i+1}. {os.path.basename(file)}")
    
    # 첫 번째 노트북을 기본으로 사용할게요
    with open(notebook_files[0], 'r', encoding='utf-8') as f:
        combined_notebook = json.load(f)
    
    # 나머지 노트북의 셀들을 하나씩 추가할게요
    for notebook_file in notebook_files[1:]:
        try:
            with open(notebook_file, 'r', encoding='utf-8') as f:
                notebook = json.load(f)
                combined_notebook['cells'].extend(notebook['cells'])
                
                # 노트북 사이에 제목 셀을 추가하여 구분해요
                # file_name = os.path.basename(notebook_file)
                # title_cell = {
                #     "cell_type": "markdown",
                #     "metadata": {},
                #     "source": [f"## 노트북: {file_name}"]
                # }
                # combined_notebook['cells'].insert(-len(notebook['cells']), title_cell)
                
        except Exception as e:
            print(f"'{notebook_file}' 파일을 처리하는 중 오류가 발생했어요: {e}")
    
    # 결과를 새 파일로 저장할게요
    output_path = os.path.join(folder_path, output_file)
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(combined_notebook, f)
    
    print(f"노트북 파일들이 성공적으로 '{output_path}'에 합쳐졌어요!")



# 사용 예시
if __name__ == "__main__":
    # 노트북 파일들이 있는 폴더 경로를 지정하세요
    folder_path = "../../202507/0716"  # 이 부분을 여러분의 폴더 경로로 바꿔주세요
    
    # 결과 파일 이름을 정해주세요
    output_notebook = "99_0716_공부내용.ipynb"
    
    combine_notebooks_in_folder(folder_path, output_notebook)
```
