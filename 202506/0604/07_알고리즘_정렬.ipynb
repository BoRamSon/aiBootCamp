# 🟩 알고리즘 - 정렬  



## 🟢 선택 정렬 (Selection Sort)  

### 🟡 선택 정렬이란?  

> 선택 정렬은 배열에서 **가장 작은(또는 가장 큰)** 값을 **선택**해서  
> **앞쪽으로 하나씩 자리를 정해가는** 정렬 알고리즘입니다.  

### 🟡 선택 정렬의 동작 방식 (오름차순 기준)  

1. **첫 번째 위치에** 올 값을 찾기 위해 **전체 배열에서 가장 작은 값을 찾음**  
2. 그 값을 **첫 번째 값과 교환**  
3. **두 번째 위치**에 올 값을 위해 **남은 값들 중에서** 가장 작은 값을 다시 찾음  
4. **두 번째 값과 교환**  
5. ...이 과정을 **배열 끝까지 반복**  

### 🟡 예시로 보는 선택 정렬  

- 초기 배열:  
  - [5, 2, 9, 1]  

- 1단계  
  - 가장 작은 값: 1  
  - 1 ↔ 5 교환  
  - 결과:  
    - [1, 2, 9, 5]  

- 2단계  
  - 남은 배열: [2, 9, 5]  
  - 가장 작은 값: 2 → 그대로 위치 (교환 X)  
  - 결과: [1, 2, 9, 5]  

- 3단계  
  - 남은 배열: [9, 5]  
  - 가장 작은 값: 5  
  - 5 ↔ 9 교환  
  - 결과: [1, 2, 5, 9]  

<br>

### 🟡 선택 정렬의 특징 요약  
| 항목 | 설명 |  
|------|------|
| **정렬 방식** | 매번 가장 작은 값을 선택하여 앞쪽으로 보냄 |  
| **시간 복잡도** | 항상 O(n²) (최선, 평균, 최악 모두 같음) |  
| **공간 복잡도** | O(1) (추가 메모리 거의 안 씀) |  
| **안정성** | ❌ 불안정 정렬 (같은 값의 순서가 바뀔 수 있음) |  
| **정렬 방향** | 오름차순 / 내림차순 모두 가능 |  
| **특징** | 간단하지만 느림, 배열이 작을 때만 사용 가치 있음 |  


### 🟡 선택 정렬이 사용되는 상황  
| 상황 | 이유 |  
|------|------|
| 배열 크기가 작을 때 | 알고리즘이 단순하므로 빠르게 구현 가능 |  
| 메모리를 아껴야 할 때 | 추가 공간 없이 정렬 가능 (in-place) |  
| 속도가 중요하지 않을 때 | O(n²)이라 느림 |  


### 🟡 선택 정렬 vs 버블 정렬  
| 비교 항목 | 선택 정렬 | 버블 정렬 |  
|-----------|-----------|------------|
| 비교 횟수 | 고정적 (n²) | 많음 (최악 n², 최선 n) |  
| 교환 횟수 | 적음 (필요할 때만 교환) | 비교할 때마다 교환 발생 |  
| 직관성 | 비교적 쉬움 | 아주 직관적 |  
| 속도 | 느림 | 더 느림 |  

```python
# 내가 직접 만들어보기
# 예시 숫자 배열 = [9,7,25,6,8,4,3]

def selection_sort(arr):
  for i in range(0, len(arr)-1):   # 마지막 것은 해줄 필요 없으니, -1을 해준 것
    for j in range(i+1, len(arr)):   # 뒤에꺼를 비교하면서 계속 바꿔나갈 것이다.
      if arr[i] > arr[j]:          
        arr[i], arr[j] = arr[j], arr[i]
        print("변화 관찰하기 = ", arr[i], arr[j])
  print("결과 = ", arr)

arr = [9,7,25,6,8,4,3]
selection_sort(arr)


```
