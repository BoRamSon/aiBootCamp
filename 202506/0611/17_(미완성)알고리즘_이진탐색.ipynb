# 🟩 알고리즘 - 이진탐색  


### 이진탐색과 이진탐색트리는 완전히 다른 내용입니다.  

| 항목           | 이진 탐색 (Binary Search)   | 이진 탐색 트리 (Binary Search Tree) |  
| ------------ | ----------------------- | ----------------------------- |
| **무엇인가요?**   | 정렬된 배열에서 값을 찾는 **알고리즘** | 데이터를 저장하고 관리하는 **자료구조**       |  
| **동작 대상**    | **배열/리스트** (정렬되어 있어야 함) | **트리 구조** (노드로 이루어짐)          |  
| **탐색 방법**    | 중간 인덱스를 기준으로 탐색         | 노드를 왼쪽/오른쪽 자식으로 따라가며 탐색       |  
| **시간복잡도**    | O(log N)                | O(log N) (균형 잡힌 경우)           |  
| **정렬 필요 여부** | 필요함 (입력 배열이 정렬되어야 함)    | 정렬은 자동 관리됨 (삽입 시 위치 결정됨)      |  
| **데이터 수정**   | 없음 (배열 기반)              | 삽입/삭제 가능                      |
