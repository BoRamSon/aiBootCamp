## 🟢 자연어 1  


```python
# 우리가 직접 벡터화를 해보자!
import string

# print("구두점", string.punctuation)


class MyVectorize:
    # 🟡 표준화 함수
    def standardize(self, text):
        text = text.lower()  # 1. 전부 소문자를 만든다.
        return "".join(c for c in text if c not in string.punctuation)
        # 구두점을 제거한 문장을 만들어서 반환함

    # 🟡 토큰화 함수
    def tokenize(self, text):
        return text.split()  # 잘라서 보낸다.

    # 🟡 어휘사전 만들기
    def make_vocab(self, dataset):
        # 1.  전체 데이터셋을 순회하며 단어 사전을 만든다.
        # 2. 기본 단어 사전을 만든다.
        self.vocab = {"<PAD>": 0, "<UK>": 1}
        # 3. 데이터셋을 순회하며 단어 사전을 만든다.
        for text in dataset:
            text = self.standardize(text)
            tokens = self.tokenize(text)
            for text in dataset:
                text = self.standardize(text)
                tokens = self.tokenize(text)
                for token in tokens:
                    if token not in self.vocab:
                        self.vocab[token] = len(self.vocab)

        # 역순서   단어 + 숫자  =>  숫자+단어로 바꾼다.
        self.inverse_vocabulary = dict((v, k) for k, v in self.vocab.items())

    def encode(self, int_sequence):
        return "  ".join(self.inverse_vocabulary.get(i, "[UK]") for i in int_sequence)


my = MyVectorize()

dataset = [
    "I write, erase, rewrite",
    "Erase again, and then",
    "A poppy blooms",
    "Dog is pretty",
]


# 🟡 표준화!
print("========== 표준화 ==========")
text = my.standardize(dataset[0])
print(text)
# 이렇게 표준화를 해보았습니다.

# 🟡 토큰화!
print("\n========== 토큰화 ==========")
text2 = my.standardize(dataset[1])
print(text2)
tokenized = my.tokenize(text2)
print(tokenized)


# 🟡 어휘사전 만들기
print("\n========== 어휘사전 만들기 ==========")
my.make_vocab(dataset)
print(my.vocab)
print(my.inverse_vocabulary)


# 🟡 인코딩!
print("\n========== 인코딩 ==========")
encoded = my.encode([1, 2, 3, 4, 5])
print(encoded)
```

## 🟢 자연어 2  

```python
import re  # 정규식 처리
import string
import tensorflow as tf
from keras.layers import TextVectorization


def custom_standardization_fn(text):
    lower_text = tf.strings.lower(text)
    return tf.strings.regex_replace(
        lower_text, f"[{re.escape(string.punctuation)}]", ""
    )


def custom_split_fn(text):
    return tf.strings.split(text)


# 객체 만들때 파라미터 값들을 지정하면 된다.
text_vectorization = TextVectorization(
    output_mode="int",  # 출력이 시퀀스임
    standardize=custom_standardization_fn,  # 표준화에 사용할 함수를 전달할 수 있다.
    split=custom_split_fn,  # 토큰화에 사용할 함수를 전달할 수 있다
)

dataset = [
    "I write, erase, rewrite",
    "Erase again, and then",
    "A poppy blooms",
    "Dog is pretty",
]

text_vectorization.adapt(dataset)  # 학습시킬 데이터가 았으면 adapt에 전달하면 된다.

vocaburary = text_vectorization.get_vocabulary()
# 알아서 단어 빈도수로 정리됨
print(vocaburary)

# 인코딩
text = "I wruite, rewrite, and still rewrite agian"
encoded = text_vectorization(text)
print(encoded)

# 디코딩
decoded_voca = dict(enumerate(vocaburary))
print(decoded_voca)

decoded_sen = " ".join(decoded_voca[int(i)] for i in encoded)
print(decoded_sen)
```
