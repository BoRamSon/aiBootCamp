### 🟡 2차원 배열

```python
# 행렬 연산

import numpy as np

# 수학적으로 행렬 (matrix)
m1 = np.array([[1,2], [3,4]])
m2 = np.array([[5,6], [7,8]])

"""
<생김새 그려보기>
1   2      5   6   =   1+5    2+6      6   8
3   4      7   8   =   3+7    4+8      10  12 
각각의 자리끼리 서로 계산을 해준다.
"""


# 행렬 덧셈
print('\n-----행렬 덧셈-----')
m3 = m1 + m2   # 행렬 덧셈
print(m3)

# 행렬 인덱싱
print('\n-----행렬 인덱싱-----')
print(m1[0,0]) # 1
print(m1[0,1]) # 2
print(m1[1,0]) # 3
print(m1[1,1]) # 4

# 행렬 슬라이싱
print('\n-----행렬 슬라이싱-----')
print(m1[0]) # [1,2]
print(m1[1]) # [3,4]  

# 행렬 연산 (곱셈, 벡터연산)
print('\n-----행렬 연산 (곱셈, 벡터연산)-----')
print(m1 * m2) # 행렬 곱셈 (각각의 자리끼리 곱셈)
print(m1.dot(m2)) # 행렬 곱셈 (벡터 연산)

# 요소의 크기
print('\n-----요소의 크기-----')
print (m1.shape) # (2,2)
row, col = m1.shape
print(row, col) # 2 2
print( m1.dtype) #데이터 타입 확인하기 - int32

# 행렬 요소 출력
print('\n-----행렬 요소 출력-----')
for i in range(0,row):
    for j in range(0,col):
        print(m1[i,j], ',', end='')
    print()





```

### 🟡 복사의 개념

```python
import numpy as np

# 🟡 range 대신  =>  arange(시작값, 끝값, 증가값)
# for문이나 list 대신 사용할 수 있다.
print('\n-----arange-----')
print( np.arange(1,11))     # 1부터 10까지 1씩 증가
print( np.arange(0,10,0.1)) # 0부터 10까지 0.1씩 증가
print( np.arange(0,10,2))   # 0부터 10까지 2씩 증가


# 🟡 복사하기 - 얕은 복사
print('\n-----얕은 복사-----')
a = np.arange(1,11)
b = a  # 얕은 복사

a[0] = 100
print(a)
print(b)

# 🟡 복사하기 - 깊은 복사
print('\n-----깊은 복사-----')
c = a.copy() # 깊은 복사
c[0] = 100
print(a)
print(c)


# 🟡 얕은 복사와 깊은 복사의 차이점
# 얕은 복사는 원본 데이터를 복사하는 것이 아니라 참조를 복사하는 것이다.
# 깊은 복사는 원본 데이터를 복사하는 것이다.
# 얕은 복사는 원본 데이터를 변경하면 복사한 데이터도 변경된다.
# 깊은 복사는 원본 데이터를 변경하면 복사한 데이터는 변경되지 않는다.
# 얕은 복사는 메모리 사용량이 적다. (참조를 복사하기 때문에)
# 깊은 복사는 메모리 사용량이 많다. (원본 데이터를 복사하기 때문에)
# 얕은 복사는 원본 데이터를 변경하면 복사한 데이터도 변경된다.
# 깊은 복사는 원본 데이터를 변경하면 복사한 데이터는 변경되지 않는다.



```

### 🟡 특정값 채우기 및 random 함수

```python

import numpy as np


# 🟡 메모리 할당을 해서 요소의 값이 0이나 1로 채워진 ndarray 타입 만들기
# N-dimensional array - 1차원부터 N차원까지 모든 형태의 배열
print('\n-----메모리 할당을 해서 요소의 값이 0이나 1로 채워진 ndarray 타입 만들기-----')
a = np.zeros(10) # 10개의 요소를 가진 1차원 배열 만들기
print(a)

b = np.zeros((3,4)) # 3x4 행렬 만들기 (3행 4열)  / 튜플로 만들어줘야 합니다.
print(b)

c = np.ones((3,4)) # 3x4 행렬 만들기 (3행 4열)  / 튜플로 만들어줘야 합니다.
print(c)




# 🟡 rand 함수
# 0부터 1사이의 랜덤 값을 생성
# 가우스분포(정규분포)를 따르는 랜덤 값을 생성
# 요소가 랜덤인 배열 또는 행렬을 생성할 때 사용

print('\n-----random 함수 사용-----')
print(np.random.rand(10))    # 10개의 요소를 가진 1차원 배열 만들기
print(np.random.rand(3,4))   # 3x4 행렬 만들기 (3행 4열)  / 튜플로 만들어줘야 합니다.
print(np.random.rand(3,4,5)) # 3x4x5 행렬 만들기 (3행 4열 5층)  / 튜플로 만들어줘야 합니다.


# 정수발생 np.random.randint(시작값, 끝값, 개수)  low ~ high-1 사이의 정수를 발생
print('\n-----random 정수발생-----')
print(np.random.randint(1,10)) # 1부터 10까지의 정수 발생 (기본 1개 발생)
print(np.random.randint(1,10,5)) # 1부터 10까지의 5개의 정수 발생
print(np.random.randint(1,10, (2,3))) # 2x3 행렬 만들기 (2행 3열)  / 튜플로 만들어줘야 합니다.





```
