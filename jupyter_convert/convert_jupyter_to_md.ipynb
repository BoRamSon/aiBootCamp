# 🟩 nbconvert 라이브러리  

### 🟢 이 명령어는 누구에 의해 만들어졌나요?  

이 명령어는 Jupyter 프로젝트에서 공식적으로 제공하는 CLI (Command Line Interface) 도구인 **nbconvert**에서 나왔습니다.  

Jupyter는 단순히 "코드 실행 환경"이 아니라, 교육/논문/보고서 작성까지 모두 가능한 문서형 시스템입니다.  

그래서 "노트북을 HTML, PDF, Markdown, LaTeX 등으로 변환할 수 있는 기능"을 CLI로도 지원하게 된 것입니다.  

<br>

### 🟢 nbconvert 설치  

pip install nbconvert  

<br>

### 🟢 명령어  

jupyter-nbconvert <파일명.ipynb> --to markdown --output <원하는 명칭.md>  

<br>

### 🟢 명령어 구조 해석  

```bash
jupyter-nbconvert 03_pandas_2.ipynb --to markdown --output 03_pandas_2.md  
```

| 구성 요소           | 의미                                      |  
| ------------------- | ----------------------------------------- |
| `jupyter-nbconvert` | Jupyter의 공식 변환 도구 실행 명령        |  
| `03_pandas_2.ipynb` | 변환 대상 Jupyter Notebook 파일           |  
| `--to markdown`     | Markdown 파일로 변환하라는 옵션           |  
| `--output xxx.md`   | 출력 파일명을 지정 (확장자까지 명시 가능) |  

<br>

# 🟩 Jupyter -> Markdown으로 변환하기  

<br>

## 🟢 1단계: 폴더 안 모든 .ipynb → .md 로 변환  

파이썬 스크립트 사용  


````python
import os
import subprocess

# 폴더 경로 지정
folder_path = '../../202507/0715'
output_file = '99_0715_공부내용.md'
````

````python
# 폴더 내 모든 .ipynb 파일을 반복
for filename in os.listdir(folder_path):
    if filename.endswith('.ipynb'):
        name = filename[:-6]  # 확장자 제거

        input_path = os.path.join(folder_path, filename)
        output_name = name + '.md'
        
        subprocess.run([
            'jupyter-nbconvert',
            input_path,
            '--to', 'markdown',
            '--output', output_name
        ])
    print(f"🟡 {filename} 파일 변환 완료")
````

<br>

## 🟢 2단계: 변환된 .md 파일들을 하나로 합치기  


````python
# md 파일만 수집하고 정렬
md_files = sorted([f for f in os.listdir(folder_path) if f.endswith('.md')])

with open(os.path.join(folder_path, output_file), 'w', encoding='utf-8') as outfile:
    for fname in md_files:
        with open(os.path.join(folder_path, fname), 'r', encoding='utf-8') as infile:
            outfile.write(infile.read())
            # outfile.write('\n\n---\n\n')  # 각 파일 사이 구분선 추가
````
