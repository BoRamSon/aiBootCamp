## 🟢 데이터 필터링

```python
import pandas as pd

df = pd.read_csv('./data/auto-mpg.csv')

# print(df.describe())

print('---------- 8기통 자동차 데이터 및 개수 ------------')
df2 = df[df["cylinders"]==8]
# print(df2)
print(df.shape)
print(df2.shape)

# value_counts() - 카테고리형 2 4 8
print('---------- 기통 데이터 개수 ------------')
print(df2["cylinders"].value_counts())
print(df["cylinders"].value_counts())

print('---------- 8기통 자동차의 mpg 데이터 ------------')
print(df[df["cylinders"]==8]["mpg"])

print('---------- 8기통 자동차의 mpg 데이터 평균 ------------')
print(df[df["cylinders"]==8]["mpg"].mean())


# 모델발표연도가 70이고 연비가 25이상(and, or - python 연산자는 사용 못한다.)
import numpy as np
df3 = df[ np.logical_and(df["model-year"]==70, df["mpg"]>=25) ]
print(df3)

# 모델발표연도가 70이고 연비가 25이상(and, or - python 연산자는 사용 못한다.)  데이터 건수
print(df3.shape)
```

## 🟢 문제 풀어보기

- 모델과 모델별 제품 개수
- 1. 80년대에 나온 제품들은 모델과 개수로 나타내기
- 2. 73, 78, 76년에 나온 제품만 모두 출력하기
- 3. 80년대에 출시한 제품중에서 연비가 25 이상인 제품의 정보만 출력하기
- 4. 70년대에 출시된 모델중에서 실린더가 4개인 제품의 정보만 출력하기
- 5. 80년대에 출시된 모델들의 실린더 개수와 제품개수 출력하기

```python
import pandas as pd
import numpy as np

df = pd.read_csv('./data/auto-mpg.csv')
# print(df.head())



# 1. 70년대에 나온 제품들은 모델과 개수로 나타내기
print('\n---------- 1. 70년대에 나온 제품들은 모델과 개수로 나타내기 ------------')
# df5 = df[df["model-year"]>=80]   # 이거는 80년대 부터 전부 출력이기 때문에 안된다.
df5 = df[ np.logical_and(df["model-year"]>=80, df["model-year"]<=89) ]
print(df5)
print(df5.shape)

# 정답
# print(df[df["model-year"]//10==8].value_counts())  # 모델별 개수 확인



# 2. 73, 78, 76년에 나온 제품만 모두 출력하기
print('\n---------- 2. 73, 78, 76년에 나온 제품만 모두 출력하기 ------------')
# df6 = df[ np.logical_and(df["model-year"]==73, df["model-year"]==76, df["model-year"]==78) ]
df6 = df[df["model-year"].isin([73, 76, 78])]   # 🔥 isin 은 리스트 형태로 출력한다.
print(df6)
print(df6.shape)



# 3. 80년대에 출시한 제품중에서 연비가 25 이상인 제품의 정보만 출력하기
print('\n---------- 3. 80년대에 출시한 제품중에서 연비가 25 이상인 제품의 정보만 출력하기 ------------')
df7 = df[ np.logical_and(df["model-year"]>=80, df["model-year"]<=89) ]
# 🔥 and를 3번 사용해야하는데 numpy and가 2번밖에 안먹히는 문제가 있음.
print(df7)
print(df7.shape)

df8 = df7[df7["mpg"]>=25]
print(df8)
print(df8.shape)

# 정답
# print(df[df["model-year"]//10==8].value_counts())  # 모델별 개수 확인



# 4. 70년대에 출시된 모델중에서 실린더가 4개인 제품의 정보만 출력하기
print('\n---------- 4. 70년대에 출시된 모델중에서 실린더가 4개인 제품의 정보만 출력하기 ------------')
df9 = df[ np.logical_and(df["model-year"]>=70, df["model-year"]<=79) ]
df10 = df9[df9["cylinders"]==4]
print(df10)
print(df10.shape)



# 5. 80년대에 출시된 모델들의 실린더 개수와 제품개수 출력하기
print('\n---------- 5. 80년대에 출시된 모델들의 실린더 개수와 제품개수 출력하기 ------------')
df11 = df[ np.logical_and(df["model-year"]>=80, df["model-year"]<=89) ]
print(df11["cylinders"])    # 실린더 개수???
print(df11.shape)           # 제품 개수???

```

## 🟢 데이터 분석2

```python
import pandas as pd

# 분석에서 데이터가 크게 두종류가 있다  -  분석방법이 다르다

# 연속형 데이터, 평균갑이 중요하다.   연비 40, 30, 25   회귀분석이라고 한다.

# 분연속형 데이터, 범주형, 카테고리, 평균값이 중요하지 않다. 빈도수가 중요하다.

# 발생빈도수 - 실린더수 3 5 4 6 8,  분류분석, 로지스틱회귀, 분류(classifier)


# header가 3번쨰 줄에 있음
data = pd.read_csv('./data/auto-mpg.csv')

# value_counts() - 각 데이별 고유 카운트-빈도수, 발생빈도수를 카운트한다.
print(data["model-year"].value_counts())

# 평균, 최대, 최소 등등 통계량
print('연비평균 : ', data["mpg"].mean())
print('연비최대 : ', data["mpg"].max())
print('연비최소 : ', data["mpg"].min())
print('연비중앙값 : ', data["mpg"].median())
print('연비분산 : ', data["mpg"].var())
print('연비최빈값 : ', data["mpg"].mode())
print('연비표준편차 : ', data["mpg"].std())

print('1사분위수 : ', data["mpg"].quantile(0.25))
print('2사분위수 : ', data["mpg"].quantile(0.5))
print('3사분위수 : ', data["mpg"].quantile(0.75))

```
