### 🟡 linspace 


#### ⚫ 1. 정의 (Definition)
- 특정 범위에서 균등하게 간격을 나눠서 숫자들을 생성하는 함수
- numpy.linspace()는 지정한 시작값과 끝값 사이를 일정한 간격으로 나눠서, 원하는 개수만큼 숫자를 만들어주는 함수입니다.

- 예를 들어:
    - 0부터 1까지 5개 숫자를 일정한 간격으로 나누면 → [0. , 0.25, 0.5, 0.75, 1.]



#### ⚫ 2. 기본 형태 (Syntax)

```python
numpy.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)
```
##### 매개변수 설명 (하나씩 쉽게)
| 매개변수       | 뜻        | 설명 예시                            |
| ---------- | -------- | -------------------------------- |
| `start`    | 시작값      | 숫자 생성 시작점 (예: 0)                 |
| `stop`     | 끝값       | 숫자 생성 끝점 (예: 1)                  |
| `num`      | 숫자의 개수   | 총 몇 개의 숫자를 만들지 (기본: 50개)         |
| `endpoint` | 끝값 포함 여부 | `True`면 stop 포함, `False`면 포함 안 함 |
| `retstep`  | 간격 반환 여부 | `True`면 생성된 숫자 사이 간격도 함께 반환      |
| `dtype`    | 자료형      | float, int 등 직접 지정 가능            |
| `axis`     | 다차원에서 사용 | 거의 안 쓰지만 축 지정 가능                 |



#### ⚫ 4. 기초 예제
- 예제 1: 0부터 1까지 5개의 숫자 만들기
```python
import numpy as np

arr = np.linspace(0, 1, 5)
print(arr)


# 출력
[0.   0.25 0.5  0.75 1.  ]

```
0부터 1까지 5개의 값을 균등하게 생성했습니다.  


```python
arr = np.linspace(0, 1, 5, endpoint=False)
print(arr)

# 출력
[0.  0.2  0.4  0.6  0.8]
```
