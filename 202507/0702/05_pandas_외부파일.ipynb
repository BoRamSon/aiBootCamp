# 🟩 pandas - 외부파일 읽고 쓰기

- Pandas는 다양한 외부 파일 형식을 지원하여 데이터를 읽고(read_*) 쓰기(to_*) 매우 편리하게 처리할 수 있습니다.


### 🟡 Pandas 외부 파일 형식
| 순위 | 파일 형식       | 읽기 함수 (`read_*`)    | 쓰기 함수 (`to_*`)    | 주요 사용 용도                     |
| -- | ----------- | ------------------- | ----------------- | ---------------------------- |
| ①  | **CSV**     | `pd.read_csv()`     | `df.to_csv()`     | 가장 보편적, 텍스트 기반 데이터 저장        |
| ②  | **Excel**   | `pd.read_excel()`   | `df.to_excel()`   | 업무 데이터 공유, 보고서 자료            |
| ③  | **JSON**    | `pd.read_json()`    | `df.to_json()`    | API 응답, 웹 연동 데이터 처리          |
| ④  | **Parquet** | `pd.read_parquet()` | `df.to_parquet()` | 대용량 데이터 저장, 빠른 입출력           |
| ⑤  | **SQL**     | `pd.read_sql()`     | `df.to_sql()`     | 데이터베이스 연동 (예: SQLite, MySQL) |
| ⑥  | **Pickle**  | `pd.read_pickle()`  | `df.to_pickle()`  | 빠른 저장/복원, 모델 결과 저장           |
| ⑦  | **HTML**    | `pd.read_html()`    | `df.to_html()`    | 웹 테이블 크롤링, 보고서 출력            |

## 🟢 파일의 경로 표현 방법


### 🟡 파일의 경로는 절대경로와 상대경로 모두 가능합니다. 
경로를 기술할 경우 주위할 점은 \ 절대경로를 처리 경우에는 c:\pandas_workspace\data\score.csv 처럼 모든 경로를 다 써줘야 하는데 문제는 파이썬에 \는 다른 문자와 결합하여 제어문자로 사용된다는데 있습니다. 
이럴경우에는 경로를 제대로 인식하지 못해 문제가 될 수 도 있습니다. 

### 🟡 파이썬에 이를 해결하는 방법은 3가지가 있습니다. 

#### 1) \ 를 하나 더 붙여서 \\ 로 해야 합니다. 
c:\\pandas_workspace\\data\\score.csv

#### 2) 경로앞에 r을 쓰면 \문자가 가지고 있는 기능이 무력화 됩니다. 
r”c:\pandas_workspace\data\score.csv”

#### 3)마지막은 \(역슬래쉬) 대신에 /(슬래쉬)를 사용하면 됩니다.
윈도우os 는 경로를 나타낼때 \(역슬래쉬)를 사용하고 리눅스 os는 /(슬래쉬)를 사용했는데 요즘은 윈도우 os가 /(슬래쉬)도 지원합니다.

---

### 🟡 1. 상대경로 
- 상대경로는 현재 어플리케이션이  가동중인 폴더를 기준으로 경로를 배정합니다. 
    - . (도트) : 현재 어플리케이션이 가동중인 폴더
    - ..(도트 두개) : 자신보다 하나 위의 폴더를 나타냅니다. 

- 현재 파이썬 파일이 있는 폴더가 uni_10 인데 이 폴더 안에 data 폴더가 있을 경우에는 
    - ./data/score.csv  형태로 기술하는것을 상대경로라 합니다. 

### 🟡 2. 절대경로는 /(루트)부터 모든 경로를 기술하는것을 말합니다 .
- 예) c:/pandas_workspace/uni_10/data/score.csv    c:는 생략해도 됩니다. 

- 절대경로 보다는 상대경로를 사용하는 것이 더 좋습니다. 
- 절대 경로를 사용할 경우 폴더 이동시 모든 경로를 다시 작성해야 하는데 반해서 상대경로는 폴더를 이동해도 굳이 경로를 수정할 필요가 없습니다.






## 🟢 파일 읽기

```python
import pandas as pd

# data 디렉토리에 있는 파일 불러오기와서 데이터 프레임으로 만들기
df = pd.read_csv('./data/score.csv')


# 데이터 프레임으로 만들면 자동으로 행 인덱스가 부여된다.
print('------------------------ 헤더 확인 ------------------------')
print(df.head())

print('\n------------------------ 열 확인 ------------------------')  
print(df.columns)

print('\n------------------------ 행 확인 ------------------------')
print(df.index)

print('\n------------------------ 필드 데이터 2개 추가 ------------------------')
df['total'] = df.kor + df.eng + df.mat  # 키 값에 하이픈이 들어갈 경우 이렇게 쓸 수 없다.
df['avg'] = df.total / 3
print(df)

```

## 🟢 파일 데이터 분석

- 데이터 파일을 읽어서 DataFrame으로 저장

- 데이터의 앞부분과 뒷부분 샘플 출력 (head(), tail())

- 데이터의 행과 열 개수 확인 (shape)

- 데이터 타입과 결측치 포함한 요약 정보 확인 (info())

- 각 수치형 컬럼에 대한 통계 요약량 확인 (describe())

```python
# 분석1.py
import pandas as pd

df = pd.read_csv('./data/auto-mpg.csv')

print('\n------------------------ 헤더 확인 ------------------------')
print(df.head())

print('\n------------------------ 테일 확인 ------------------------')
print(df.tail())

print('\n------------------------ 행과 열 확인 ------------------------')
print(df.shape)
row, col = df.shape
print(f'행의 개수: {row}, 열의 개수: {col}')


# info ->  데이터 정보를 요약해서 보여준다.
print('\n------------------------ 데이터 정보 확인 ------------------------')
print(df.info())
# 마치 db 테이블 정보를 보는 느낌입니다.


# 통계량
print('\n------------------------ 통계량 확인 ------------------------')
print(df.describe())
# mean 평균 
# std 표준편차
# min 최소값
# max 최대값
# count 개수




```
