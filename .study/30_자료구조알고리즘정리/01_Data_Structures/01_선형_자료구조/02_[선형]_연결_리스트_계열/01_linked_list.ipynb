- 정의 : 사람이 손을 잡고 줄을 선 것"이라고 생각하세요.  
  - ex. [철수] → [영희] → [민수]  
    - 철수는 영희 손을 잡고 있고,  
    - 영희는 민수 손을 잡고 있어요.  
    - 이렇게 각 노드(Node)가 다음 노드를 가리키며 연결되는 구조입니다.  

- 단일 연결 리스트 구조  
  - [값 | 다음 노드 주소]  
    - ex. [1 | O] → [3 | O] → [5 | None]  
      - [1 | ●] ──▶ [3 | ●] ──▶ [5 | X]  
      -        next       next     (None)  
        - 1 → 3 → 5 순서로 연결됨  
        - 마지막 노드는 None을 가리켜서 끝임을 표시  
        





### 🟡 Python 리스트의 insert 기능을 수동으로 구현해보기  
- "insert(index, value)의 내부 동작 이해하기"  
- "리스트 삽입 시 발생하는 메모리 이동 과정 직접 구현"  

```python
# arr.insert(3, 88)처럼 삽입(insert) 기능을 직접 수동으로 구현해보았습니다.
# Python 리스트의 내부 동작을 수동으로 따라해봅시다.


arr = list()
#미리 메모리를 10개만 확보한다.
for i in range(0, 10):
  arr.append(i+1)

print('데이터 만들기 : ',arr)


# 3번방에 데이터 88을 넣어보자
# 컴퓨터가 쓸데없는 일을 많이 해야 한다.  overhead
for i in range(9, 3, -1):  # 마치 뒤로 한칸씩 이동한 느낌을 줍니다. 그렇다고 마지막 10이 보이지는 않는다.
  arr[i] = arr[i-1]

arr[3] = 88

print('결과 : ',arr)




```

### 🟡 단일 연결 리스트 클래스를 만들고, 노드 조작하기  

  - appendleft(x): 연결 리스트의 처음에 x를 추가한다.  
  - append(x): 연결 리스트의 끝 x를 추가한다.  
  - popleft(): 연결 리스트에서 첫 노드의 값을 반환하고, 노드는 삭제한다.  
  - pop(): 연결 리스트에서 마지막 노드의 값을 반환하고, 노드는 삭제한다.  
  - insert(i, x): 연결 리스트의 i번 인덱스에 x를 추가한다.  
  - remove(x): 연결 리스트에서 값이 x인 노드를 찾아 삭제한다.  
  - reverse(): 연결 리스트를 제자리에서 순서를 뒤집는다.  




```python
# 단일 링크드 리스트를 더미노드 형태로 코드를 만들고 이해합니다.
# --------------------------------
# 🟡 <초기 상태>
#     head                tail
#   더미 node            더미 node
# [None | next]   →   [data | None]  
# -------------------------------------------------------
# 🟡 <insert_head의 결과>
#     head                                    tail
#   더미 node              node              더미 node       
# [None | next]   →   [data | next]   →   [data | None]  
# -------------------------------------------------------
# [ data: 5 | next: 주소 ] → 다음 노드로 연결
# -----------------------------------------



class Node:   # 노드 하나의 정의
  def __init__(self, data=None):
    self.data = data  # 데이터 파트
    self.next = None  # 다음번 요소의 주소


class SinglyLinkedList:   # 연결 리스트 전체 구조 정의
  # head - 리스트의 시작을 가리킨다.
  # tail - 리스트이 마지막을 가리킨다.
  # ----------------------------------
  #     head                tail
  #   더미 node            더미 node
  # [None | next]   →   [data | None]  
  # ----------------------------------
  def __init__(self):
    self.head = Node()  # 줄의 맨 앞 사람 (첫 노드)
    self.tail = Node()  # 줄의 맨 끝 사람 (마지막 노드)
    self.head.next = self.tail
    self.tail.next = self.tail
    
  # tail에서는 추가하기 힘들고, head에서는 쉽게 추가할 수 있다.
  # tail에서 추가하기 위해서는 head부터 찾아들어가며서 마지막까지 가야 마지막을 알 수 있습니다.
  def insert_head(self, data):    # 3. 값을 리스트에 추가하는 메서드
    # -------------------------------------------------------
    #     head              (추가된 것)              tail
    #   더미 node              node               더미 node      
    # [None | next]   →   [data | next]   →   [data | None]  
    # -------------------------------------------------------
    print(f'Insert Data = {data}')
    # 노드개체를 새로 생성한다.
    new_node = Node(data)
    # 연결 작업을 한다. 단일이기 때문에 2개만 해주면 되죠!!!
    new_node.next = self.head.next   # 먼저해야한다.
    self.head.next = new_node   # 그 다음에 해줘야합니다.
    self.print_all()

  def print_test(self):         # 4. 리스트를 모두 출력하는 메서드
    print(self.head.next.data)
    print(self.head.next.next.data)
    print(self.head.next.next.next.data)

  def print_all(self):          # 4. 리스트를 모두 출력하는 메서드
    # (head의 값과 tail을 값은 움직이면 안된다. 절대 바뀌면 안된다.)
    # 그래서 추적용 node 하나를 선언
    print('🍎 SinglyLinkedList = HEAD'," > ", end='')
    trace = self.head.next
    while trace != self.tail:
      print(trace.data, " > ", end='')
      trace = trace.next        # 이렇게 반복적으로 가면서 tail까지 갑니다.
    print('TAIL', end='')
    print('\n')

  def delete_head(self):
    if self.head.next == self.tail:
      print('🚫 이미 아무것도 없음')
      return    #이미 다 삭제되어서 없음
    # head > ✅ < 어떤 node   =   그냥 주소만 다다음으로 지정해주면 사이에 데이터 추적이 불가해진다.
    # 메모리는 자동 관리된다.
    self.head.next = self.head.next.next
    self.print_all()

  def delete_all(self):
    # 메모리 관리를 알아서 하니깐, 
    self.head.next = self.tail
    # head 다음을 바로 tail로 연결하라는 뜻
    # head -> 노드1 -> 노드2 -> 노드3 -> tail
    # head -> tail   다 사라졌음

  def insert_order(self, data):
    print('<특정 위치에 데이터 넣기>')
    temp = Node(data)
    # 1. 위치 찾기
    t1 = self.head.next   # 현재 노드 (앞으로 갈 노드)
    t2 = self.head        # 이전 노드 (뒤에서 따라가는 노드)

    # 위치를 찾거나, 마지막에 도달 했을 때 끝내야함.
    flag = False   # not flag : flag 값이 False인 동안 반복하라
    
    while not flag and t1 != self.tail:
      if t1.data > temp.data:   # 🔥 지금 보고 있는 노드(t1.data)가 새로 넣을 값(temp.data)보다 크다면,
        flag = True     # 삽입할 위치 찾았으니 반복 중단
      else:
        t1 = t1.next
        t2 = t2.next

    # 2. 노드 연결 - t2와 t1 사이에 temp를 끼운다
    temp.next = t1
    t2.next = temp
    self.print_all()



m1 = SinglyLinkedList()

print("---------------------------------------------")
print("<데이터 넣기>")
m1.insert_head('E')
m1.insert_head('C')
m1.insert_head('B')
m1.insert_head('A')
# head → [C] → [B] → [A] → tail
# 왜 거꾸로 들어가냐면 항상 맨 앞에 넣기 때문입니다.
# m1.print_all()

# print("\n---------------------------------------------")
# print("<데이터 삭제하기>")
# m1.delete_head() # c 삭제
# m1.delete_head() # b 삭제
# m1.delete_head() # a 삭제
# m1.delete_head() # 삭제할게 없음

print("\n---------------------------------------------")
m1.insert_order("D")
# m1.insert_order("B")
# m1.insert_order("C")
# m1.insert_order("D")
# m1.print_all()
```
