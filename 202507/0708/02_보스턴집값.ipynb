## 🟢 sklearn.datasets - fetch_openml - 보스턴 집값


```python
from sklearn.datasets import fetch_openml  
# fetch로 되는 함수들은 너무 커서 별도의 라이브러리를 사용한다.
# 참조 사이트
# https://scikit-learn.org/stable/modules/generated/sklearn.datasets.fetch_openml.html#sklearn.datasets.fetch_openml

# ===================================================
# 데이터 불러오기
    # fetch_openml 함수는 데이터를 불러오는 함수이다.
    # name : 데이터 이름
    # version : 데이터 버전
    # 
boston = fetch_openml(name="boston", version=1) # 데이터 불러오기

# 데이터 확인
print(f"type(boston) : {type(boston)}")
print(f"boston.keys() : {boston.keys()}")
print(f"boston.DESCR : {boston.DESCR}")
print(f"boston.data.shape : {boston.data.shape}")
print(f"boston.target.shape : {boston.target.shape}")
print(f"boston.feature_names : {boston.feature_names}")

x = boston.data  # 특성값
y = boston.target  # 타겟값
    
# 데이터 확인
print(f"x.shape : {x.shape}")
print(f"y.shape : {y.shape}")
print(f"x[:10] : {x[:10]}") # 10개의 특성값
print(f"y[:10] : {y[:10]}") # 10개의 타겟값




```
