## 🟢 DataFrame 열삭제 행삭제

```python
import pandas as pd

data = {
'fruits':['망고', '딸기', '수박', '파인애플'],
'price':[2500, 5000,10000, 7000],
'count':[5, 2, 2, 4],
}

# 기존 DataFrame
df = pd.DataFrame(columns=['fruits', 'price', 'count'])

# 새로운 행 추가
df.loc[len(df)] = {'fruits': '사과', 'price': 3500, 'count': 10}

# print(df)


# ===========================================
# 삭제하기
print('\n--------------price 열삭제-----------------')
df1 = df.drop(columns=['price'], axis=1)  # axis - 축    0 - 행   1 - 열
print(df1)

print('\n--------------원본-----------------')
print(df)

print('\n--------------행삭제------------------')
df2 = df.drop(0, axis=0)
print(df2)

print('\n--------------인덱스 다시 부여------------------')
df2 = df.reset_index()  # 인덱스를 다시 부여해라
print('인덱스 다시 부여 완료')
print(df2)

```

## 🟢 문제 풀어보기

다음 테이블 데이터를 데이터프레임 객체로 만들어 봅시다.  
 
data = """  
X1 X2 X3 X4  
2.9 9.2 13.2 2  
2.4 8.7 11.5 3  
2 7.2 10.8 4  
2.3 8.5 12.3 2  
3.2 9.6 12.6 3  
"""  

위 테이블에 마지막 요소로 다음 데이터를 추가하세요. ( 10, 20, 30, 40)    

위 테이블에 total 필드를 만들고 각 필드들의 합을 구하여 total 필드에 적용하세요.   


```python
import pandas as pd

# data = """
# X1 X2 X3 X4
# 2.9 9.2 13.2 2
# 2.4 8.7 11.5 3
# 2 7.2 10.8 4
# 2.3 8.5 12.3 2
# 3.2 9.6 12.6 3
# """

# 나는 위 데이터를 사용할 줄 모르겠다. 그래서 그냥 데이터를 만들었다.

data = {
    'X1':[2.9, 2.4, 2, 2.3, 3.2],
    'X2':[9.2, 8.7, 7.2, 8.5, 9.6],
    'X3':[13.2, 11.5, 10.8, 12.3, 12.6],
    'X4':[2, 3, 4, 2, 3],
}


# 데이터프레임 객체 생성
df = pd.DataFrame(data)

# 위 테이블에 마지막 요소로 다음 데이터를 추가하세요. ( 10, 20, 30, 40)   
# '행' 추가
print('----------- 행(숫자) 추가 ----------')
df.loc[len(df)] = [10, 20, 30, 40]
# loc = 데이터 접근 인덱싱
    # loc[행, 열]
    # loc[행]  >>>>>  이 행에 len(df)는 total개수를 index로 보면 추가될 수 있는 자리를 의미한다.
    # = loc[5] 이런 느낌이 되는 것이다.
print(df)

# 위 테이블에 total 필드를 만들고 각 필드들의 합을 구하여 total 필드에 적용하세요   
# '열' 추가
print('\n----------- 필드(total)(문자) 추가 ----------')
df['total'] = df['X1'] + df['X2'] + df['X3'] + df['X4']
print(df)

```
