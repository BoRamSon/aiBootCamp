## 🟢 JadenCase  

JadenCase란 모든 단어의 첫 문자가 대문자이고, 그 외의 알파벳은 소문자인 문자열입니다.  
단, 첫 문자가 알파벳이 아닐 때에는 이어지는 알파벳은 소문자로 쓰면 됩니다.  
(첫 번째 입출력 예 참고)  

문자열 s가 주어졌을 때, s를 JadenCase로 바꾼 문자열을 리턴하는 함수, solution을 완성해주세요.  

- 제한 조건  
  - s는 길이 1 이상 200 이하인 문자열입니다.  
  - s는 알파벳과 숫자, 공백문자(" ")로 이루어져 있습니다.  
  - 숫자는 단어의 첫 문자로만 나옵니다.  
  - 숫자로만 이루어진 단어는 없습니다.  
  - 공백문자가 연속해서 나올 수 있습니다.  

```python
def is_in_length(s):
  if len(s) >= 1 and len(s) <= 200:
    return False
  print('문자열 길이가 200 이상으로 초과했습니다.')
  return True

def is_start_num(s):
  for i in range(len(s)):
    print(f"확인 : {i}")
    print(f"확인 : {s[i][1:]}")
    for j in s[i][1:]:
      if j.isdigit():
        print('숫자가 단어의 첫 글자에만 있지 않습니다.')
        return True
  return False

def convert_upper(s):
  result = []
  for i in s:
    if not i[0].isdigit():
      result.append(i[0].upper() + i[1:])
    else : 
      result.append(i)
  end = ' '.join(result)
  return end

def solution(s):
    if is_in_length(s):
      return False
    value = s.lower()
    print(value)
    value = value.split(' ')
    print(value)
    value2 = []
    for i in value:
        value2.append(i.replace(" ", ""))
    if is_start_num(value2):
      return False
    result = convert_upper(value2)
    print(result)
    print(type(result))
    return result

target = "3people    Unfollowed Me"  # 공백 존재에 대해서 해결하기
# target = "for the last week"	
solution(target)
```

### 🟡 제출  

```python
def solution(s):
    words = s.split(' ')
    answer_list = []
    for word in words:
        if not word:
            answer_list.append('')
        else:
            jaden_word = word[0].upper() + word[1:].lower()
            answer_list.append(jaden_word)
    return ' '.join(answer_list)
```

N개의 아이템이 있다. 각 아이템은 다음과 같은 속성을 가진다:  
value (가치): 아이템을 배낭에 넣었을 때 얻게 되는 이익  
weight (무게): 아이템의 무게  

이때, 각 아이템은 분할해서 넣을 수 있다.  
가치를 최대로 담았을때 어느 정도 까지 담을 수 있는지?  

목적: 배낭에 담을 수 있는 아이템들의 조합 중 총 가치(value)의 합을 최대화하는 것  
가치/무게 비율로 아이템 정렬 (내림차순)  
하나씩 넣되, 전체를 다 넣을 수 없으면 남은 용량만큼 비례해서 담는다.  
누적 가치 합산 → 최종 정답  

```python

backpack = 50
print(f"backpack : {backpack}")
items = [(60, 10), (100, 20), (120, 30)]  # 나중에는 이거 정렬까지 해야합니다.

items = sorted(items, key=lambda x: x[0], reverse=True)

best_value = 0



# 나중에 for문으로 처리할 수 있도록 만들어야 합니다.  (이거는 굉장히 규칙적이네요.)

best_value += items[2][0] * int(backpack / items[2][1])
print(best_value)

backpack = backpack - (items[2][1] * int(backpack / items[2][1]))
print(backpack)

best_value += items[1][0] * int(backpack / items[1][1])
print(best_value)
backpack = backpack - (items[1][1] * int(backpack / items[1][1]))
print(backpack)

best_value += items[0][0] * int(backpack / items[0][1])
print(best_value)
backpack = backpack - (items[0][1] * int(backpack / items[0][1]))
print(backpack)



print(f"best value : {best_value}")
```
