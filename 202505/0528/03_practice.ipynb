```python
from sqlalchemy import create_engine

theEngine = create_engine(
    "mysql+pymysql://root:qhfkal0513@localhost:3306/sakila",
    pool_size=10,  # 최대 연결 수
    max_overflow=5,  # 초과 시 추가 연결 수
    pool_recycle=3600,  # 재활용 시간
)
```

```python
from sqlalchemy import text

def statistics():
    with theEngine.begin() as conn:
        sql = """
            SELECT 
            concat(C.last_name, " ", C.first_name) AS name,
            A.postal_code, A.address, A.district, A.phone
            FROM customer C
            INNER JOIN  address A ON C.address_id = A.address_id
            LIMIT 10;
        """
        result = conn.execute(text(sql)).mappings().all()
        # print(result)
        # name / postal_code / address / district / phone
        for i in result:
            print(f"{i['name']} | {i['postal_code']} | {i['address']} | {i['district']} | {i['phone']}")

statistics()
```
