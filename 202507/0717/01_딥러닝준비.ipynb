## 🟢 Deep Learning 준비하기  

### 🟡 Tensorflow와 Keras가 무엇인가?  

| 항목    | TensorFlow                                                | Keras                                |  
| ----- | --------------------------------------------------------- | ------------------------------------ |
| 정의    | Google에서 만든 **딥러닝 프레임워크**                                 | TensorFlow 위에서 동작하는 **고수준 API (도구)** |  
| 역할    | 신경망을 구성, 학습, 예측할 수 있는 **엔진**                              | 신경망을 쉽게 만들 수 있도록 도와주는 **간편한 인터페이스**  |  
| 난이도   | 복잡하고 강력함 (코드가 길고 구조적)                                     | 간단하고 직관적 (코드가 짧고 이해하기 쉬움)            |  
| 사용 예시 | 산업용, 연구용에 많이 사용                                           | 교육, 프로토타이핑, 간단한 실험에 많이 사용            |  
| 현재 관계 | **Keras는 TensorFlow 안에 포함되어 있음** (`tensorflow.keras`로 사용) | 별도 설치 없이 TensorFlow만 설치하면 같이 사용 가능   |  

#### ⚫️ TensorFlow는 왜 중요한가?  
| 기능         | 설명                                                                 |  
| ---------- | ------------------------------------------------------------------ |
| GPU/TPU 지원 | 그래픽카드나 AI용 칩으로 학습 속도 매우 빠름                                         |  
| 자동 미분      | 수학적 계산을 자동으로 해줌 (역전파 자동 처리)                                        |  
| 유연성        | 복잡한 모델 구조, 사용자 정의 연산 등 고급 기능 가능                                    |  
| 배포 가능      | 학습한 모델을 **모바일, 웹, 서버**에 배포 가능 (예: TensorFlow Lite, TensorFlow\.js) |  


#### ⚫️ Keras는 왜 좋은가?  
| 이유             | 설명                             |  
| -------------- | ------------------------------ |
| 쉬운 문법          | 초보자도 딥러닝 개념만 알면 바로 모델을 만들 수 있음 |  
| 빠른 실험          | 모델을 바꾸고 테스트하기 매우 쉬움            |  
| 직관적 구조         | 마치 레고처럼 층을 쌓는 방식으로 모델 구성       |  
| TensorFlow와 통합 | 안정성과 성능 모두 확보됨 (`tf.keras` 사용) |

### 🟡 딥러닝 이론  

| 이론 항목                               | 쉬운 요약 설명                   | 상세 설명                                                                                                              |  
| ----------------------------------- | -------------------------- | ------------------------------------------------------------------------------------------------------------------ |
| **퍼셉트론 (Perceptron)**               | 입력을 계산해서 출력하는 작은 뇌세포 같은 단위 | 입력값에 가중치를 곱하고 편향을 더한 뒤, 그 결과를 활성화 함수에 통과시켜 출력하는 딥러닝의 가장 기본 구성 요소입니다. 이 퍼셉트론들이 여러 개 연결되어 신경망(Neural Network)을 이룹니다. |  
| **가중치와 편향 (Weights & Bias)**        | 얼마나 중요하고 어디로 치우칠지를 조절하는 값  | 가중치는 입력값이 결과에 미치는 영향 정도를, 편향은 전체 결과를 위로 또는 아래로 이동시키는 역할을 합니다. 학습을 통해 조정되며, 결과의 정확도를 높이기 위해 매우 중요합니다.               |  
| **활성화 함수 (Activation Function)**    | 뉴런이 켜질지 말지를 결정하는 스위치       | 입력값을 받아서 어떤 식으로 출력할지를 결정합니다. 선형함수로는 복잡한 패턴을 학습할 수 없기 때문에, 비선형 함수인 Sigmoid, ReLU 등이 주로 사용됩니다.                       |  
| **시그모이드 함수 (Sigmoid)**              | 값을 0\~1로 바꿔주는 곡선 함수        | 출력값을 0과 1 사이로 제한해서 이진 분류 문제에 자주 사용됩니다. 하지만 출력이 너무 작아지는 문제(기울기 소실)가 있어 요즘엔 ReLU가 더 많이 쓰입니다.                         |  
| **ReLU 함수 (Rectified Linear Unit)** | 0보다 크면 그대로, 작으면 0으로 출력     | 매우 간단하지만 효과적인 함수로, 빠르게 학습되며 현재 딥러닝에서 가장 널리 사용됩니다. 기울기 소실 문제도 적습니다.                                                 |  
| **손실함수 (Loss Function)**            | 예측이 얼마나 틀렸는지 알려주는 계산기      | 모델의 예측과 실제값 사이의 차이를 수치로 표현합니다. 이 값이 작을수록 모델이 잘 학습하고 있다는 뜻이며, 학습은 이 값을 줄이는 방향으로 진행됩니다.                              |  
| **경사하강법 (Gradient Descent)**        | 틀린 정도를 조금씩 줄여가는 방법         | 손실함수를 최소화하기 위해 가중치와 편향을 어떻게 바꿔야 할지 계산하는 알고리즘입니다. 마치 산을 내려가며 가장 낮은 지점을 찾는 것과 비슷합니다.                                 |  
| **역전파 (Backpropagation)**           | 틀린 원인을 거꾸로 따라가서 고치는 과정     | 출력에서부터 입력 방향으로 오차를 전파하여 각 층의 가중치와 편향을 얼마나 조정해야 할지 계산합니다. 딥러닝 학습의 핵심입니다.                                            |  
| **옵티마이저 (Optimizer)**               | 학습을 똑똑하게 도와주는 조정기          | 경사하강법을 구현하는 다양한 방식이 있으며, 대표적으로 SGD, Adam, RMSprop 등이 있습니다. Adam은 보통 가장 많이 사용됩니다.                                   |  
| **에포크 (Epoch)**                     | 전체 데이터를 한 번 다 학습하는 것       | 한 에포크는 전체 학습 데이터를 모델이 한 번 다 본 것입니다. 일반적으로 여러 번(수십\~수백 번) 반복해서 학습시킵니다.                                              |  
| **배치 (Batch)**                      | 데이터를 나눠서 조금씩 학습하는 단위       | 데이터를 모두 한 번에 학습시키는 대신, 여러 개의 작은 묶음으로 나눠서 학습합니다. 이렇게 하면 메모리도 절약되고 학습도 더 안정적입니다.                                     |  
| **지도학습 (Supervised Learning)**      | 정답을 알려주며 학습하는 방법           | 입력과 정답(Label)을 함께 주고 학습합니다. 예: "이 사진은 강아지야!"라고 가르쳐주는 방식. 분류, 회귀 등에 사용됩니다.                                          |  
| **비지도학습 (Unsupervised Learning)**   | 정답 없이 스스로 패턴을 찾는 학습        | 입력만 주고, 모델이 데이터 안의 구조나 패턴을 스스로 학습하게 합니다. 예: 비슷한 뉴스 기사 묶기, 이미지 압축 등.                                                |  
| **과적합 (Overfitting)**               | 외운 건 잘하지만 새로운 건 못 푸는 모델    | 학습 데이터에는 정확하지만, 새로운 데이터에는 성능이 떨어지는 현상입니다. 학습을 너무 많이 하거나, 데이터가 부족할 때 발생합니다.                                         |  
| **정규화 (Regularization)**            | 모델이 너무 외우지 않게 제어하는 기법      | 과적합을 방지하기 위한 기법으로, 대표적으로 L1/L2 정규화가 있습니다. 모델이 너무 복잡해지는 걸 막아줍니다.                                                    |  
| **드롭아웃 (Dropout)**                  | 일부 뉴런을 임시로 꺼서 과적합 방지       | 학습 중에 무작위로 일부 뉴런을 꺼서 지나치게 특정 패턴만 학습하지 않도록 합니다. 성능 향상에 매우 효과적입니다.                                                   |  
| **배치 정규화 (Batch Normalization)**    | 학습을 빠르고 안정적으로 만드는 비법       | 각 층의 입력을 평균 0, 분산 1로 맞춰서 학습이 잘 되도록 도와줍니다. 일반적으로 학습 속도가 빨라지고, 성능도 좋아집니다.                                            |  
| **하이퍼파라미터 튜닝**                      | 모델 학습 설정값을 최적화하는 작업        | 학습률, 배치 크기, 에포크 수 등은 사람이 설정하는 값이며, 실험을 통해 가장 좋은 값을 찾아야 합니다.                                                        |  
| **CNN (합성곱 신경망)**                   | 이미지 인식에 특화된 딥러닝 구조         | 이미지를 처리할 때 주변 픽셀 간의 관계를 잘 포착할 수 있어서 얼굴 인식, 자율주행 등에 사용됩니다.                                                          |  
| **RNN (순환 신경망)**                    | 시간의 흐름이 있는 데이터를 잘 처리함      | 텍스트, 음성, 주가처럼 순서가 중요한 데이터에 사용됩니다. 이전 정보를 기억할 수 있도록 설계되었습니다.                                                        |  
| **LSTM / GRU**                      | RNN보다 똑똑한 기억력 있는 구조        | RNN의 문제점(기억 상실)을 해결한 구조로, 긴 문장도 잘 이해할 수 있게 만들어졌습니다.                                                                |
