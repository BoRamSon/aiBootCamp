
# 🟩 NumPy (넘파이)

## 🟢 개념
NumPy는 Numerical Python의 줄임말로, 고성능의 수치 계산을 위한 라이브러리입니다.  
**다차원 배열 객체(ndarray)**를 중심으로, 수학, 통계, 선형대수, 푸리에 변환 등 다양한 기능을 제공합니다.


## 🟢 핵심 특징
| 항목        | 설명                            |
| --------- | ----------------------------- |
| `ndarray` | N차원 배열 객체. 빠르고 효율적인 데이터 저장 구조 |
| 브로드캐스팅    | 배열 간의 연산을 자동으로 맞춰 수행          |
| 벡터화 연산    | 반복문 없이 전체 배열에 연산 적용 가능        |
| 다양한 함수    | 선형대수, 난수, 통계 등 수치 계산 함수 내장    |


## 🟢 예시
```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)        # 배열 간 원소별 덧셈
print(a * 2)        # 배열 전체에 곱셈
print(np.mean(a))   # 평균


```

# 🟩 pandas (판다스)

## 🟢 개념
pandas는 **표 형식의 데이터(데이터프레임)**를 쉽게 다룰 수 있는 라이브러리입니다.
엑셀처럼 행과 열로 구성된 데이터를 읽고, 쓰고, 정리하고, 분석할 수 있게 도와줍니다.


## 🟢 핵심 구조
| 구조          | 설명                     |
| ----------- | ---------------------- |
| `Series`    | 1차원 데이터 (컬럼 하나)        |
| `DataFrame` | 2차원 데이터 (엑셀 시트처럼 행과 열) |


## 🟢 주요 기능
- CSV, Excel, SQL 등 다양한 데이터 파일 읽기/쓰기
- 결측치 처리 (NaN 제거, 채우기 등)
- 필터링, 정렬, 그룹화, 피벗 등 강력한 데이터 조작 기능
- 통계 요약 및 시계열 처리


## 🟢 예시
```python
import pandas as pd

# 딕셔너리를 데이터프레임으로 변환
data = {'name': ['홍길동', '이몽룡'], 'age': [30, 25]}
df = pd.DataFrame(data)

print(df)
print(df['age'].mean())   # 평균 나이 계산

```
