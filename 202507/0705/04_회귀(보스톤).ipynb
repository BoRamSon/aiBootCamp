## 🟢 회귀 분석 : 보스톤

다중회귀분석 : 공분산(특성 간에 서로 영향을 주고 받는 것)을 따져보고 필요없는 특성은 제거하는게 맞다.

R은 기본적으로 제거해준다.

파이썬은 우리가 해야한다.


```python
from heapq import heapreplace
import pandas as pd
import numpy as np

url = "http://lib.stat.cmu.edu/datasets/boston"

# 분리문자가 공백이 아니고, \s+, skiprows=22 줄 건너뛰기 제목 줄이 없다.
df = pd.read_csv(url, sep='\s+', skiprows=22, header=None)
# print(df.head(10))

# np에 hstack 함수가 있음, 수평방향으로 배열을 
# 짝수행에 홀수를 갖다가 붙인다. df.values[::2, : ] 짝수행 0,2,4,6,8 : 전체 컬럼
# 홀수행에 앞에 열 2개만 df.values[1::2, :2] 홀수행 1,3,5,7,9 : 앞에 2개 컬럼
X = np.hstack( [df.values[::2, : ], df.values[1::2, :2]] )
print(X.shape, X[:10])

Y = df.values[1::2, 2]  # 이 열이 target
print(Y.shape, Y[:10])  # 데이터가 잘 못 왔을 수 있으니 꼭 shape를 확인해보아야한다.


# ========================================================
# 또 데이터를 훈련셋과 테스트셋으로 나누자!!!!
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state=42)

# 선형 회귀 모델을 만든다 = LinearRegression
from sklearn.linear_model import LinearRegression
model = LinearRegression()   # 선형 회귀 모델을 만든다
model.fit(X_train, Y_train)  # 훈련셋을 넣어서 모델을 훈련시킨다.
print("LinearRegression")
print("훈련셋 점수 : ", model.score(X_train, Y_train))
print("테스트셋 점수 : ", model.score(X_test, Y_test))

# 기울기와 절편을 찍어보자
print("기울기들 : ", model.coef_)
print("절편 : ", model.intercept_)
```

- 선형회귀분석 : 다중공선성문제, 여러 특성간에 서로 너무 밀접해서 필요없는 요소들 이걸 고려하를 않는다.
-
- 특성의 개수가 많을 경우에 처리능력이 떨어진다.
- 보스톤 주택가격데이터의 특성의 개수는 13개임, 즉 가중치도 13개가 나온다.

- 누군가가 가중치를 규제하면 과대적합을 막을 수 있지 않을까?
- 가중치가 너무 크거나 작으면 훈련테이터셋에 초점이 맞춰진다. = 과대적합
- 가중치를 규제하자 / 라쏘는 가중치를 0에 가깝게 하다가 불필요한 요소가 있으면 아예 0으로 만들기도 한다.

- 모델을 심플하게 만든다. L2 정규화
- 리지는 계수를 완전히 0으로 만들지는 못한다. L1 정규화
- 하리퍼파라미터 alpha라는 값이 있는데, 이걸 0으로 놓으면 규제를 아무것도 안하겠다. LinearRegression하고 똑같음 alpha를 키우면 규제가 높아진다.

- 라쏘하고 리지라는 알고리즘이 있다.
- 라쏘는 쓸데없는 계수 (기울기들)


```python
from sklearn.linear_model import Ridge
model = Ridge(alpha=0.1) 
model.fit(X_train, Y_train)  # 훈련셋을 넣어서 모델을 훈련시킨다.
print("Ridge")   # 0에 한없이 가까이 가기는 하는데 실제로 0으로 가지는 않는다.
print("훈련셋 점수 : ", model.score(X_train, Y_train))
print("테스트셋 점수 : ", model.score(X_test, Y_test))

# 기울기와 절편을 찍어보자
print("기울기들 : ", model.coef_)
print("절편 : ", model.intercept_)


print("================================================")


from sklearn.linear_model import Lasso
model = Lasso(alpha=0.1)    # 숫자가 커질수록 규제가 커진다.
model.fit(X_train, Y_train)  # 훈련셋을 넣어서 모델을 훈련시킨다.
print("Lasso")
print("훈련셋 점수 : ", model.score(X_train, Y_train))
print("테스트셋 점수 : ", model.score(X_test, Y_test))

# 기울기와 절편을 찍어보자
print("기울기들 : ", model.coef_)
print("절편 : ", model.intercept_)


# 🔥 alpha 값이 적절해야 한다.
# 머신러닝 알고리즘은 하이퍼파라미터가 있다. => 우리가 수작업으로 찾아야 한다.
    # 이걸 자동으로 찾으면 딥러닝 이다.
# 이미지, 흑백사진의 경우는 머신러닝이나 딥러닝이나 비슷
```
