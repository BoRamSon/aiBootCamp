# 🟩 Stack 문제 풀어보기 1  
  > **괄호의 짝이 맞는지 확인하기**  
    괄호의 짝이 바르면 True, 바르지 않으면 False를 반환하는 함수를 작성하라. <br> 예를 들어 ((a*(b+c))-d) / e는 괄호의 짝이 올바르지만, (((a*(b+c))-d) / e 는 괄호의 짝이 맞지 않는다. 괄호는 소괄호(())만 사용한다.  


#### ⚫️ 스택(Stack) 자료구조에서 사용하는 기본 연산(메서드, 함수)  
- **is_full()**: 스택이 가득 찼는지 확인하는 작업 (가득 찼으면 True, 주로 용량 제한이 있을 때 사용)  
- **is_empty()**: 스택이 비어있는지 확인하는 작업 (비었으면 True)  
- **push(x)**: 자료 x를 스택에 넣는 작업 (삽입)  
- **pop()**: 스택에서 가장 마지막에 넣은 자료를 꺼내는 작업 (삭제)  
- **peek()**: 스택의 가장 마지막 자료를 확인하는 작업 (꺼내지 않고 보기)  



```python
# stack 클래스 가져오기
class Stack:
    def __init__(self, size):        
        self.size = size 

        self.stack = []
        for i in range(0, self.size): 
            self.stack.append(0)

        self.top = -1  

    # is_full
    def is_full(self):
        if self.size-1 == self.top:
            return True
        return False

    # is_empty
    def is_empty(self):
        if self.top == -1:
            return True
        return False

    # push
    def push(self, data):
        if self.is_full():
            print('이미 가득 참')
            return 
        self.top += 1
        if len(self.stack) == 0:
            self.stack.append(data)
        else:   
            self.stack[self.top] = data  

    # peek
    def peek(self):  # 스택의 가장 마지막 자료를 확인하는 작업 (꺼내지 않고 보기)  
        if self.is_empty():
            return None
        return self.stack[self.top]  # 맨 뒤 값을 반환 (삭제는 안 함)

    # pop
    def pop(self):
        if self.is_empty():
            return None

        return self.stack.pop()  # 맨 뒤에서 꺼냄  # 🔥 이게 원래 주석처리임
                                    # ⛔️ 이 코드는 잘 못된 코드입니다.

        # item = self.stack[self.top] 
        # self.top -= 1
        # return item

    # print_all
    def print_all(self):
        i = 0 
        while i <= self.top:
            print(self.stack[i], end=" ")
            i+=1
        print()



    

# 🟩 문제 풀기
# 괄호의 짝이 바르면 True, 바르지 않으면 False를 반환하는 함수를 작성하라.
# 예를 들어 ((a*(b+c))-d) / e는 괄호의 짝이 올바르지만, (((a*(b+c))-d) / e 는 괄호의 짝이 맞지 않는다. 괄호는 소괄호(())만 사용한다. 

def collect_bracket(target):
    size = target.count('(')
    stack = Stack(size)
    for i in target:
        if i == '(':
            stack.push(i)
        elif i == ')':
            stack.pop()
    print("짝의 결과 = ", stack.stack)
    if len(stack.stack) == 0:
        return True
    return False

target = '(((a*(b+c))-d)'
collect_bracket(target)


```
