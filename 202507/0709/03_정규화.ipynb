## 🟢 정규화
머신러닝에서 **정규화(Normalization)**는 매우 중요한 개념입니다. 데이터의 **스케일(범위)**을 조정하여, 모델이 더 잘 학습되도록 도와주는 전처리 기법 중 하나예요.




### 🟡 지도학습·비지도학습·강화학습 상세 설명
<br>

### 1. 지도학습 (Supervised Learning)

#### 1.1 개념
- 입력 데이터(Input)와 정답(Label) 쌍을 이용하여 관계를 학습합니다.  
- 목표: **새로운 입력에 대한 정확한 예측**.

#### 1.2 학습 과정
1. 데이터 수집  
2. 전처리(결측치 처리, 정규화 등)  
3. 학습·검증·테스트 세트 분할  
4. 모델 학습(손실 최소화)  
5. 검증 및 하이퍼파라미터 튜닝  
6. 최종 테스트 및 배포

#### 1.3 대표 알고리즘 (세부 설명 추가)

- **회귀(Regression) 계열**
  - **선형회귀 (Linear Regression)**
    - 핵심: 입력과 출력의 선형 관계를 가정, MSE 최소화.
    - 파라미터: 정규화 방식(Ridge/Lasso), α 값.
    - 장단점: 해석 용이·학습 빠름 / 비선형 데이터 부적합.
    - 활용: 가격 예측, AB 테스트 효과 추정.
  - **Lasso / Ridge / Elastic Net**
    - L1, L2, 혼합 규제 적용으로 과적합 완화·변수 선택.
    - α(규제 강도), l1_ratio(Elastic Net) 조정이 핵심.
  - **Gradient Boosting Regressor (XGBoost, LightGBM, CatBoost)**
    - 순차적 오차 보정, 학습률(eta)·트리 깊이 등 수십 개 파라미터.
    - 소량 튜닝으로도 SOTA 급 성능. Kaggle·실무 최다 사용.

- **분류(Classification) 계열**
  - **로지스틱 회귀 (Logistic Regression)**
    - 시그모이드/소프트맥스로 확률 예측, BCE(Cross-Entropy) 최적화.
    - 정규화(C, penalty)·클래스 가중치(class_weight)가 주 파라미터.
    - 단순·빠르며 베이스라인으로 좋음.
  - **서포트 벡터 머신 (SVM)**
    - 최대 마진 초평면, 커널트릭(RBF, Poly).
    - C(규제), γ(커널 폭) 튜닝으로 성능 좌우.
    - 소규모·고차원에 강함 / 대용량 데이터 느림.
  - **K-NN**
    - 학습 대신 예측 시 거리 기반 다수결.
    - K 값·거리 척도·가중치 방식이 성능 결정.
    - 구현 간단 / 고차원·대용량 시 속도·정확도 저하.
  - **결정트리 & 랜덤포레스트**
    - 결정트리: 정보이득 기준 분할, 최대 깊이·min_samples_leaf 설정.
    - 랜덤포레스트: 배깅+랜덤 분할, 트리 개수·max_features로 과적합 제어.
    - 변수 중요도 제공, 해석 쉬움.
  - **Gradient Boosting (XGBoost·LightGBM·CatBoost)**
    - 트리 기반 부스팅, 강력한 성능·결측치 자동 처리.
    - 학습률·트리 수·subsample 등이 핵심 파라미터.
  - **딥러닝 (MLP, CNN, RNN 등)**
    - MLP: 완전연결층, 비선형(ReLU). 범용 함수 근사.
    - CNN: 합성곱으로 공간 패턴 추출(이미지·음성).
    - RNN/LSTM/GRU: 순차 데이터(텍스트·시계열) 모델링.
    - 층 수·노드 수·학습률·배치 크기·옵티마이저(Adam 등) 중요.

#### 1.4 장점
- 정답이 있어 **안정적 성능** 확보 용이  
- 평가 지표(정확도, MSE 등)가 명확

#### 1.5 단점
- **대량의 라벨링** 비용·시간 소요  
- 라벨 품질에 민감

#### 1.6 활용 사례
- 스팸 메일 분류, 이미지 인식, 주가 예측, 질병 진단

## 2. 비지도학습 (Unsupervised Learning)

#### 2.1 개념
- **정답 라벨이 없는** 데이터에서 **잠재 구조·패턴**을 탐색합니다.

#### 2.2 학습 과정
1. 데이터 준비  
2. 목표 설정(군집화, 차원 축소 등)  
3. 알고리즘 학습  
4. 결과 해석·검증

#### 2.3 대표 알고리즘 (세부 설명 추가)

- **군집화(Clustering)**
  - **K-Means**
    - 중심(centroid) 기반, 군집 내 제곱 오차 최소화.
    - K(군집 수), 초기화 방법(k-means++) 조정.
    - 대규모 데이터에 빠르나 비구형 군집·잡음에 약함.
  - **DBSCAN**
    - 밀도 기반, eps·minPts가 군집 형성 기준.
    - 임의 형태·잡음 식별 / 고차원 데이터 성능 저하.
  - **계층적 군집**
    - Agglomerative 병합 방식, 덴드로그램으로 클러스터 트리 시각화.
    - 거리 행렬 O(n²) 메모리 필요 → 대용량에 부적합.

- **차원 축소(Dimensionality Reduction)**
  - **PCA**
    - 분산 최대화 선형 투영, 고유값 분해.
    - 주성분 수 혹은 누적 설명 분산 비율로 차원 결정.
  - **t-SNE**
    - 국소 거리 보존 임베딩, KL Divergence 최소화.
    - perplexity, learning_rate 설정 민감‧대용량 느림.
  - **UMAP**
    - 위상 공간 근사 + 그래프 레이아웃, t-SNE 대비 빠르고 글로벌 구조 보존.
    - n_neighbors, min_dist 파라미터 직관적.

- **생성·재구성 모델**
  - **오토인코더(AE) / 변분 오토인코더(VAE)**
    - Encoder–Decoder 구조, 잠재 공간 학습.
    - bottleneck 크기·노이즈 주입(Denoising AE)으로 표현력 조정.
    - 차원 축소·이상 탐지·데이터 생성에 사용.

- **연관 규칙 학습**
  - **Apriori / FP-Growth**
    - 지지도·신뢰도 기반 빈발 아이템 집합 탐색.
    - FP-Growth가 메모리 사용 효율+속도 우수.

- **이상 탐지(Anomaly Detection)**
  - **Isolation Forest**
    - 랜덤 분할 트리로 고립 정도 계산.
    - 트리 개수·샘플 비율 조정, 대규모·고차원에 적합.

#### 2.4 장점
- **라벨링 불필요** → 데이터 확보 용이  
- 탐색적 데이터 분석에 유용

#### 2.5 단점
- 평가 지표가 모호·주관적  
- 결과 해석 난이도 존재

#### 2.6 활용 사례
- 고객 세분화, 토픽 모델링, 이상 거래 탐지, 이미지 압축

## 3. 강화학습 (Reinforcement Learning)

#### 3.1 개념
- **에이전트**가 환경과 상호작용하며 **보상**을 최대화하도록 학습합니다.  
- 동적 의사결정 문제에 특화.

#### 3.2 핵심 요소
| 요소 | 설명 |
|------|------|
| 상태(State) | 환경의 현재 정보 |
| 행동(Action) | 에이전트의 선택 |
| 보상(Reward) | 행동 결과 신호 |
| 정책(Policy) | 상태→행동 매핑 |
| 가치함수(Value) | 장기 보상 기대값 |

#### 3.3 학습 흐름
1. 초기 상태 진입  
2. 정책에 따라 행동 선택  
3. 환경이 다음 상태·보상 반환  
4. 정책/가치함수 업데이트  
5. 종료 조건까지 반복

#### 3.4 대표 알고리즘 (세부 설명 추가)

- **가치 기반(Value-Based)**
  - **Q-Learning**
    - 테이블/함수 근사로 Q(s,a) 갱신, 오프폴리시.
    - 학습률 α, 할인율 γ, ε-greedy 탐험.
    - 단순하지만 연속 상태·액션 처리 어려움.
  - **DQN (Deep Q-Network)**
    - Q-Learning + CNN 함수 근사, 경험 재플레이·타깃 네트워크로 안정화.
    - 리플레이 버퍼 크기, 타깃 네트워크 교체주기 설정 필요.

- **정책 기반(Policy-Based)**
  - **REINFORCE**
    - 정책 πθ(a|s) 직접 최적화, 고분산 Monte-Carlo 방식.
    - 학습률, 배치 크기, 베이스라인 사용 여부(variance 감소) 중요.
  - **A2C / A3C (Advantage Actor-Critic)**
    - Actor-Critic 병렬 학습, n-step 리턴으로 안정·속도↑.
    - 엔트로피 보너스 계수, 스레드 수(A3C) 조정.

- **근사 정책 최적화**
  - **PPO (Proximal Policy Optimization)**
    - 클리핑(clip range) 또는 KL 패널티로 업데이트 폭 제한.
    - 학습률, clip range, GAE(Generalized Advantage Estimation) λ 조정.

- **연속 행동(Continuous Action)**
  - **DDPG (Deep Deterministic Policy Gradient)**
    - Actor-Critic + deterministic 정책, 연속 제어.
    - OU 노이즈, 타깃 네트워크 τ, 버퍼 크기.
  - **SAC (Soft Actor-Critic)**
    - 엔트로피 보너스 포함, 자동 온도 α 조정으로 탐험-활용 균형.
    - 샘플 효율·안정성에서 최신 SOTA.

- **모델 기반(Model-Based)**
  - **AlphaZero / MuZero**
    - MCTS + 신경망(정책·가치·동적 모델) 공동 학습.
    - 대규모 연산 자원 필요하지만 장기적 전략 최적화 탁월.

#### 3.5 장점
- **명시적 라벨 불필요**, 보상만 있으면 학습 가능  
- 게임·로봇 제어 등 **순차적 의사결정** 문제 해결

#### 3.6 단점
- 높은 샘플·연산 비용  
- 보상 설계 어려움 → 의도치 않은 행동 가능

#### 3.7 활용 사례
- AlphaGo/AlphaZero, 로봇 제어, 자율주행, 실시간 광고, 포트폴리오 운용

## 4. 세 방법 비교

| 구분 | 지도학습 | 비지도학습 | 강화학습 |
|------|----------|------------|----------|
| 필요 데이터 | 입력 + 정답 | 입력만 | 환경, 보상 |
| 목표 | 정확한 예측 | 패턴 발견 | 누적 보상 극대화 |
| 대표 문제 | 분류·회귀 | 군집화·차원 축소 | 순차 의사결정 |
| 평가 지표 | 정확도, MSE 등 | 군집 품질 등 | 평균 보상, 승률 |
| 장점 | 안정적·객관적 | 라벨 비용 無 | 동적 문제 해결 |
| 단점 | 라벨 비용 큼 | 해석·평가 난해 | 학습 비용 高 |

## 5. 학습 형태 선택 시 고려 사항

- **데이터 라벨 유무·품질**  
- **문제 유형**: 예측 vs 패턴 탐색 vs 순차 의사결정  
- **실험 환경**: 시뮬레이터 존재 여부(강화학습)  
- **평가 지표** 설계 가능성  
- **자원·시간**: 강화학습은 특히 고비용

## 6. 요약

- **지도학습**: 라벨이 있을 때 가장 안정적이며 예측 성능이 우수합니다.  
- **비지도학습**: 라벨 없이 데이터 구조를 탐색하고, 전처리·이상 탐지에 유용합니다.  
- **강화학습**: 동적 환경에서 장기 보상을 최적화하는 데 강력하지만, 높은 계산 비용과 보상 설계 난이도가 있습니다.

세 방법은 상호 보완적이므로 **문제 특성·데이터·자원**을 종합적으로 고려하여 적절한 학습 방식을 선택하시기 바랍니다.

---

```python
import pandas as pd

#스케일링을 하기위한 가짜데이터 만들기
data = {
    "feature1" : [160, 180, 170, 190, 165, 190, 172, 168, 175, 182],
    "feature2" : [3000, 3200, 3100, 3300, 3050, 3250, 3150, 3350, 3075, 3275],
    "feature3" : [3, 2, 1, 4, 4, 6, 12, 13, 11, 6]
}

df = pd.DataFrame(data)
print(df)


# ================================================

#StandardScalar : 데이터가 정규분포를 따른다고 가정할때 많이 쓰임, 모델특성이 스케일링에 민감할때 
#                 서포트벡터머신, 로지스틱, 딥러닝 일때 유용하다 , 이상치에 민감하다.  

from sklearn.preprocessing import StandardScaler, RobustScaler, MinMaxScaler, Normalizer
# preprocessing(전처리)는 머신러닝이나 데이터 분석에서 모델에 데이터를 입력하기 전에 데이터를 정제하고 변환하는 과정입니다.
# 대표적인 전처리 작업에는 결측치 처리, 이상치 처리, 인코딩, 정규화, 표준화 등이 있습니다.
# 이 예제에서는 정규화(표준화)를 통해 각 feature의 값의 범위를 맞추어 모델의 성능을 높이고, 특정 feature가 과도하게 영향을 미치는 것을 방지합니다.

# StandardScaler는 데이터의 각 feature(특성)를 평균이 0, 표준편차가 1이 되도록 변환(표준화)해주는 전처리 도구입니다.
# 즉, 각 feature의 값에서 평균을 빼고 표준편차로 나누어줍니다.
# 이렇게 하면 서로 다른 단위를 가진 feature들도 동일한 스케일로 맞춰져서,
# 머신러닝 모델이 특정 feature에 치우치지 않고 학습할 수 있게 도와줍니다.
# 특히, 거리 기반 알고리즘(예: KNN, SVM, 선형회귀 등)에서 효과적입니다.

ss = StandardScaler() #객체 생성 
df_scaled = ss.fit_transform(df)  #학습하고 바로 변경된값 반환 
print(df_scaled)

#RobustScaler : 데이터에 이상치에 많으면 StandardScaler 대신 쓸 수 있다. 
rb = RobustScaler()
df_scaled = rb.fit_transform(df)  #학습하고 바로 변경된값 반환 
print(df_scaled)

#MinMaxScaler : 특성값의 범위가 명확히 0~1 사이에 와야할때(이미지,특정신경망)
rb = MinMaxScaler()
df_scaled = rb.fit_transform(df)  #학습하고 바로 변경된값 반환 
print(df_scaled)

#Normalizer - 주로 텍스트분석에 유용하다. 클러스터링(군집분석)
rb = Normalizer()
df_scaled = rb.fit_transform(df)  #학습하고 바로 변경된값 반환 
print(df_scaled)

#별짓 다해도 데이터가 많아야 한다. 
```
