#### ⚫ `형태소` 란  
**형태소(Morpheme)**는 **"의미를 가지는 가장 작은 말의 단위"**를 말합니다.  

- 📌 **예시 1**  
  - 문장: 먹었습니다  
  - 형태소 단위로 나누면:  
      - 먹 (동사 어간)  
      - 었 (과거 시제 선어말 어미)  
      - 습니다 (공손한 종결 어미)  


#### ⚫ 주요 형태소 분석기 종류  
| 분석기명                   | 특징                  | 장점                     | 단점                        |  
| ---------------------- | ------------------- | ---------------------- | ------------------------- |
| Okt (Open Korean Text) | 트위터 형태소 분석기 기반      | 빠르고 비교적 정확, 신조어 인식 가능  | 일부 문법적 오류 존재              |  
| Kkma                   | 서울대 연구실 개발          | 자세한 품사 태그, 문장 단위 분석 가능 | 느린 속도                     |  
| Komoran                | 자바 기반, 사용자 사전 추가 가능 | 비교적 빠르고 정확             | 상대적으로 무거움                 |  
| Hannanum               | 한나눔 형태소 분석기         | 안정적, 사전 업데이트 꾸준히 이루어짐  | 다소 오래된 품사 체계 사용           |

### 🟡 tag는 품사 태깅 예시 보기  

- tag는 품사 태깅(POS tagging) 기능 관련 모듈
- Okt(Open Korean Text)는 konlpy.tag 안에 있는 하나의 형태소 분석기 클래스입니다.

```python
from konlpy.tag import Okt

okt = Okt()
text = "아버지가 방에 들어가신다."

# 형태소 단위 토큰화
morphs = okt.morphs(text)
print(morphs)
# ['아버지', '가', '방', '에', '들어가', '신다', '.']

# 품사 태깅
pos = okt.pos(text)
print(pos)
# [('아버지', 'Noun'), ('가', 'Josa'), ('방', 'Noun'), ('에', 'Josa'), ('들어가', 'Verb'), ('신다', 'Eomi'), ('.', 'Punctuation')]

# 명사만 추출
nouns = okt.nouns(text)
print(nouns)
# ['아버지', '방']
```

## 🟢 konlpy.tag - Okt, Kkma 비교

- tag는 품사 태깅(POS tagging) 기능 관련 모듈

| 분석기      | 특징                                   |
| -------- | ------------------------------------ |
| Okt      | SNS용 문장 처리에 강함. 빠르고 경량화됨. 비정형 문장에 적합 |
| Kkma     | 구문 분석까지 가능. 정확도 높지만 느림               |
| Hannanum | KAIST에서 개발. 빠름. 일부 단어 처리에 약간 부족함     |
| Komoran  | 성능 우수, 실시간 처리에 적합                    |
| Mecab    | 매우 빠르고 정확함 (별도 설치 필요)                |


```python
from konlpy.tag import Kkma

msg = """
오픈소르를 이용하여 형태소 분석을 배워봅시다. 형태소 분석을 지원하는 라이브러리가 많습니다.
각자 어떻게 분석하는지 살펴보겠습니다. 이건 Kkma 모듈입니다.
"""

kkma = Kkma()
print(kkma.sentences(msg))  # <- 문장으로 나눈다
print(kkma.morphs(msg) )     # <- 형태소로 나눈다 
print(kkma.nouns(msg))        # <- 명사를 추출한다 
print(kkma.pos(msg))           # <- 품사태깅을 한다.


print("\n\n")

from konlpy.tag import Okt
twitter = Okt()
print("--- 형태소로 분해 --- ")
print(twitter.morphs(msg)) 
print("--- 명사분해 --- ")
print(twitter.nouns(msg))
print("--- 품사태깅 --- ")
print(twitter.pos(msg))


```
