# BFS

```python
from collections import deque

# 정말 어렵다.

# 큐를 사용한다. 방문위치를 


# 큐를 사용한다.


N, M = map(int, input().strip())
arr = []
for i in range(N):
    temp = list(input())  # str -> lsit
    temp = list(map(int, temp))
    arr.append(temp)

def printArray(arr, N):
    for i in range(N):
        print(arr[i])


visited = [ [False] * M for _ in range(N) ]

def bfs(y, x):  #시작값 y행, x열 
    dx = [-1, 1, 0, 0, 0]
    dy = [ 0,0, -1, 1] 
    visited[y][x] = True # 방문을 했음
    # bfs는 큐를 활용한 탐색기법이다.
    
q = deque()
q.append((y,x))
while q: # 큐가 빌 때까지
    for i in range(len(dy)):
        # 새로운 좌표 만들기
        ny = dy[i] + y
        nx = dx[i] + x
        if not (0 <= nx < M and 0 <= ny < N):
            continue
        if arr[ny][nx]==1:
            continue
        if not visited[ny][nx]:
            visited[ny][nx] = True
            # 큐에 방금 좌표값 추가
            q.append((ny, nx))

bfs(0,0)
printArray(visited)


```
