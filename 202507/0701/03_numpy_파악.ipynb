# 🧠 NumPy 완전 정리

---

## 🟢 1. NumPy 기본 구조

| 개념     | 설명                                 |
|----------|--------------------------------------|
| `ndarray` | N차원 배열 객체 (NumPy의 핵심 구조)  |
| `np.array()` | 리스트를 ndarray로 변환             |
| `dtype`   | 배열의 자료형 (정수형, 실수형 등)     |
| `shape`   | 배열의 차원 (행, 열 등)               |
| `ndim`    | 배열의 차원 수                        |
| `size`    | 전체 원소 개수                        |
| `itemsize`| 각 원소의 메모리 크기 (바이트 단위)  |

### 🟡 예시

```python
import numpy as np
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)    # (2, 3)
print(arr.ndim)     # 2
print(arr.dtype)    # int64
```

## 🟢 2. 배열 생성 함수

| 함수                          | 설명                              |
|-------------------------------|-----------------------------------|
| `np.array()`                  | 리스트 → ndarray                   |
| `np.zeros(shape)`            | 0으로 채운 배열 생성               |
| `np.ones(shape)`             | 1로 채운 배열 생성                 |
| `np.full(shape, value)`      | 지정한 값으로 채움                 |
| `np.eye(n)`                  | 단위 행렬 (대각선 1)               |
| `np.arange(start, stop, step)`| 연속된 값 생성 (range와 비슷함)   |
| `np.linspace(start, stop, num)`| 일정 간격으로 나눈 배열 생성      |
| `np.random.rand(shape)`      | 0~1 사이 난수                      |
| `np.random.randint(low, high, size)` | 정수 난수 배열              |

---

## 🟢 3. 배열 연산

| 연산               | 설명                |
|--------------------|---------------------|
| `+`, `-`, `*`, `/` | 기본 수학 연산      |
| `np.sum()`         | 전체 합             |
| `np.mean()`        | 평균                 |
| `np.std()`         | 표준편차             |
| `np.min()` / `np.max()` | 최소/최대값     |
| `np.dot(a, b)`     | 행렬 곱             |
| `np.transpose(a)` / `a.T` | 전치행렬      |

---

## 🟢 4. 인덱싱 & 슬라이싱

| 문법        | 설명                       |
|-------------|----------------------------|
| `a[i]`      | 1차원 인덱스              |
| `a[i, j]`   | 2차원 인덱스              |
| `a[:, 0]`   | 모든 행의 첫 번째 열      |
| `a[1, :]`   | 두 번째 행의 모든 열      |
| `a[::2]`    | 간격을 두고 슬라이싱      |

---

## 🟢 5. 조건 & 불리언 연산

| 함수/문법                        | 설명                           |
|----------------------------------|--------------------------------|
| `arr > 3`                        | 조건 배열 (`True`, `False`)   |
| `np.where(condition, x, y)`     | 조건을 만족하면 x, 아니면 y    |
| `np.logical_and(x, y)`          | 두 조건 모두 참일 때           |
| `np.logical_or(x, y)`           | 둘 중 하나 참이면              |
| `np.any()` / `np.all()`         | 하나라도 참 / 전부 참인지      |

---

## 🟢 6. 배열 변형

| 함수                    | 설명                         |
|-------------------------|------------------------------|
| 🟢`reshape()`             | 배열 모양 바꾸기 N차원으로 변경            |
| `flatten()` / `ravel()` | 1차원으로 변환               |
| `transpose()` / `.T`    | 전치 (행 <-> 열)             |
| `resize()`              | 크기 변경 (원본 변경)        |
| `expand_dims()`         | 차원 추가                    |
| `squeeze()`             | 차원 축소                    |

## 🟢 7. 배열 결합 / 분할

| 함수                                | 설명                      |
|-------------------------------------|---------------------------|
| `np.concatenate([a, b], axis=0)`   | 축 기준으로 붙이기        |
| `np.stack([a, b], axis=0)`         | 새로운 차원으로 결합       |
| `np.hstack()` / `vstack()`         | 수평/수직 결합             |
| `np.split(a, 3)`                   | 배열을 균등하게 나누기     |
| `np.array_split()`                 | 비균등 분할                |

## 🟢 8. 난수 생성

|---------------------------|-----------------------------|

| `np.random.shuffle()`     | 섞기 (원본 변경)            |
| `np.random.permutation()` | 섞기 (새 배열 반환)         |
