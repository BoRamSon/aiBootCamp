## 🟢 나온 단어  

- RNN (Recurrent Neural Network(리커런트 뉴럴 네트워크))  
    - RNN은 "기억을 할 수 있는 뇌" 같은 인공지능이에요. 보통의 인공지능은 "지금 한 번만 보고 판단" 하지만, RNN은 "이전에도 뭘 봤는지 기억하면서 판단" 해요.  
    - 예를 들어 "나는 학교에..." 까지 썼으면 그 다음은 보통 "갔다" 같은 말이 어울려요. 이런 식으로 앞에 있던 말을 기억해서 다음 말을 예측하는 것, 이게 바로 RNN이 잘하는 일이에요!  
    - 일반 신경망의 경우 매번 입력을 새로 받아서 판단함.. 과거를 기억 못함.  
    - 시간 순서대로 이어지는 정보 (문장, 음악, 날씨 예측 등)를 잘 처리해요.  
    - 단점 : 기억력이 너무 약함 (멀리 있는 과거는 잘 잊어버림 😅)... 그래서 더 발전된 형태인 LSTM, GRU 같은 모델이 나왔어요.
