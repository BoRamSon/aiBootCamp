# 🟩 Database 이론 수업


## 1. 데이터베이스란?

- 여러 사용자가 **공유**하고 사용할 수 있는 **데이터의 집합**
- 목적: **중복 최소화**, **무결성 유지**, **효율적인 검색/저장**

## 2. DBMS (Database Management System)

- 데이터베이스를 **생성, 관리, 검색, 보안** 등을 수행하는 **소프트웨어**
- 예시: MySQL, PostgreSQL, Oracle, SQLite, MongoDB 등

## 3. 데이터 모델

### ✅ 관계형 모델 (가장 널리 사용됨)
- 데이터를 **행(Row)**과 **열(Column)**의 형태로 저장 (테이블 구조)
- **SQL** 언어를 통해 조작

## 4. 스키마 (Schema)

> 데이터베이스의 **구조(설계도)**

### 📌 예시:
```sql
CREATE TABLE Students (
  student_id INT PRIMARY KEY,
  name VARCHAR(100),
  birth_date DATE
);
```

## 5. 키(Key)

| 키 종류               | 설명                   |
|  | -- |
| 기본키(Primary Key)   | 행을 유일하게 식별하는 키       |
| 후보키(Candidate Key) | 기본키 후보가 될 수 있는 속성들   |
| 대체키(Alternate Key) | 후보키 중 기본키로 채택되지 않은 키 |
| 외래키(Foreign Key)   | 다른 테이블의 기본키를 참조하는 키  |


## 6. 무결성 제약 조건 (Integrity Constraints) 

| 제약조건        | 설명                  |
| ----------- | ------------------- |
| NOT NULL    | 값이 반드시 존재해야 함       |
| UNIQUE      | 중복 값 허용 불가          |
| PRIMARY KEY | NOT NULL + UNIQUE   |
| FOREIGN KEY | 외래키 설정              |
| CHECK       | 조건 제한 (ex. age > 0) |



## 7. 정규화 (Normalization)

| 정규형                   | 조건                             | 목적 및 특징                    |
|  |  | -- |
| **제1정규형 (1NF)**       | 모든 컬럼이 \*\*원자값(Atomic)\*\*만 가짐 | 반복 그룹 제거, 하나의 셀에는 하나의 값만   |
| **제2정규형 (2NF)**       | 1NF 만족 + **부분 함수 종속 제거**       | 기본 키의 일부분에만 종속된 컬럼 제거      |
| **제3정규형 (3NF)**       | 2NF 만족 + **이행적 함수 종속 제거**      | 기본 키가 아닌 다른 속성에 종속된 컬럼 제거  |
| **BCNF (보이스-코드 정규형)** | 3NF 만족 + **모든 결정자가 후보키**       | 결정자가 후보키가 아닌 경우 제거         |
| **제4정규형 (4NF)**       | BCNF 만족 + **다치 종속 제거**         | 한 키에 여러 독립적인 다중값이 있는 경우 제거 |
| **제5정규형 (5NF)**       | 4NF 만족 + **조인 종속성 제거**         | 데이터 분해 후 재조인이 가능하도록        |


## 8. SQL (Structured Query Language)

| 분류  | 명령어                            | 설명                   |
| --- | ------------------------------ | -------------------- |
| DDL | CREATE, DROP, ALTER            | 데이터 정의 (테이블 생성/삭제 등) |
| DML | SELECT, INSERT, UPDATE, DELETE | 데이터 조작               |
| DCL | GRANT, REVOKE                  | 권한 제어                |
| TCL | COMMIT, ROLLBACK               | 트랜잭션 제어              |


## 9. 트랜잭션(Transaction)과 ACID
- 하나의 작업 단위. 모두 성공하거나 모두 실패해야 함

| 요소                | 설명              |
| ----------------- | --------------- |
| 원자성 (Atomicity)   | 모두 성공하거나 모두 실패  |
| 일관성 (Consistency) | 무결성 제약 조건 유지    |
| 독립성 (Isolation)   | 동시 트랜잭션 간 간섭 없음 |
| 지속성 (Durability)  | 성공한 트랜잭션은 영구 저장 |


## 10. 데이터베이스 설계 절차
1. 요구사항 분석
2. 개념적 설계 (ERD 등)
3. 논리적 설계 (정규화, 키 설정)
4. 물리적 설계 (인덱스, 저장 방식 등)
5. 구현 (SQL로 테이블 생성)

## 🟢 python 코드 작성 1

````python
import pymysql
import pymysql.cursors


# MySQL 서버에 연결을 설정한다.
# pymysql.connect()는 MySQL 서버에 접속할 때 필요한 정보들을 넣는다.
conn = pymysql.connect(
    host="localhost",  # MySQL 서버 주소 (로컬 PC면 'localhost' 또는 '127.0.0.1')
    user="root",  # MySQL 사용자 이름 (기본은 'root')
    password="",  # MySQL 사용자 비밀번호 (여기에 실제 패스워드를 입력해야 함)
    db="mydb",  # 접속하려는 데이터베이스 이름 (예: 'mydb')
    port=3306,  # MySQL 포트 번호 (기본값 3306)
)

# 커서를 통해
print("접속 성공")
````

### 🟡 문제 해결
실제 MySQL Server 설정 비밀번호로 입력되어야 합니다.  


````python
# 연결된 데이터베이스에서 SQL 명령을 실행할 수 있는 '커서'를 만든다.
cursor = conn.cursor()


# SQL 쿼리문을 문자열로 작성한다.
# 여기서는 'emp' 테이블의 모든 데이터를 가져오는 쿼리이다.
sql = "select * from emp"

# 작성한 SQL 쿼리를 실행한다.
cursor.execute(sql)
# 이 시점에서 select 결과는 cursor 내부에 임시 저장된다.

# cursor에 저장된 결과 전체를 가져온다.
rows = cursor.fetchall()

# 결과를 한 줄씩 반복하면서 출력한다.
for row in rows:
    print(type(row), row)
    # 각 row는 tuple 형태로 출력된다. (예: (8001, '홍길동', 'MANAGER', ...))
````

````python
import pymysql.cursors

print("======== 하나만 가져오기 ========")
sql = "select * from emp where empno=8001"
cursor.execute(sql)
row = cursor.fetchone()
print(row)
````

````python
import pymysql.cursors

sql = "select * from emp where empno<8000"
cursor.execute(sql)
rows = cursor.fetchmany(3)  # 앞에 3개만
for row in rows:
    print(row)
````
