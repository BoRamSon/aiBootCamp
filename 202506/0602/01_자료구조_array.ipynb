### 🟡 문제 - 정수 배열에서 가장 큰 두 수를 찾기  
- 정수로 이루어진 배열이 주어질 때, 가장 큰 두 수를 찾아 [가장 큰 값, 둘째로 큰 값]을 반환하는 함수를 완성하라.  
  - 예시  
    - 입력: [3, -1, 5, 0, 7, 4, 9, 1], 출력: [9, 7]  
    - 입력: [7], 출력: [7]  


```python
# 단순하게 구현하기

def sort_list():
  # input_value = input('숫자를 입력해주세요(','콤마로 구분) :' )
  input_value = "3, -1, 5, 0, 7, 4, 9, 1"
  convert_list = [int(x.strip()) for x in input_value.split(',')]
  if len(convert_list) <= 1:
    return convert_list[0]
  sorted_list = sorted(convert_list, reverse=True)
  return [sorted_list[0], sorted_list[1]] 


sort_list()
```

```python
# 같이 풀어보기

def getMax1Max2(arr:list)->tuple:
  """
    함수에 대한 설명 써주기
  """
  
  # 하나밖에 없다는 그게 가장 큰 값이겠지... 종료!
  if len(arr) < 2:
    return arr
  
  # 내가 원하는 값은 가장 큰값과 그 다음 큰값
  # max1, max2 = arr[:2]
  max1 = arr[0]
  max2 = arr[1]

  # 🔥 그런데.. max2의 값이 더 크다면... 위치를 바꿔주라는 의미 (지속적인 변화를 준다.)
  if max1 < max2:
    max1, max2 = max2, max1
  print('현재값 확인 :', max1, max2)

  # 위에 max 2개를 해줬을 때... 저 2개의 값보다 더 큰 값이 존재하는지 확인을 하고 대치시켜줘야지??
  for i in range(0, len(arr)):
    if max1 < arr[i]:
      max1 = arr[i]
  print("가장 큰 값 = ",max1)

  arr.remove(max1)
  for i in range(0, len(arr)):
    if max2 < arr[i]:
      max2 = arr[i]
  print("두번째로 큰 값 = ", max2)
      

input_value = "3, -1, 5, 0, 7, 4, 9, 1"  # 나중에 input으로 입력 받아서 한다고 생각하면 됩니다.
arr = [int(x.strip()) for x in input_value.split(',')]
getMax1Max2(arr)

```

### 🟡 문제 - 회문(Palindromes) 찾기  

```python
# 쉬운 코드

word = "racecar"

if word == word[::-1]:
    print(True)
else:
    print(False)

#결과값 : True
```

```python
# 하나하나 직접 비교해보기

def palindromes(word):
  for i in range(1, len(word)+1):
    if word[i-1] == word[len(word)-i]:
      print(f"{word[i-1]} | {word[len(word)-i]}  = 잘 맞습니다.")
    else:
      return print('회문(Palindromes)이 아닙니다.')
  return word

word = "racecar"    
palindromes(word)
```

```python
# 같이 풀어보기

import math

def palindromes2(s):
  ln = len(s)-1
  # 6   6/2  -  3
  for i in range(0, math.ceil(ln//2)):
    if s[i] != s[ln-i]:
      return False
  return True

print( palindromes2('madam'))
print( palindromes2('tomato'))
print( palindromes2('abba'))
```
