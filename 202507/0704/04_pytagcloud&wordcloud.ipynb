# 🟩 tokenizer (nltk) 사용하기

- 토크나이저는 텍스트(문장, 문서)를 작은 단위(토큰, token)로 쪼개는 도구입니다.

- 여기서 토큰(token)은
  - 보통 단어(word)
  - 혹은 형태소(morpheme)
  - 혹은 문장부호 등 의미를 갖는 최소 단위


## 🟢 nltk(토크나이저) - KoNLPy, wordcloud 사용

토큰이 문장에서 단어를 하나씩 분리  
토크나이저 - 문장에서 단어를 분리해내는 라이브러리

### 🟡 영어로 출력


```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import nltk  # 영어 자연어 처리 모듈
from collections import Counter    # 단어 개수
from konlpy.corpus import kolaw     # 한국 법률 말뭉치
from konlpy.tag import Okt  # 한국 형태소 분석기
import matplotlib.pyplot as plt
import numpy as np
from wordcloud import ImageColorGenerator
from PIL import Image
import random

nltk.download('punkt_tab')
# 다운로드 완료 (windows에서는 현재 여기에 저장되었다고 함.)
# [nltk_data] Downloading package punkt to
# [nltk_data]     C:\Users\user\AppData\Roaming\nltk_data...
# [nltk_data]   Unzipping tokenizers\punkt.zip.

from nltk.tokenize import sent_tokenize     # 문장으로 쪼재주는 토크나이저
from nltk.tokenize import word_tokenize     # 단어로 쪼재주는 토크나이저

# 파일을 읽어와서 토큰화 하는 것입니다.
c = open('./data/alice.txt', 'r', encoding='utf-8').read()
tokens_en = word_tokenize(c)
print(tokens_en)

# 각 단어들의 갯수를 세는 것입니다.
data = Counter(tokens_en)
print(data)

# 앞에서 100개만 추려서 쓰자!
data = data.most_common(100)
temp_data = dict(data)

# 워드클라우드 생성
font_path = 'C:/Windows/fonts/malgun.ttf'
wordcloud = WordCloud(font_path='C:/Windows/fonts/malgun.ttf', width=800, height=400, background_color='white').generate_from_frequencies(temp_data)

# 이미지 저장
wordcloud.to_file('image2.png')

# 이미지 정보를 리턴
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()


```
