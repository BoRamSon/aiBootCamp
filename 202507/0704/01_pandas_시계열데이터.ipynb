## 🟢 pd.to_datetime 
pd.to_datetime("2025-07-01")  
pd.to_datetime(["2025-07-01", "2025/07/02", "20250703"])  
"날짜(또는 시간)로 변환해줘!"  

#### 📌 목적:
- 문자열, 숫자, 객체 등을 Datetime 형식으로 변환

#### 📥 입력:
- 문자열 "2025-07-01"
- 리스트 ["2025-07-01", "2025-07-02"]
- Series, DataFrame의 열

#### 📤 출력:
- Timestamp 또는 DatetimeIndex

```python
import pandas as pd 
import numpy as np  

date_str = ["2019-10-01", "2019-10-02", "2019/10/03", "20191004"]
date_obj = pd.to_datetime(date_str, format="mixed") 
print(date_obj)


```

## 🟢 pd.date_range
pd.date_range(start="2025-07-01", periods=5, freq='D')  
"시작일과 간격, 개수를 기준으로 날짜 리스트를 만들어줘!"  

#### 📌 목적:
일정한 주기(freq)로 구성된 Datetime 시계열 생성  

#### 📥 입력:
- 시작일(start) 또는 종료일(end)
- 기간 수(periods)
- 간격(freq)

#### 📤 출력:
DatetimeIndex (날짜 리스트)

### 🟡 frequency 인수 – 주요 주기 옵션 정리표

| **코드**                 | **단위**      | **설명**                    | **예시 값**                          |
| ---------------------- | ----------- | ------------------------- | --------------------------------- |
| `'Y'`                  | 연           | 매년 (12월 31일 기준)           | `'2025-12-31'`                    |
| `'YS'`                 | 연 시작        | 매년 (1월 1일 기준)             | `'2025-01-01'`                    |
| `'Q'`                  | 분기 끝        | 매 분기 마지막 날 (3, 6, 9, 12월) | `'2025-03-31'`, `'2025-06-30'`    |
| `'QS'`                 | 분기 시작       | 매 분기 첫날 (1, 4, 7, 10월)    | `'2025-01-01'`, `'2025-04-01'`    |
| `'M'`                  | 월 끝         | 매월 마지막 날                  | `'2025-01-31'`, `'2025-02-28'`    |
| `'MS'`                 | 월 시작        | 매월 첫날                     | `'2025-01-01'`, `'2025-02-01'`    |
| `'W'`                  | 주           | 매주 일요일 기준                 | `'2025-07-06'`                    |
| `'W-MON'`, `'W-TUE'` 등 | 주(요일 지정)    | 지정된 요일마다 한 주              | `'W-FRI'` → 매주 금요일                |
| `'D'`                  | 일           | 매일                        | `'2025-07-01'`, `'2025-07-02'`    |
| `'B'`                  | 평일(비즈니스 데이) | 월\~금만 포함                  | `'2025-07-01'`, `'2025-07-02'`    |
| `'H'`                  | 시간          | 매시간                       | `'2025-07-01 01:00'`              |
| `'T'`, `'min'`         | 분           | 매분                        | `'2025-07-01 01:30'`              |
| `'S'`                  | 초           | 매초                        | `'2025-07-01 01:30:15'`           |
| `'L'`                  | 밀리초         | 1/1,000초                  | `'2025-07-01 01:30:15.123'`       |
| `'U'`                  | 마이크로초       | 1/1,000,000초              | `'2025-07-01 01:30:15.123456'`    |
| `'N'`                  | 나노초         | 1/1,000,000,000초          | `'2025-07-01 01:30:15.123456789'` |


```python
import pandas as pd

print('---------- 일 기준 반환 ---------')
today = pd.date_range(start='2025-01-01', end='2025-01-31', freq='D')
print(today)


print('---------- 분 기준 반환 ---------')
d = pd.date_range(start='2025-07-04', periods=30, freq='T')
print(d)
# periods는 반환할 날짜or분or주 등등의 개수를 지정한다.
# periods 값을 주어지지 않으면 기본값으로 1이 된다.



# 각 주의 첫번째 날짜를 반환한다.
# 기본값으로 일요일 기준으로 반환한다.
print('---------- 주(일요일) 기준 반환 ---------')
d = pd.date_range(start='2025-07-04', periods=30, freq='W')
# print(d)


# 각 주의 첫번째 날짜를 반환한다.
# 월요일 기준으로 반환한다.
print('---------- 월요일 기준 반환 ---------')
d = pd.date_range(start='2025-07-04', periods=30, freq='W-MON')  
# print(d)




```

## 🟢 가짜데이터 생성

- 가짜데이터 생성방법
  - `np.random.randn` 함수는 가우스분포(정규분포)를 따르는 값을 랜덤하게 생성한다.

  - 가우스분포를 따르는 실수값이 중요한 이유:
    - 통계학자들은 자연계에서 얻어진 수많은 데이터를 분석했다.
    - 대부분의 경우 데이터들이 평균 근처에 몰려 있고, 양쪽으로 퍼져 있는 정규분포 형태를 보였다.
    - 예를 들어:
      - 사람의 키, 몸무게
      - 학생들의 시험 점수
      - 지역의 기온 분포
    - 이런 데이터는 정규분포로 모델링하면 예측과 분석이 쉬워진다.
  - 그래서 실험이나 시뮬레이션에서 데이터를 만들 때, 정규분포를 따르는 가짜 데이터를 생성하면 현실적인 결과를 흉내내는 데 유리하다.
 
  - 머신러닝/딥러닝에서도:
    - 가중치 초기화 시 정규분포 사용
    - 노이즈 추가 또는 샘플링 등에 사용됨
    - 많은 알고리즘이 입력 데이터가 정규분포라고 가정하고 설계되어 있음




```python
import pandas as pd
import numpy as np

ts = pd.series( np.random.randn(12),
                index=pd.date_range(start='2025-01-01', periods=12, freq='D' ))

# 12개의 가짜 데이터를 만들었음
# 날짜가 인덱스임

print(ts)








```
