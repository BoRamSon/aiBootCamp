# 🟩 표현 가능한 차트 종류 파악하기

| 순위   | 라이브러리          | 언어/환경      | 특징 요약                         | 주 사용 분야                |
| ---- | -------------- | ---------- | ----------------------------- | ---------------------- |
| 🥇 1 | 🔥**Chart.js**   | JavaScript | 가볍고 쉬움. 반응형. 막대/선/파이 등 기본 차트  | 웹 프론트엔드 (React/Vue 포함) |
| 🥈 2 | **Matplotlib** | Python     | Python 시각화의 표준, 거의 모든 분석에서 사용 | 데이터 분석, 교육, 과학         |
| 🥉 3 | 🔥**Seaborn**    | Python     | 통계 시각화에 특화, Matplotlib 기반     | 통계 분석, 머신러닝, 논문        |
| 4    | 🔥**Plotly**     | Python/JS  | 대화형 그래프, 웹 지원, 다양한 언어 지원      | 리포트, 대시보드, 분석 웹앱       |
| 5    | **D3.js**      | JavaScript | SVG 기반 저수준 시각화, 자유도 최강        | 맞춤형 시각화, 데이터 저널리즘      |
| 6    | **Tableau**    | GUI        | 기업 BI 도구, 클릭으로 대시보드 제작        | 리포트, 비즈니스 인텔리전스        |
| 7    | **Highcharts** | JavaScript | 고급 상업용 차트, 기업용에 최적화           | 고급 대시보드, 실무 웹          |
| 8    | **Dash**       | Python     | Plotly 기반 대시보드 앱 개발           | 데이터 앱, 내부 대시보드         |
| 9    | **Altair**     | Python     | 선언형 시각화, 문법 간결                | 보고서 자동화, 데이터 분석        |
| 10   | **ECharts**    | JavaScript | 대용량 데이터, 인터랙티브 차트             | 실시간 시각화, 중국 내 인기       |
| 11   | **Recharts**   | React      | React 친화적인 컴포넌트 기반 차트         | React 웹앱, SaaS 프론트엔드   |
| 12   | **Bokeh**      | Python     | 웹 위젯 포함, 실시간 스트리밍 대응 가능       | 실시간 분석, 스트리밍 대시보드      |
| 13   | **Holoviews**  | Python     | 복잡한 그래프를 간단히 생성               | 과학 시각화, 모델링            |
| 14   | **Plotnine**   | Python     | ggplot 스타일 시각화                | R 유저 대상 통계 시각화         |
| 15   | **Pygal**      | Python     | SVG 기반, 가볍고 정적                | 정적 웹 문서, 블로그           |
| 16   | **Mayavi**     | Python     | 3D 과학 시각화 특화                  | 물리/공학 시뮬레이션            |
| 17   | **VisPy**      | Python     | GPU 가속, 고속 3D 렌더링             | 고성능 과학 시각화             |
| 18   | **Plotext**    | Python     | 텍스트 기반 CLI 시각화                | 서버 환경, 터미널 출력          |
