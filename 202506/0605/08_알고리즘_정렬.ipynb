# 🟩 버블 정렬 (Bubble Sort)  

### 🟡 버블 정렬 (Bubble Sort) 정의  
- 버블 정렬은 인접한 두 원소를 반복적으로 비교하고, 순서가 잘못된 경우 서로 교환(swap) 하면서 정렬하는 알고리즘입니다.  
- 작은 값이 점점 앞으로 "버블처럼 떠오르는" 모습에서 Bubble Sort 라는 이름이 붙었습니다.  

### 🟡 동작 방식 (작동 형태)  
- 리스트의 첫 번째 원소부터 시작해서 인접한 두 값을 비교합니다.  
- 앞의 값이 크면 두 값을 교환합니다.  
- 다음 쌍으로 이동해서 다시 비교 및 교환을 반복합니다.  
- 한 바퀴를 돌면 가장 큰 값이 리스트 끝으로 이동합니다.  
- 위 과정을 리스트가 정렬될 때까지 반복합니다.  
- 이 과정을 통해 가장 큰 값이 오른쪽으로, 그다음 큰 값이 그 앞쪽으로 점점 "떠오릅니다".  


## 🟡 버블 정렬 관련 문제 풀어보기  

### Hint  
if arr[j] > arr[j+i]:  
  arr[j], arr[j+1] = arr[j+1], arr[j]  

````python
list = [9,7,10,23,45,6,3,4,5]

def bubble_sort(arr):
  for i in range(len(arr)):  # 몇 번 반복할지
      for j in range(len(arr) - 1):  # 앞에서부터 비교.  🔥 완벽하지 않음.
          if arr[j] > arr[j + 1]:
              arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

bubble_sort(list)
````

````python
list = [9,7,10,23,45,6,3,4,5]

def bubble_sort2(arr):
  for i in range(len(arr)):     # 몇 번 반복할지
      for j in range(len(arr) - i - 1):   # 🔥 이 부분 range가 어려워서 이해를 못하고 있었다.
          if arr[j] > arr[j + 1]:
              arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

bubble_sort2(list)
````

#### ⚫️ 버블정렬 시간복잡도  

| 경우           | 비교 횟수 | 시간복잡도                         |  
| ------------ | ----- | ----------------------------- |
| 최선 (Best)    | O(n)  | **이미 정렬되어 있는 경우** (개선된 버전일 때) |  
| 평균 (Average) | O(n²) |                               |  
| 최악 (Worst)   | O(n²) | **역순으로 완전히 뒤바뀐 경우**           |  


````python
# 개선된 버블 정렬이라고 함.
# flag는 "더 이상 정렬할 필요가 없는지"를 판단하기 위해 사용하는 장치예요.
# 한마디로, 버블 정렬을 "불필요하게 계속 반복하지 않기 위해" 쓰는 것입니다.

def bubble_sort3(arr):
    for i in range(len(arr)):
        flag = False  # 이 반복에서 데이터가 바뀌었는지 추적
        for j in range(len(arr) - i - 1):  # 정렬된 뒤쪽은 점점 줄여감
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                flag = True
        if not flag:
            break  # 데이터가 바뀌지 않았으면 정렬 끝났음
    return arr  # 정렬된 리스트 반환

# 테스트
data = [3, 5, 7, 6, 8, 9, 25]
sorted_data = bubble_sort3(data)
print(sorted_data)
````

| -------------- | ------------------------- | --------------------------- |

| **자리를 바꾸는 시점** | 매번 비교할 때마다 자리 바꿈          | 한 바퀴 돌고 나서 한 번만 자리 바꿈       |  
| **비교 횟수**      | 많음 (교환도 많음)               | 비교는 많지만 교환은 적음              |  
| **특징**         | 큰 수가 오른쪽으로 ‘버블’처럼 밀려감     | 가장 작은 수를 ‘선택’해서 맨 앞으로 보냄    |  
| **속도 (Big-O)** | O(n²)                     | O(n²) (하지만 실제로는 버블보다 약간 빠름) |  

 결론적으로 버블은 근접한 2개를 비교하면서 바꿔대고, 선택정렬은 맨 앞을 기준으로 그 뒤를 모두 비교하여 가장 작은 값을 가져온다.

# 🟩 퀵 정렬 (Quick Sort)  

### 🟡 여기서 알아보는 알고리즘에 특징  
  - 속도가 빠르면 메모리를 많이 차지함  
  - 속도가 느리면 메모리 덜 차피  
  - 속도가 빠르면 코딩도 어려움  
  - Trade-off  
  - 최근에는 메모리가 엄청 싸다 속도위주의 알고리즘을 선택한다.  


### 🟡 정의  
- 속도가 매우 빠르고 자주 쓰이는 알고리즘  
- 하나의 기준값(pivot)을 중심(쪼개기)으로, 작은 값은 왼쪽, 큰 값은 오른쪽으로 나누고, 이 과정을 재귀적으로 반복하여 정렬  


### 🟡 형태 (생김새)  



### 🟡 예시 설명  

#### ⚫️ 기준이 되는 피벗 선택의 기준  
| 전략 이름               | 설명                                               |  
| ------------------- | ------------------------------------------------ |
| **첫 번째 요소**         | 리스트의 첫 번째 값을 피벗으로 선택                             |  
| **마지막 요소**          | 마지막 값을 피벗으로 선택                                   |  
| **중간 요소**           | 가운데 위치의 값을 피벗으로 선택                               |  
| **랜덤 선택**           | 리스트에서 임의의 값을 피벗으로 선택 (보통 최악 시간복잡도를 피하기 위해 사용)    |  
| **Median of Three** | 첫 값, 중간 값, 마지막 값 중 "중간 크기" 값을 피벗으로 선택 → 성능 향상 목적 |  

- 피벗은 임의로 선택할 수 있다.  
- 그러나 잘못된 피벗 선택은 퀵 정렬의 최악 시간복잡도 O(n²) 를 유발할 수 있음.  
- 그래서 실전에서는 랜덤 피벗, median-of-three 전략을 사용하는 것이 일반적입니다.

### 🟡 퀵정렬 코드  

#### 만들 코드에 대한 설명  
| 코드 줄                | 의미                              |  
| ------------------- | ------------------------------- |
| `if len(arr) <= 1:` | 리스트가 비었거나 요소가 1개면 바로 리턴 (정렬 완료) |  
| `pivot = arr[-1]`   | 피벗을 리스트의 마지막 요소로 선택             |  
| `left = ...`        | 피벗보다 작거나 같은 값들을 왼쪽 그룹으로         |  
| `right = ...`       | 피벗보다 큰 값들을 오른쪽 그룹으로             |  
| `quick_sort(left)`  | 왼쪽 그룹에 다시 퀵 정렬 (재귀 호출)          |  
| `quick_sort(right)` | 오른쪽 그룹에 다시 퀵 정렬 (재귀 호출)         |  


````python
def quick_sort(arr):
    if len(arr) <= 1:  # 리스트가 비었거나 1개면 정렬할 필요 없음
        return arr
    
    pivot = arr[-1]  # 지금 마지막 요소를 기준으로 잡았음
    left = [x for x in arr[:-1] if x <= pivot]
    right = [x for x in arr[:-1] if x > pivot]

    # 재귀적으로 자기 자신을 호출!
    return quick_sort(left) + [pivot] + quick_sort(right)


data = [7, 2, 1, 6, 8, 5, 3, 4]
sorted_data = quick_sort(data)
print(sorted_data)


# 이 코드는 내부 구조가 보이지 않습니다. 그래서 한번 내부 구조가 작동하는 형태를 볼 수 있게 만들어보겠습니다.
````

````python
# 위 코드의 동작이 하나하나 작동하는 것을 print로 볼 수 있도록 만든 코드

def quick_sort(arr, depth=0):
    indent = "  " * depth  # 재귀 깊이에 따라 들여쓰기 조정

    if len(arr) <= 1:
        print(f"{indent}리스트 {arr} → 정렬 완료 (리턴)")
        return arr
    
    pivot = arr[-1]
    left = [x for x in arr[:-1] if x <= pivot]
    right = [x for x in arr[:-1] if x > pivot]

    print(f"{indent}정렬 대상: {arr}")
    print(f"{indent}피벗: {pivot}")
    print(f"{indent}왼쪽 그룹: {left}")
    print(f"{indent}오른쪽 그룹: {right}")
    print()

    # 🔥 재귀호출 - 재귀적으로 자기 자신을 호출
    # 이 코드를 분리했다고 보면 됨 = return quick_sort(left) + [pivot] + quick_sort(right)
    sorted_left = quick_sort(left, depth + 1)
    sorted_right = quick_sort(right, depth + 1)

    result = sorted_left + [pivot] + sorted_right
    print(f"{indent}합치기: {sorted_left} + [{pivot}] + {sorted_right} → {result}")
    print()
    return result


data = [7, 2, 1, 6, 8, 5, 3, 4]
sorted_data = quick_sort(data)
print(sorted_data)
````

# 🟩 트리(Tree)  

- 계층적 구조를 가진 자료구조  
- **노드(Node)**로 구성되고, 노드들은 부모-자식 관계를 가짐  
- 루트 노드(root)가 있고, 루트를 제외한 각 노드는 부모 노드가 하나씩 있음  
- 사이클(순환)이 없는 그래프의 특별한 형태  


### 🟡 트리의 기본 용어  
| 용어            | 설명                      |  
| ------------- | ----------------------- |
| 노드(Node)      | 트리의 각 요소                |  
| 루트(Root)      | 트리의 최상단 노드              |  
| 부모(Parent)    | 바로 위에 연결된 노드            |  
| 자식(Child)     | 바로 아래 연결된 노드들           |  
| 형제(Sibling)   | 같은 부모를 가진 노드들           |  
| 단말 노드(Leaf)   | 자식이 없는 노드               |  
| 깊이(Depth)     | 루트에서 해당 노드까지의 거리 (층)    |  
| 높이(Height)    | 해당 노드에서 가장 먼 단말 노드까지 거리 |  
| 서브트리(Subtree) | 어떤 노드를 루트로 하는 부분 트리     |  


### 🟡 트리 종류  
| 종류                              | 설명                                    |  
| ------------------------------- | ------------------------------------- |
| 이진 트리 (Binary Tree)             | 자식 노드가 최대 2개인 트리                      |  
| 이진 탐색 트리 (BST)                  | 왼쪽 자식 < 부모 < 오른쪽 자식 조건 만족             |  
| 균형 이진 트리 (Balanced BST)         | 높이가 균형 잡힌 이진 탐색 트리 (예: AVL, 레드-블랙 트리) |  
| 완전 이진 트리 (Complete Binary Tree) | 마지막 레벨을 제외한 모든 레벨이 꽉 찬 트리             |  
| 포화 이진 트리 (Full Binary Tree)     | 모든 노드가 0개 또는 2개의 자식만 가진 트리            |  
| 힙 (Heap)                        | 부모가 자식보다 크거나 작은 완전 이진 트리              |  
| B-트리, B+트리                      | 다중 자식을 가진 균형 잡힌 트리 (데이터베이스에서 사용)      |  
| 트라이(Trie)                       | 문자열 검색에 특화된 트리 구조                     |  


### 🟡 트리 탐색 방법 (Traversal)  
| 방법                  | 설명                       | 방문 순서 (이진 트리 기준)    |  
| ------------------- | ------------------------ | ------------------- |
| Preorder            | 루트 → 왼쪽 자식 → 오른쪽 자식      | 현재 → 왼쪽 → 오른쪽       |  
| Inorder             | 왼쪽 자식 → 루트 → 오른쪽 자식      | 왼쪽 → 현재 → 오른쪽       |  
| Postorder           | 왼쪽 자식 → 오른쪽 자식 → 루트      | 왼쪽 → 오른쪽 → 현재       |  
| 레벨 순회 (Level-order) | 각 레벨을 왼쪽에서 오른쪽으로 순서대로 방문 | 루트부터 레벨별로 방문 (큐 사용) |  


### 🟡 트리 활용 분야  
- 파일 시스템 구조  
- 웹 DOM(Document Object Model)  
- 데이터베이스 인덱스 (B-트리, B+트리)  
- 표현식 계산 (수식 트리)  
- 네트워크 라우팅  
- 검색 엔진, 사전, 자동완성 (트라이)
