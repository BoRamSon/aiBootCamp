# 🟩 Tree 자료구조

자료구조 중 트리(Tree)는 일렬로 바르게 나열되어 있는 것이 아닌 계층적(hierarchical) 구조를 표현하는 데 사용되는 비선형 자료구조입니다.  
트리는 자연 세계에서의 "나무" 구조와 유사하게 생겼지만, 컴퓨터 과학에서는 루트(root)가 위에 있고, 가지(branch)가 아래로 뻗는 구조로 나타냅니다.





## 🟢 트리(Tree)의 정의
- 트리는 노드(Node)들의 계층적 구조로 이루어져 있으며, 다음과 같은 특징을 갖습니다.
    - 하나의 루트 노드(Root Node) 가 존재하고,
    - 나머지 노드는 하나의 부모 노드를 가지며,
    - 사이클(cycle) 이 없는 연결 구조입니다.
    - 즉, 트리는 사이클이 없는 연결 그래프 중에서 루트가 있는 형태라고 할 수 있습니다.



## 🟢 트리의 기본 용어





| 용어                  | 설명                         |
| ------------------- | -------------------------- |
| **노드(Node)**        | 트리의 각 원소 (데이터를 저장)         |
| **루트 노드(Root)**     | 트리의 시작점, 부모가 없는 첫 노드       |
| **부모 노드(Parent)**   | 다른 노드를 직접 연결한 상위 노드        |
| **자식 노드(Child)**    | 부모로부터 연결된 하위 노드            |
| **형제 노드(Sibling)**  | 같은 부모를 공유하는 노드들            |
| **리프 노드(Leaf)**     | 자식이 없는 노드 (말단 노드)          |
| **내부 노드(Internal)** | 최소 하나 이상의 자식을 가지는 노드       |
| **서브트리(Subtree)**   | 어떤 노드를 루트로 하는 트리의 부분 집합    |
| **레벨(Level)**       | 루트로부터 얼마나 떨어져 있는지, 루트는 0레벨 |
| **높이(Height)**      | 루트에서 리프까지의 최대 거리           |
| **깊이(Depth)**       | 노드가 루트로부터 얼마나 깊은지 (레벨과 같음) |
| **차수(Degree)**      | 한 노드가 가지는 자식의 수            |


## 🟢 트리의 자료구조적 성질
- 하나의 루트 노드와 0개 이상의 하위 트리로 구성되어 있습니다.
- 데이터를 순차적으로 저장하지 않기 때문에 비선형 자료구조입니다.
- 노드가 n개인 트리는 항상 n−1개의 간선(edge)을 가집니다.
    - 예를 들어 만약 subtree만 봤을 때, 루트노드가 있고 자식노드가 2개 있다고 가정한다면 연결되는 선은 2개입니다.
    - 노드 3에서 -1을 하면 간선은 2라는 것을 알 수가 있습니다.
- 루트부터 어떤 노드까지 유일한 경로가 존재합니다. (사이클이 없음)
- 비선형 구조로 배열이나 연결리스트보다 더 복잡한 관계를 표현할 수 있습니다.



## 🟢 목차

**비선형 자료구조**  
  - **트리 계열**  
    - 기본 트리  
      1. 이진 트리 (Binary Tree)  
      2. 이진 탐색 트리 (Binary Search Tree)  
    - 자가 균형 트리  
      1. AVL 트리  
      2. 레드-블랙 트리 (Red-Black Tree)  
      3. 스플레이 트리 (Splay Tree)  
    - B-트리 계열  
      1. B-트리 (B-Tree)  
      2. B+ 트리 (B+ Tree)  
      3. B* 트리 (B* Tree)  
    - 특수 트리  
      1. 트라이 (Trie)  
      2. 세그먼트 트리 (Segment Tree)  
      3. 펜윅 트리 (Fenwick Tree)  
      4. 힙 (Heap) 
      5. K-D 트리 (K-Dimensional Tree)
      6. 유니온-파인드 (Union-Find / Disjoint Set)
