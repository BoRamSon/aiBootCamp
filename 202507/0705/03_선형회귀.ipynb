# 🟩 회귀 분석 (Regression Analysis)

회귀 분석은 **지도학습(Supervised Learning)**의 일종으로,  
**하나 또는 그 이상의 독립 변수(x)**가 **종속 변수(y)**에 어떤 영향을 미치는지를 모델링하고,  
이를 통해 **종속 변수 값을 예측**하는 통계 기법입니다.

### 🟡 회귀 분석의 목적

- **독립 변수**가 **한 단위 증가**할 때, **종속 변수**에 미치는 영향을 측정
- **독립 변수의 값**에 기반해 **종속 변수의 값을 예측**
- 변수 간의 **인과관계, 상관관계, 추세**를 모델링

### 🟡 주요 유형

| 구분               | 설명                           |
| ------------------ | ------------------------------ |
| **단순 선형 회귀** | 독립 변수가 1개일 때 사용      |
| **다중 회귀 분석** | 독립 변수가 2개 이상일 때 사용 |

### 🟡 회귀 분석의 활용 분야

- 📊 **논문, 학술 연구**에서 인과관계 분석
- 📈 **마케팅**에서 광고비와 매출, 가격과 수요의 관계 분석
- 🕒 **시간에 따른 변화 예측** (예: 기온, 매출, 주가 등 시계열 데이터)
- 🔬 **가설 검정 및 원인-결과 모델링**
- 🤖 머신러닝 모델 학습 및 평가 지표로도 활용

### 🟡 회귀 분석의 특징

- **정량적 데이터 분석**에 특화됨
- 데이터 간의 **선형 혹은 비선형 관계를 수학적으로 모델링**
- 결과를 통해 **변수 간의 민감도 또는 영향력 해석 가능**

### 🟡 예시 수식

\[  
y = W_1x_1 + W_2x_2 + \dots + W_nx_n + b  
\]

- \( x_i \): 독립 변수
- \( W_i \): 가중치 (각 변수의 영향도)
- \( b \): 편향 (절편)
- \( y \): 예측되는 종속 변수


<br>

## 🟢 선형 회귀 (Linear Regression) 설명


선형 회귀는 **종속 변수 `y`와 독립 변수 `x` 간의 선형 관계**를 모델링하는 대표적인 회귀 알고리즘입니다.  
실제 값과 예측 값 사이의 차이를 최소화하도록 **가중치(W)**와 **편향(b)**을 학습합니다.  
<br>

### 🟡 1. 단순 선형 회귀 (Simple Linear Regression)

- 독립 변수(x)가 **1개**일 때 사용
- 예측 모델 수식:

\[  
y = Wx + b  
\]

| 항목 | 의미                    |
| ---- | ----------------------- |
| `W`  | 가중치 (weight): 기울기 |
| `b`  | 편향 (bias): y절편      |

- 학습 목적: 데이터를 가장 잘 설명하는 **직선**을 찾는 것
- 시각화: 2차원 평면상의 **직선 회귀선**  
  <br>

### 🟡 2. 다중 선형 회귀 (Multiple Linear Regression)

- 독립 변수(x)가 **2개 이상**일 때 사용
- 예측 모델 수식:

\[  
y = W_1x_1 + W_2x_2 + \dots + W_nx_n + b  
\]

- 각 독립 변수에 대해 가중치(W₁, W₂, ..., Wₙ)를 곱해서 더하고, 마지막에 편향(b)을 더함

| 항목              | 의미             |
| ----------------- | ---------------- |
| `W₁, W₂, ..., Wₙ` | 각 변수의 기울기 |
| `x₁, x₂, ..., xₙ` | 독립 변수들      |
| `b`               | 편향 (y절편)     |

- 시각화:
  - 독립변수가 2개일 경우: **3차원 평면(plane)**
  - 독립변수가 3개 이상일 경우: 고차원 공간  
    <br>

### 🟡 핵심 개념 요약

| 구분           | 단순 선형 회귀                                 | 다중 선형 회귀                                 |
| -------------- | ---------------------------------------------- | ---------------------------------------------- |
| 독립 변수 개수 | 1개                                            | 2개 이상                                       |
| 예측 함수 형태 | \( y = Wx + b \)                               | \( y = W_1x_1 + W_2x_2 + \dots + W_nx_n + b \) |
| 시각화         | 직선 (2D)                                      | 평면 이상 (3D 이상)                            |
| 학습 목표      | 선형 관계에 가장 잘 맞는 회귀계수(W, b)를 찾기 |

<br>

### 🟡 머신러닝 관점에서

- 선형 회귀는 지도학습(Supervised Learning) 방법 중 **회귀(Regression)** 문제에 사용
- 모델 훈련은 손실함수(예: MSE)를 최소화하는 **W, b 파라미터 최적화 과정**  
  <br>

### 🟡 활용 예시

- 키에 따른 몸무게 예측
- 공부 시간에 따른 시험 점수 예측
- 광고비에 따른 판매량 예측 등  
  <br>

## 🟢 선형 회귀 분석 : 데이터 분석


````python
## 🟢 데이터 분석

### 🟡 공부 시간 
X = [[20], [19], [18], [17], [16], [15], [14], [13], [12], [11], [10], [9], [8], [7], [6]]

# 평균값
Y = [100, 100, 90, 90, 60, 70, 40, 40, 70, 30, 45, 40, 30, 20, 10]

import numpy as np
X = np.array(X)
Y = np.array(Y)

print(X.shape)
print(Y.shape)


# 또 데이터를 훈련셋과 테스트셋으로 나누자!!!!
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state=1)

from sklearn.linear_model import LinearRegression # 선형 회귀 모델
model = LinearRegression()   # 하이퍼파라미터 없음 과대던 과소던 할 수 있는 건 테이터셋 늘려주기 밖에 없다.
model.fit(X_train, Y_train)

print("--------------------------------")
y_pred = model.predict(X_test)
print("훈련셋 점수 : ", model.score(X_train, Y_train))
print("테스트셋 점수 : ", model.score(X_test, Y_test))
print("기울기 : ", model.coef_)
print("절편 : ", model.intercept_)



print("--------------------------------")
y_pred2 = X_test * model.coef_ + model.intercept_
print("실제값 : ", Y_test)
print("예측값 : ", y_pred)   
print("예측값2 : ", y_pred2)

# 다중회귀분석의 경우 가중치가 많다.
# 각 독립변수마다 별도의 가중치를 가져온다.
# 가중치를 통해 예측값을 계산할 수 있다.
# 가중치를 통해 예측값을 계산할 수 있다.


# ========================================================
# knn 이웃 회귀 알고리즘
from sklearn.neighbors import KNeighborsRegressor
model = KNeighborsRegressor(n_neighbors=3)
model.fit(X_train, Y_train)
y_pred = model.predict(X_test)
print("훈련셋 점수 : ", model.score(X_train, Y_train))
print("테스트셋 점수 : ", model.score(X_test, Y_test))
print("실제값 : ", Y_test)
print("예측값 : ", y_pred)
````

````python
import mglearn  #사이킷런 책 쓴사람이 차트그리기 편하고 가끔 가짜 데이터 만들어보라고 만든 라이브러리 
X, y = mglearn.datasets.make_wave(n_samples=200)
print(X[:10])
print(y[:10])


from sklearn.model_selection import train_test_split 
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=1)

from sklearn.linear_model import LinearRegression
model = LinearRegression()   #하이퍼파라미터 없음 과대던 과소던 할수있는건 데이터셋 늘려주기 밖에 없다. 
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("훈련셋 : ", model.score(X_train, y_train))
print("테스트셋 : ", model.score(X_test, y_test))
print("기울기 : ", model.coef_) 
print("절편 : ", model.intercept_)
y_pred2 = X_test * model.coef_ + model.intercept_
print(y_test)
print(y_pred)
print(y_pred2)
#다중회귀분석의 경우 가중치가 많다. 각 독립변수마다 별도의 가중치를 가져온다 
# w1x1 + w2x2 + w3x3 ......... + wnxn

"""
# (w1,w2,w3,...wn)  X  (x1,
                        x2, 
                        x3, 
                        x4, ......
                        xn)

"""
#Knn 이웃 회귀알고리즘 
from sklearn.neighbors import KNeighborsRegressor
model = KNeighborsRegressor(n_neighbors=3)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("훈련셋 : ", model.score(X_train, y_train))
print("테스트셋 : ", model.score(X_test, y_test))
print(y_test)
print(y_pred)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 데이터 생성
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([1.5, 3.8, 6.1, 8.2, 10.1])

# 모델 학습
model = LinearRegression()
model.fit(x, y)

# 예측
y_pred = model.predict(x)

# 시각화
plt.scatter(x, y, color='blue', label='실제 데이터')
plt.plot(x, y_pred, color='red', label='회귀선')
plt.xlabel('x')
plt.ylabel('y')
plt.title('선형 회귀')
plt.legend()
plt.show()
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
import numpy as np
import matplotlib.pyplot as plt

# 데이터 생성
np.random.seed(0)
x = np.linspace(0, 1, 15)
y = np.sin(2 * np.pi * x) + np.random.normal(0, 0.1, x.shape)
x = x.reshape(-1, 1)

# 다양한 모델 비교
degrees = [1, 4, 10]
plt.figure(figsize=(15, 4))

for i, d in enumerate(degrees):
    poly = PolynomialFeatures(degree=d)
    x_poly = poly.fit_transform(x)

    model = LinearRegression()
    model.fit(x_poly, y)
    y_pred = model.predict(x_poly)

    plt.subplot(1, 3, i+1)
    plt.scatter(x, y, label='실제 데이터')
    plt.plot(x, y_pred, color='red', label=f'예측 (deg={d})')
    plt.title(f'{d}차 다항 회귀')
    plt.legend()

plt.show()
````
