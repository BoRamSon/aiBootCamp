# 🟩 자료구조(Data Structure)와 알고리즘(Algorithms)  

## 🟢 자료구조 (Data Structures)  

- **1. 선형 자료구조**  
  - **배열 계열**  
    1. 정적 배열 (Static Array)  
    2. 동적 배열 (Dynamic Array)  
    3. 다차원 배열 (Multi-dimensional Array)  
    4. 희소 배열 (Sparse Array)  
    5. (python의 `tuple`도 같이 참고하세요.)  
  - **연결 리스트 계열**  
    1. 단일 연결 리스트 (Singly Linked List)  
    2. 이중 연결 리스트 (Doubly Linked List)  
    3. 원형 연결 리스트 (Circular Linked List)  
    4. 이중 원형 연결 리스트 (Doubly Circular Linked List)  
  - **스택과 큐 계열**  
    1. 스택 (Stack)  
    2. 큐 (Queue)  
       - 선형 큐 (Linear Queue)  
       - 원형 큐 (Circular Queue)  
       - 덱 (Deque)  
       - 우선순위 큐 (Priority Queue)  

<br>

- **2. 비선형 자료구조**  
  - **트리 계열**  
    - 기본 트리  
      1. 이진 트리 (Binary Tree)  
      2. 이진 탐색 트리 (Binary Search Tree)  
    - 자가 균형 트리  
      1. AVL 트리  
      2. 레드-블랙 트리 (Red-Black Tree)  
      3. 스플레이 트리 (Splay Tree)  
    - B-트리 계열  
      1. B-트리 (B-Tree)  
      2. B+ 트리 (B+ Tree)  
      3. B* 트리 (B* Tree)  
    - 특수 트리  
      1. 트라이 (Trie)  
      2. 세그먼트 트리 (Segment Tree)  
      3. 펜윅 트리 (Fenwick Tree)  
      4. 힙 (Heap)  
  - **그래프 계열**  
    1. 방향 그래프 (Directed Graph)  
    2. 무방향 그래프 (Undirected Graph)  
    3. 가중 그래프 (Weighted Graph)  
    4. 완전 그래프 (Complete Graph)  
    5. 이분 그래프 (Bipartite Graph)  

<br>

- **3. 해시 기반 자료구조**  
  1. 해시 테이블 (Hash Table)  
  2. 해시 맵 (Hash Map) (python의 `dictionary`와 동일한 개념)  
  3. 해시 셋 (Hash Set) (python의 `set`과 동일한 개념)  
  4. 블룸 필터 (Bloom Filter)  

<br>

- **4. 고급/특수 자료구조**  
  - **집합 관련**  
    1. 유니온-파인드 (Union-Find)  
    2. 비트셋 (Bitset)  
  - **문자열 관련**  
    1. 서픽스 배열 (Suffix Array)  
    2. 서픽스 트리 (Suffix Tree)  
    3. LCP 배열 (LCP Array)  
  - **기하학적 자료구조**  
    1. KD-트리 (K-D Tree)  
    2. 쿼드트리 (Quadtree)  
    3. 옥트리 (Octree)  
    4. R-트리 (R-Tree)  
  - **동시성 관련**  
    1. 락-프리 자료구조 (Lock-free Data Structures)  
    2. 원자적 자료구조 (Atomic Data Structures)  

<br>

- **5. 확률적 / 근사 자료구조**  
  1. 스킵 리스트 (Skip List)  
  2. 카운트-민 스케치 (Count-Min Sketch)  
  3. 하이퍼로그로그 (HyperLogLog)

## 🟢 알고리즘(Algorithms)  

- **1. 정렬 알고리즘 (Sorting Algorithms)**  
  - **비교 기반 정렬**  
    1. 버블 정렬 (Bubble Sort)  
    2. 선택 정렬 (Selection Sort)  
    3. 삽입 정렬 (Insertion Sort)  
    4. 병합 정렬 (Merge Sort)  
    5. 퀵 정렬 (Quick Sort)  
    6. 힙 정렬 (Heap Sort)  
    7. 셸 정렬 (Shell Sort)  
  - **비교하지 않는 정렬**  
    1. 계수 정렬 (Counting Sort)  
    2. 기수 정렬 (Radix Sort)  
    3. 버킷 정렬 (Bucket Sort)  

<br>

- **2. 탐색 알고리즘 (Search Algorithms)**  
  - **선형 탐색**  
    1. 순차 탐색 (Linear Search)  
  - **이진 탐색 계열**  
    1. 이진 탐색 (Binary Search)  
    2. 삼진 탐색 (Ternary Search)  
    3. 지수 탐색 (Exponential Search)  
  - **해시 기반 탐색**  
    1. 해시 탐색 (Hash Search)  

<br>

- **3. 그래프 알고리즘 (Graph Algorithms)**  
  - **그래프 순회**  
    1. 깊이 우선 탐색 (DFS, Depth-First Search)  
    2. 너비 우선 탐색 (BFS, Breadth-First Search)  
  - **최단 경로 알고리즘**  (탐욕(Greedy) 알고리즘)  
    1. 다익스트라 알고리즘 (Dijkstra's Algorithm)  
    2. 벨만-포드 알고리즘 (Bellman-Ford Algorithm)  
    3. 플로이드-워셜 알고리즘 (Floyd-Warshall Algorithm)  
    4. A* 알고리즘  
  - **최소 신장 트리**  (탐욕(Greedy) 알고리즘)  
    1. 크러스칼 알고리즘 (Kruskal's Algorithm)  
    2. 프림 알고리즘 (Prim's Algorithm)  
  - **위상 정렬**  
    1. 칸의 알고리즘 (Kahn's Algorithm)  
    2. DFS 기반 위상 정렬  

<br>

- **4. 동적 계획법 (Dynamic Programming)**  
  - **기본 DP 패턴**  
    1. 피보나치 수열  
    2. 최장 공통 부분 수열 (LCS)  
    3. 최장 증가 부분 수열 (LIS)  
    4. 배낭 문제 (Knapsack Problem)  
  - **고급 DP**  
    1. 구간 DP (Interval DP)  
    2. 트리 DP (Tree DP)  
    3. 비트마스킹 DP  

<br>

- **5. 탐욕 알고리즘 (Greedy Algorithms)**  
  1. 활동 선택 문제 (Activity Selection)  
  2. 분할 가능 배낭 문제  
  3. 허프만 코딩 (Huffman Coding)  
  4. 회의실 배정 문제  

<br>

- **6. 분할 정복 (Divide and Conquer)**  
  1. 병합 정렬 (Merge Sort)  
  2. 퀵 정렬 (Quick Sort)  
  3. 거듭제곱 (Fast Exponentiation)  
  4. 최대 부분 배열 합  

<br>

- **7. 백트래킹 (Backtracking)**  
  1. N-Queen 문제  
  2. 스도쿠 해결  
  3. 미로찾기  
  4. 부분집합 생성  
  5. 순열과 조합  

- **7. 분기 한정(Branch and Bound)**  
  - 백트래킹과 유사하지만, 더 나은 해를 찾기 위해 가지를 치는 방법입니다.  
  1. 외판원 문제(Travelling Salesman Problem)  
  2. 0-1 배낭 문제  

<br>

- **8. 문자열 알고리즘 (String Algorithms)**  
  - **문자열 매칭**  
    1. 브루트 포스 (Brute Force)  
    2. KMP 알고리즘 (Knuth-Morris-Pratt)  
    3. 보이어-무어 알고리즘 (Boyer-Moore)  
    4. 라빈-카프 알고리즘 (Rabin-Karp)  
  - **고급 문자열 처리**  
    1. Z 알고리즘  
    2. 매나커 알고리즘 (Manacher's Algorithm)  
    3. 서픽스 배열 (Suffix Array)  

<br>

- **9. 수학적 알고리즘 (Mathematical Algorithms)**  
  - **수론**  
    1. 유클리드 호제법  
    2. 확장 유클리드 알고리즘  
    3. 에라토스테네스의 체  
    4. 모듈러 지수법  
  - **조합론**  
    1. 파스칼의 삼각형  
    2. 카탈란 수  

<br>

- **10. 기하 알고리즘 (Geometric Algorithms)**  
  1. 컨벡스 헐 (Convex Hull)  
  2. 그레이엄 스캔  
  3. 선분 교차 판정  
  4. 최근접 점 쌍  

<br>

- **11. 고급 알고리즘**  
  - **네트워크 플로우**  
    1. 포드-풀커슨 알고리즘  
    2. 에드몬드-카프 알고리즘  
    3. 디닉 알고리즘  
  - **이분 매칭**  
    1. 헝가리안 알고리즘  
    2. 쿤의 알고리즘  

<br>

각 알고리즘은 특정 문제 유형에 최적화되어 있으므로, 문제의 특성을 파악하고 적절한 알고리즘을 선택하는 것이 중요합니다.
