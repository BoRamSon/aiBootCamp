## 🟢 트리 순회(Tree Traversal)
트리에서 모든 노드를 방문하는 방법

1. DFS (깊이 우선 탐색)
- 이진 트리 기준
    - 전위 순회(Pre-order traversal): 루트 → 왼쪽 → 오른쪽
    - 중위 순회(In-order traversal): 왼쪽 → 루트 → 오른쪽
    - 후위 순회(Post-order traversal): 왼쪽 → 오른쪽 → 루트

2. BFS (너비 우선 탐색)
    - 레벨 순서대로 방문 (큐 자료구조 사용)

## 🟢 트리의 주요 활용 사례
 
- 디렉터리(폴더) 구조 표현
   ```markdown
         Users (루트)
         └── Kim (Users의 자식)
            ├── Documents (Kim의 자식)
            │   ├── Report.docx (리프 노드)
            │   └── Resume.pdf  (리프 노드)
            ├── Downloads (Kim의 자식)
            └── Pictures (Kim의 자식)
               └── Hawaii_2023.jpg (리프 노드)
   ```
   
- 이진 탐색 트리 (BST)로 빠른 검색/삽입/삭제
   - 루트를 기준으로 들어갈 수록 왼쪽에 루트보다 작은 값을.. 오른쪽에 루트보다 높은 값을 놓기 때문에 50%를 잘라내고 들어갈 수 있기 때문에 빠르다고 합니다.
   - 
- 힙으로 우선순위 큐 구현
- 파싱 트리 (컴파일러, 수식 계산 등)
- 트라이로 문자열 검색 최적화
- 데이터베이스 인덱싱 (B 트리, B+ 트리)
- 머신러닝에서 결정 트리(Decision Tree)
