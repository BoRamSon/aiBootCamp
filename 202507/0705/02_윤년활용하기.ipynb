## 🟢 윤년 활용해보기


```python
def isLeaf(year):
    if year%4==0 and year%100!=0 or year%400==0:
        return 1
    return 0

x = []
y = []
for i in range (1, 2026):
    x.append(i)
    y.append(isLeaf(i))
    # print(x[2000:])
    # print(y[2000:])



# 머신러닝, x는 반드시 ndarray(2d), y는 1d array
import numpy as np

x = np.array(x)
# 차원 하나를 추가해서 2차원으로 만들자
x = x.reshape(-1, 1)
print(x.shape)

y = np.array(y)
y = y.reshape(-1, 1)
print(y.shape)




# 훈련셋과 테스트셋을 나누자!
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
                                        x, y, test_size=0.3, 
                                        random_state=1
                                    )
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3)  # knn 이웃 분류 알고리즘
model.fit(x_train, y_train)
print(f"훈련셋 : {model.score(x_train, y_train)}")
print(f"테스트셋 : {model.score(x_test, y_test)}")




```
