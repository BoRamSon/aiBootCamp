# 🟩 자료구조 - 이진 탐색 트리 (Binary Search Tree)  

이분검색 - 배열  

- 데이터를 넣어서 트리를 만들 때  
- 순서를 지키면서 만든다.  
- left, right 2개의 edge를 갖는다.  
- 나보다 작은 값은 왼쪽으로  
- 나보다 큰 값은 오른쪽으로  

```markdown
16 8 9 2 4 12 17 21 23   이렇게 있다고 했을 때  

                     16  
                     
        8                      17  
    2       9                       21  
4               12                      23  
```


## 🟢 코드로 만들기  

````python
class Data:
    def __init__(self, num):
        self.num = num


# 이진트리 구축하기
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# 이진 탐색 트리
class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, data):
        # 1. root Node가 있는지 확인을 하고 None이면 root node를 빨리 만든다.
        if not self.root:
            self.root = TreeNode(data)
            return  # 루트노드 만들고 끝낸다.

        # 1. 내 노드가 들어갈 위치를 찾기
        parent = None
        current = self.root # 추적해서 들어가다 보면 None인 곳에 추가가능하다.

        while current:   # current가 None이 아닌 동안
            if current.data.num == data.num:
                return   # 중복데이터 배제
            parent = current  # 현재 위치값을 저장을 해놓는다. 나중에 parenet에 연결을 시킨다.
            

            # 나보다 작은 값은 왼쪽으로 나보다 큰 값은 오른쪽으로 결정을 해줍니다.
            if data.num < current.data.num:
                current = current.left
            else:
                current = current.right
            
        # 터미널 노드는 edge가 없다. 
        # current 값이 None일 때까지 왼쪽 오른쪽 움직이면서 찾아간다.
        # 그래서 뒤에 parent가 따라가야한다.
        
        # 노드 만들어서 Parent에 연결하기
        newNode = TreeNode(data)
        if data.num < parent.data.num:
            parent.left = newNode
        else:
            parent.right = newNode

    
    # 🟢 데이터 검색함수 
    def search(self, key):
        current = self.root #검색시작위치 
        count = 0 #찾은 회수 
        while current:
            if key.num == current.data.num: #찾았으니까 
                return count 
            count+=1 
            if key.num < current.data.num:
                current = current.left 
            else:
                current = current.right 

        return -1


    def find(self, key):
        parent = None         # 삭제될 노드의 부모와 작시을 연결해야 한다.
        current = self.root   # 삭제될 노드의 위치를 찾는다.
        find = False          # 못 찾음
        while current and not find:
            if current.data.num == key.num:
                find = True
            else:
                parent = current
                if key.num < current.data.num:
                    current = current.left
                else:
                    current = current.right
        return find, parent, current

    
    def delete(self, key):
    # 내부에서 사용할 최소값 노드를 찾는 함수 (오른쪽 서브트리에서 가장 작은 노드)
        def find_min(node):
            while node.left:
                node = node.left
            return node

        # 실제 삭제를 수행하는 재귀 함수
        def delete_node(node, key):
            # 노드가 없다면 삭제할 것이 없다
            if node is None:
                return None

            # 삭제할 값이 현재 노드보다 작으면 왼쪽 서브트리로 이동
            if key.num < node.data.num:
                node.left = delete_node(node.left, key)

            # 삭제할 값이 현재 노드보다 크면 오른쪽 서브트리로 이동
            elif key.num > node.data.num:
                node.right = delete_node(node.right, key)

            else:
                # ---------------------------------------
                # Case 1: 자식이 없는 경우 (단말 노드)
                if node.left is None and node.right is None:
                    return None  # 그냥 None을 반환하면 현재 노드가 삭제됨

                # ---------------------------------------
                # Case 2: 자식이 하나인 경우
                if node.left is None:   # 오른쪽 자식만 있는 경우
                    return node.right   # 현재 노드를 오른쪽 자식으로 대체
                if node.right is None:  # 왼쪽 자식만 있는 경우
                    return node.left    # 현재 노드를 왼쪽 자식으로 대체

                # ---------------------------------------
                # Case 3: 자식이 둘 다 있는 경우
                    # 오른쪽 서브트리에서 가장 작은 값을 찾아 현재 노드를 대체
                min_larger_node = find_min(node.right)
                node.data = min_larger_node.data  # 값만 바꾸기
                    # 중복 노드 제거를 위해 오른쪽 서브트리에서 다시 삭제 재귀 호출
                node.right = delete_node(node.right, min_larger_node.data)

            return node  # 변경된 (또는 삭제된) 노드를 부모에게 반환

        # 삭제 시작은 항상 루트부터
        self.root = delete_node(self.root, key)


    def inorder(self, node):
        if node == None:
            return
        self.inorder(node.left)
        print(node.data.num)
        self.inorder(node.right)



# ❤️ 시간만 들이면 이게 어떻게 insert 해서 inorder 방식으로 조회를 했는지, 검색하고, 삭제하는지 알 수 있습니다.
# ❤️ 나중에 시간이 나면 공부하면 되겠습니다. 
# ❤️ 일단 대충 작은값, 큰값이 나뉘는 내용만 알면 되겠습니다.

if __name__ == "__main__":

    print('-------insert-------')
    bst = BinarySearchTree()
    arr = [16,8,9,2,4,12,17,21,23]
    for i in arr:
        bst.insert(Data(i))
    bst.inorder(bst.root)

    print('-------search-------')
    print(bst.search(Data(16)))
    print(bst.search(Data(2)))
    print(bst.search(Data(26)))

    print('--------delete--------')
    bst.delete(Data(8))  # 예: 8을 삭제
    bst.inorder(bst.root)



````
