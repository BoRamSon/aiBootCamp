# 🟩 Greedy Algorithm  

- 문제를 해결할 때마다 "가장 좋아 보이는 선택(최선의 선택)"을 하는 알고리즘.  
- 현재 순간의 최적 선택을 반복해서 전체 문제의 최적 해답을 찾으려고 하는 방식.  

- 🔑 핵심 개념  
  - 지금 당장 가장 이득이 되는(좋은) 선택을 한다.  
  - 미래를 고려하지 않고, 매 순간 "이게 최선이겠지!"라고 선택한다.  
  - 항상 정답을 보장하진 않지만, 특정 조건에서 정확하고 빠르게 답을 찾을 수 있어요.  

- 그리디  
  - 욕심쟁이, 탐욕  
  - 반드시 정렬을 해야 문제을 풀 수 있다.  

- 예제

  - 거스름돈 금액 N원을 입력받아 동전의 최소 개수를 구하라.

## 🟢 거스름돈 주기  

```python

# 1580원을 거슬러준다고 가정해보자!
change = 1580
print(f"{change}원 거스름돈이 있습니다.")

fivehundredwon = 0
onehundredwon = 0
fivetenwon = 0
onetenwon = 0

while change != 0:
  if len(str(change)) >= 3:
    if (change / 500) >= 1:
      a = change / 500
      a1 = int(a)  # 500원은 3개
      fivehundredwon = a1
      a2 = change - (a1 * 500)
      change = a2
      print('500원 = ', fivehundredwon, change)
    else:
      a = change / 100
      a1 = int(a)
      onehundredwon = a1
      a2 = change - (a1 * 100)
      change = a2
      print('100원 = ', onehundredwon, change)

  elif len(str(change)) >= 2:
    if (change / 50) >= 1:
      a = change / 50
      a1 = int(a)  # 50원은 1개
      fivetenwon = a1
      a2 = change - (a1 * 50)
      change = a2
      print('50원 = ', fivetenwon, change)
    else:
      a = change / 10
      a1 = int(a)
      onetenwon = a1
      a2 = change - (a1 * 10)
      change = a2
      print('10원 = ', onetenwon, change)

```

## 🟢 회의실 사용 경우의 수 구하기

```python
meetings = [(1, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]
# 단순히 생각하면 진짜 모르겠지만, 맨 앞 1, 4 기준을 하나 잡아서 하라고 합니다.

# (1, 4), (3, 5)                              여기서 or이 왜 들어갔냐면, = 1, 4가 먼저오는 시간이 아닐 수 있잖아 (순서문제)
# initial_time[1]  <=  impossible_time[0]   or   impossible_time[1]  <=  initial_time[0]
#        4                     3                        5                      1

# 이것을 생각 못했다.
# def is_overlap(range1, range2):
#     return (range1[1] <= range2[0] or range2[1] <= range1[0])

def boram(meetings):
  initial_time = meetings[0]
  # print(initial_time)
  # print(type(initial_time))

  impossible_time = []
  impossible_time.append(initial_time)
  # print(impossible_time)
  # print(type(impossible_time))

  for i in meetings[1:]:
    temp = initial_time
    if temp[1] <= i[0] or i[1] <= temp[0]:
      temp = i
      initial_time = i
      impossible_time.append(i)

  # print(initial_time)
  print('🟢 결과 : ', impossible_time)


boram(meetings)
```

```python
meetingss = [(1, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]

def max_meetings(meetingss):
  # 종료 시간을 기준으로 정렬이 되어있다면 안해도 됩니다.
  meetingss.sort(key=lambda x:x[1])
  count = 0
  end_time = 0
  for start, end in meetingss:
    if start >= end_time:
      count += 1
      end_time = end

  return count

# 실행
print(max_meetings(meetingss))
# 출력 : 3
```
