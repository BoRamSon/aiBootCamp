    - Python 리스트의 insert 기능을 수동으로 구현해보기

      - appendleft(x): 연결 리스트의 처음에 x를 추가한다.  
      - append(x): 연결 리스트의 끝 x를 추가한다.  
      - popleft(): 연결 리스트에서 첫 노드의 값을 반환하고, 노드는 삭제한다.  
      - pop(): 연결 리스트에서 마지막 노드의 값을 반환하고, 노드는 삭제한다.  
      - insert(i, x): 연결 리스트의 i번 인덱스에 x를 추가한다.  
      - remove(x): 연결 리스트에서 값이 x인 노드를 찾아 삭제한다.  
      - reverse(): 연결 리스트를 제자리에서 순서를 뒤집는다.  

  - [**Stack**] 파이썬으로 `Stack` 구현하기

## 🟢 정의  

#### ⚫️ 알고리즘 (Algorithm)  
- 정의  
  - 문제를 해결하기 위한 절차나 방법을 정해진 순서대로 명확하게 기술한 것.  
    - 입력을 받아서 원하는 출력을 만들어내는 논리적인 절차.  
    - 반드시 유한한 단계로 끝나야 함.  
- 예시  
  - 버블 정렬, 이진 탐색, 최대값 찾기 등  
- 비유  
  - 요리 레시피  
    - "김치찌개를 만들기 위한 순서와 방법"  
  - 대중적인 코딩테스트 문제  
    - "이렇게 풀면 가장 베스트인 방식들이 존재한다."  


#### ⚫️ 자료구조 (Data Structure)  
- 정의  
  - 데이터를 효율적으로 저장하고 조직하기 위한 구조나 형태.  
    - 데이터가 어떻게 저장되는지,  
    - 어떻게 접근하고,  
    - 어떻게 수정하거나 삭제하는지를 정의함.  
- 예시  
  - 배열, 연결 리스트, 트리, 스택, 큐, 해시 테이블 등  
- 비유  
  - 냉장고/서랍의 정리 방식  
    - "같은 종류끼리 정리해두면 찾기 쉬움"  


#### ⚫️ 추상 자료형 (Abstract Data Type, ADT)  
- 정의  
  - 데이터와 그것을 다루는 연산을 함께 추상화한 것. "무엇을 할 수 있는지"는 정의하지만, "어떻게 구현되는지"는 숨김.  
  - ADT는 인터페이스만 제공하고,  
  - 내부 구조는 은닉(encapsulation)  

- 예시  
  - 스택 (LIFO 구조, push, pop만 제공)  
  - 큐 (FIFO 구조, enqueue, dequeue만 제공)  
  - 스택이 배열로 구현되든, 연결 리스트로 구현되든 상관없이, 사용자는 인터페이스만 알면 된다.  

-  비유  
  - 자동차의 페달  
    - "브레이크를 밟으면 멈춘다" → 어떻게 멈추는지는 몰라도 사용 가능

## 🟢 메모리 및 저장장치 종류 비교  

- 부팅  
  - 전원을 킨다. ->  ROM을 읽어서 기본 프로그램 (ROM BIOS)을 읽고  ->  외장하드나 SSD 또는 cd를 이용해서 OS를 읽어들인다.  
    - OS는 RAM 이라는 메모리에 올라가게 된다.  

| 구분           | 약자                   | 설명                                  | 예시                        |  
| ------------ | -------------------- | ----------------------------------- | ------------------------- |
| **SSD/HDD**  | -                    | **비휘발성 저장장치**, 데이터를 영구 저장함          | 운영체제, 사진, 문서 등            |  
| **ROM**      | Read Only Memory     | **읽기 전용 비휘발성**, 일반적으로 수정 불가         | BIOS, 펌웨어                 |  
| **RAM**      | Random Access Memory | **SRAM + DRAM 포함**, 휘발성             | 실행 중인 OS, 프로그램 데이터        |  
| **SRAM**     | Static RAM           | 빠른 휘발성 메모리, 리프레시 불필요, **비싸고 용량 작음** | CPU 캐시 (L1, L2, L3)       |  
| **DRAM**     | Dynamic RAM          | 느리지만 저렴하고 용량 큼, **리프레시 필요**         | 메인 메모리 (DDR4, DDR5 등)     |  
| **Register** | -                    | CPU 내부의 **가장 빠른 메모리**, **아주 적은 용량**, (ax,bx,cx,dx,ip - 다음 명령어를 가르팀, pc) | 프로그램 카운터(PC), 연산 중간 결과 저장 |  

### 🟡 **Register**  

| 구분                         | 역할                                       |  
| -------------------------- | ---------------------------------------- |
| `AX, BX, CX, DX`           | CPU가 데이터를 저장하고 계산할 때 사용하는 **일반 목적 레지스터** |  
| `IP (Instruction Pointer)` | 다음에 실행할 **명령어의 주소를 저장**                  |  
| `PC (Program Counter)`     | `IP`과 같은 역할 (용어만 다름)                     |  


## 🟢 언어별 변수 비교  

- C언어  
  - 값 변수  
    - int a = 6;   // a라는 변수와 값 6 전부 스택에 저장  
  - 포인터형 변수  
    - int *p;      // 값 저장 못함, 주소만 저장  
    - p = &a;      // 다른 변수의 주소를 저장하거나,  
    - p = new int  // 힙공간에 동적으로 할당을 받는다.  
    - *p = 20;  
    - int **q;     // 2중 포인터  
    - int ***r;  
    - 어떤 변수를 값다입으로도 선언할 수 있고, 또는 포인터 타입으로도 선언할 수 있다.  
    - 장점 : OS나 그 밖에 시스템 프로그램이 다 가능하다. 진짜 못하는게 없다.
