# 🟩 자료구조 - 해시(Hash)  

해시는 어떤 입력값을 고정된 길이의 숫자/문자로 바꿔주는 "정보 요약 도구"입니다.  
빠르고, 안전하며, 검색이나 보안에 매우 유용해요!  

## 🎯 개념  

**해시(Hash)**는 임의의 크기 데이터를 고정된 크기 값으로 변환하는 함수입니다.  

### 핵심 특징  
- **빠른 검색**: O(1) 평균 시간복잡도  
- **고정 크기**: 입력 크기에 관계없이 출력 크기 일정  
- **결정적**: 같은 입력은 항상 같은 출력  

---

## 💡 해시 함수 종류  

### 1. 나눗셈법 (Division Method)  
```cpp
int hash_division(int key, int table_size) {  
    return key % table_size;  
}  
```

### 2. 곱셈법 (Multiplication Method)  
```cpp
int hash_multiplication(int key, int table_size) {  
    double A = 0.6180339887; // (√5 - 1) / 2  
    return (int)(table_size * (key * A - (int)(key * A)));  
}  
```

### 3. 문자열 해싱  
```cpp
int string_hash(string s) {  
    int hash_value = 0;  
    int base = 31;  
    for (char c : s) {  
        hash_value = hash_value * base + c;  
    }  
    return hash_value;  
}  
```

## 🔧 충돌 해결 방법  

### 1. 체이닝 (Chaining)  
```cpp
class HashTable {  
    vector<list<pair<int, int>>> table;  
    int size;  
    
public:  
    HashTable(int s) : size(s), table(s) {}  
    
    void insert(int key, int value) {  
        int index = key % size;  
        table[index].push_back({key, value});  
    }  
    
    bool search(int key) {  
        int index = key % size;  
        for (auto& pair : table[index]) {  
            if (pair.first == key) return true;  
        }  
        return false;  
    }  
};  
```

### 2. 개방 주소법 (Open Addressing)  
```cpp
class OpenHashTable {  
    vector<int> table;  
    vector<bool> deleted;  
    int size;  
    
    int probe(int key, int i) {  
        return (key + i) % size; // 선형 조사  
    }  
    
public:  
    OpenHashTable(int s) : size(s), table(s, -1), deleted(s, false) {}  
    
    void insert(int key) {  
        for (int i = 0; i < size; i++) {  
            int index = probe(key, i);  
            if (table[index] == -1 || deleted[index]) {  
                table[index] = key;  
                deleted[index] = false;  
                return;  
            }  
        }  
    }  
};  
```

## 📚 실전 활용  

### 1. 문자열 패턴 매칭 (라빈-카프)  
```cpp
bool rabin_karp(string text, string pattern) {  
    int n = text.length(), m = pattern.length();  
    int base = 256, mod = 1000000007;  
    
    // 패턴 해시값 계산  
    int pattern_hash = 0;  
    for (int i = 0; i < m; i++) {  
        pattern_hash = (pattern_hash * base + pattern[i]) % mod;  
    }  
    
    // 텍스트 슬라이딩 윈도우 해시  
    int text_hash = 0;  
    int power = 1;  
    for (int i = 0; i < m; i++) {  
        text_hash = (text_hash * base + text[i]) % mod;  
        if (i < m - 1) power = (power * base) % mod;  
    }  
    
    if (text_hash == pattern_hash) return true;  
    
    for (int i = m; i < n; i++) {  
        // 슬라이딩 윈도우 해시 업데이트  
        text_hash = (text_hash - (text[i-m] * power) % mod + mod) % mod;  
        text_hash = (text_hash * base + text[i]) % mod;  
        
        if (text_hash == pattern_hash) return true;  
    }  
    
    return false;  
}  
```

### 2. 중복 검사  
```cpp
bool has_duplicate(vector<int>& arr) {  
    unordered_set<int> seen;  
    for (int num : arr) {  
        if (seen.count(num)) return true;  
        seen.insert(num);  
    }  
    return false;  
}  
```
